## A1 -- Logaritmo e ln(x): o que e DE VERDADE

### A pergunta que o log responde

Esquece formula por um segundo. Log responde UMA pergunta:

> "Quantas vezes eu preciso multiplicar a base por ela mesma para chegar no numero?"

**Exemplo concreto:**
```
log_2(8) = ?
```
Pergunta: quantas vezes multiplico 2 por ele mesmo pra dar 8?
- 2 (uma vez)
- 2 x 2 = 4 (duas vezes)
- 2 x 2 x 2 = 8 (tres vezes!)

Resposta: 3. Entao log_2(8) = 3.

### O que e ln(x) entao?

`ln` e simplesmente **log na base e** (o numero 2.718...).

```
ln(x) = log_e(x)
```

Quando voce ve `ln(x)`, leia: "quantas vezes multiplico 2.718... pra chegar em x?"

**Por que usar base e e nao base 10?** Porque o numero e tem propriedades
matematicas magicas (a derivada de e^x e o proprio e^x). Isso simplifica
calculos em estatistica e ML. Mais sobre isso em A2.

### Por que log aparece TANTO em data science?

Tres razoes praticas:

**1. Log "encolhe" numeros grandes (comprime escala).**
Renda de clientes varia de R$ 1.000 a R$ 10.000.000. Essa diferenca de
escala atrapalha modelos. Aplicando log:
- log(1000) = 6.9
- log(10000000) = 16.1

De repente a diferenca fica gerenciavel (6.9 a 16.1 em vez de 1mil a 10mi).
**No banco isso e essencial** porque renda, faturamento de empresa PJ, volume
de transacao -- tudo tem cauda longa. Log normaliza.

**2. Log transforma multiplicacao em soma.**
```
log(a x b) = log(a) + log(b)
```
Isso facilita contas e e a base de muitos algoritmos (Naive Bayes usa isso
pra nao ter underflow numerico).

**3. Log lineariza relacoes exponenciais.**
Se algo cresce exponencialmente (juros compostos, crescimento de base de
clientes), aplicar log deixa a relacao reta -- e ai regressao linear funciona.

### As propriedades que voce DECORA (e por que valem)

```
log(a x b) = log(a) + log(b)      multiplicacao vira soma
log(a / b) = log(a) - log(b)      divisao vira subtracao
log(a^n)   = n x log(a)            expoente desce
log(1)     = 0                     (qualquer base^0 = 1)
log_b(b)   = 1                     (b^1 = b)
ln(e^x)    = x                     ln cancela e
e^(ln x)   = x                     e cancela ln
```

### Na sabatina

**Pergunta provavel:** "Por que voce aplicaria log na variavel de faturamento
dos clientes PJ antes de modelar?"

**Resposta modelo:** "Faturamento de PJ tem distribuicao muito assimetrica --
a maioria fatura pouco e poucos faturam muito (cauda longa). Modelos lineares
sofrem com isso porque os outliers de faturamento dominam o ajuste. Aplicando
log, comprimo a escala, a distribuicao fica mais simetrica (mais proxima da
normal), e o modelo trata todos os clientes de forma mais equilibrada. Alem
disso, a interpretacao muda para variacao percentual, que e mais natural para
faturamento."


````python
# DEMONSTRACAO A1 -- por que log ajuda com cauda longa
np.random.seed(42)

# Simular faturamento de clientes PJ (distribuicao lognormal = cauda longa)
faturamento = np.random.lognormal(mean=11, sigma=1.5, size=5000)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Sem log: cauda longa, dificil de ver
axes[0].hist(faturamento, bins=60, color='coral', edgecolor='black')
axes[0].set_xlabel('Faturamento (R$)')
axes[0].set_title('SEM log: cauda longa\nMaioria comprimida na esquerda, outliers dominam')
axes[0].axvline(faturamento.mean(), color='blue', linestyle='--', label=f'Media: R$ {faturamento.mean():,.0f}')
axes[0].axvline(np.median(faturamento), color='green', linestyle='--', label=f'Mediana: R$ {np.median(faturamento):,.0f}')
axes[0].legend()

# Com log: vira quase normal
axes[1].hist(np.log(faturamento), bins=60, color='steelblue', edgecolor='black')
axes[1].set_xlabel('log(Faturamento)')
axes[1].set_title('COM log: quase Normal (simetrica)\nAgora o modelo trata todos de forma equilibrada')
plt.tight_layout(); plt.show()

print('OBSERVE:')
print(f'  Sem log: media (R$ {faturamento.mean():,.0f}) >> mediana (R$ {np.median(faturamento):,.0f})')
print('  Isso indica forte assimetria (cauda direita).')
print()
print('  Com log: distribuicao vira quase simetrica (Normal).')
print('  Modelos lineares assumem normalidade dos residuos -- log ajuda a chegar la.')
````

## A2 -- Exponencial e o numero e

### O que e exponencial

Exponencial e o INVERSO do log. Enquanto log pergunta "quantas vezes multiplico",
exponencial JA multiplica.

```
2^3 = 2 x 2 x 2 = 8        (exponencial)
log_2(8) = 3               (log -- o inverso)
```

### O numero e (2.718...)

e e uma constante matematica, como o pi. Vale aproximadamente 2.718.

**De onde vem?** Imagina que voce tem R$ 1 e ganha 100% de juros ao ano.
- Se paga 1x no fim do ano: R$ 2
- Se paga 2x (50% a cada 6 meses, compondo): R$ 2.25
- Se paga 12x (mensalmente): R$ 2.61
- Se paga infinitas vezes (juros continuo): **R$ 2.718... = e**

e e o resultado de juros compostos infinitamente divididos. Por isso aparece
em qualquer coisa que cresce ou decai continuamente.

### A propriedade magica de e^x

```
A derivada de e^x e o proprio e^x.
```

Nenhuma outra funcao faz isso. Isso torna e^x super conveniente em calculo,
e por isso ele aparece em estatistica (distribuicao normal, exponencial),
ML (funcao sigmoide, softmax) e em qualquer modelagem de crescimento.

### A funcao sigmoide (vai te encontrar em Regressao Logistica)

```
sigma(z) = 1 / (1 + e^(-z))
```

Essa funcao pega qualquer numero (de -infinito a +infinito) e espreme pra
um valor entre 0 e 1. Por isso e perfeita para PROBABILIDADE.

- z muito negativo -> sigma proximo de 0
- z = 0 -> sigma = 0.5
- z muito positivo -> sigma proximo de 1

### Na sabatina

**Pergunta:** "Na regressao logistica, por que usamos a funcao sigmoide?"

**Resposta modelo:** "Porque a saida de uma combinacao linear (w.x + b) pode
ser qualquer numero real, de -infinito a +infinito. Mas eu quero uma
PROBABILIDADE, que tem que estar entre 0 e 1. A sigmoide faz exatamente essa
transformacao: espreme qualquer real para o intervalo [0,1]. Alem disso, ela
e suave e diferenciavel, o que permite otimizar via gradiente."


````python
# DEMONSTRACAO A2 -- a funcao sigmoide
z = np.linspace(-10, 10, 200)
sigmoide = 1 / (1 + np.exp(-z))

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(z, sigmoide, lw=3, color='steelblue')
ax.axhline(0.5, color='gray', linestyle='--', alpha=0.7, label='0.5 (fronteira de decisao)')
ax.axvline(0, color='gray', linestyle=':', alpha=0.5)
ax.fill_between(z, 0, sigmoide, alpha=0.1, color='steelblue')
ax.set_xlabel('z = w.x + b (saida linear, pode ser qualquer numero)')
ax.set_ylabel('sigma(z) = probabilidade (entre 0 e 1)')
ax.set_title('Funcao Sigmoide: transforma qualquer numero em probabilidade')
ax.legend()
plt.tight_layout(); plt.show()

print('Exemplos:')
for zval in [-5, -2, 0, 2, 5]:
    p = 1/(1+np.exp(-zval))
    print(f'  z = {zval:+d}  ->  probabilidade = {p:.3f}')
print()
print('No banco: se z e o score de propensao de um cliente PJ contratar um produto,')
print('a sigmoide converte esse score em probabilidade de 0 a 100%.')
````

## A3 -- Derivada: a ferramenta que faz modelos aprenderem

### O que e, de verdade

Derivada = **velocidade de mudanca** = inclinacao.

Imagina que voce esta dirigindo. Sua POSICAO muda ao longo do tempo.
A derivada da posicao e sua VELOCIDADE (quao rapido a posicao muda).

Em uma funcao f(x), a derivada f'(x) diz: "se eu aumentar x um tiquinho,
quanto f sobe ou desce?"

- Derivada positiva: funcao subindo
- Derivada negativa: funcao descendo
- Derivada ZERO: funcao no topo (maximo) ou no fundo (minimo) -- nao sobe nem desce

**Esse ultimo ponto e a chave do machine learning.** Mais sobre em A4.

### A regra que resolve 80% dos casos

```
Se f(x) = x^n,  entao  f'(x) = n . x^(n-1)
```

Em portugues: "desce o expoente como multiplicador, diminui 1 no expoente".

Exemplos:
- f(x) = x^2  ->  f'(x) = 2x
- f(x) = x^3  ->  f'(x) = 3x^2
- f(x) = 5x^2 ->  f'(x) = 10x  (a constante 5 fica multiplicando)

### Regras adicionais

```
Constante:        d/dx(7) = 0       (constante nao muda)
x sozinho:        d/dx(x) = 1
Soma:             derive cada parte separada
Produto:          (f.g)' = f'.g + f.g'
ln(x):            d/dx[ln(x)] = 1/x
e^x:              d/dx[e^x] = e^x    (o unico que e igual a si mesmo)
```

### Por que "diminuir 1 no expoente" faz sentido

Pensa numa funcao x^2 (uma parabola). Quando x e pequeno, ela cresce devagar.
Quando x e grande, ela cresce rapido. A "velocidade de crescimento" depende
de onde voce esta -- e essa velocidade e 2x (a derivada). Em x=1 a velocidade
e 2; em x=10 a velocidade e 20. Faz sentido: parabola fica cada vez mais ingreme.


````python
# DEMONSTRACAO A3 -- derivada como inclinacao
x = np.linspace(-3, 3, 200)
f = x**2          # funcao
df = 2*x          # derivada

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].plot(x, f, lw=3, color='steelblue', label='f(x) = x^2')
# Reta tangente em x=1
x0 = 1
tangente = f[np.argmin(abs(x-x0))] + 2*x0*(x - x0)
axes[0].plot(x, tangente, '--', color='red', label=f'tangente em x={x0} (inclinacao = {2*x0})')
axes[0].scatter([x0], [x0**2], color='red', s=100, zorder=5)
axes[0].set_ylim(-1, 9)
axes[0].set_xlabel('x'); axes[0].set_ylabel('f(x)')
axes[0].set_title('A derivada e a INCLINACAO da reta tangente')
axes[0].legend()

axes[1].plot(x, df, lw=3, color='green', label="f'(x) = 2x")
axes[1].axhline(0, color='gray', linestyle='--')
axes[1].scatter([0], [0], color='red', s=100, zorder=5, label='derivada=0 em x=0 (minimo de f)')
axes[1].set_xlabel('x'); axes[1].set_ylabel("f'(x)")
axes[1].set_title('Derivada: negativa (desce) | zero (fundo) | positiva (sobe)')
axes[1].legend()
plt.tight_layout(); plt.show()

print('Onde a derivada e ZERO (x=0), a funcao original esta no MINIMO.')
print('Esse e o segredo de como modelos aprendem -- proximo topico.')
````

## A4 -- Por que derivada faz modelos aprenderem (otimizacao)

### O problema central do ML

Todo modelo tem PARAMETROS (pesos). Treinar = achar os pesos que fazem o
modelo errar o MENOS possivel.

"Errar o menos possivel" = MINIMIZAR uma funcao de erro (loss).

E como achar o minimo de uma funcao? **Onde a derivada e zero.**

### A intuicao do "descer a montanha"

Imagina que a funcao de erro e um vale (montanha de cabeca pra baixo).
Voce esta em algum lugar da encosta, vendado, e quer chegar no fundo.

Como faz? Voce sente a INCLINACAO do chao sob seus pes (a derivada) e da
um passo na direcao da descida. Repete ate o chao ficar plano (derivada zero
= fundo do vale).

Isso e o **Gradient Descent** (Descida do Gradiente):

```
peso_novo = peso_velho - taxa_aprendizado x derivada
```

- A derivada diz a direcao da subida
- O sinal MENOS faz voce ir na direcao oposta (descida)
- A taxa de aprendizado (eta) controla o tamanho do passo

### Por que o sinal e MENOS

A derivada aponta pra onde a funcao CRESCE. Como queremos MINIMIZAR (descer),
vamos na direcao OPOSTA -- por isso subtraimos.

Se fosse +, estariamos subindo a montanha (maximizando o erro). Pessimo.

### Na sabatina

**Pergunta:** "Como um modelo de regressao linear 'aprende' os coeficientes?"

**Resposta modelo:** "Ele minimiza uma funcao de erro -- tipicamente a soma
dos quadrados dos residuos. Pra achar o minimo, calculamos a derivada da
funcao de erro em relacao a cada coeficiente e igualamos a zero. Na regressao
linear simples ha solucao fechada (formula direta). Em modelos maiores, usamos
gradient descent: calculamos o gradiente (derivadas) e damos passos iterativos
na direcao que reduz o erro, ate convergir no minimo."

**Pergunta de follow-up provavel:** "E se a taxa de aprendizado for muito alta?"

**Resposta:** "O passo fica grande demais e o modelo pode 'pular' o minimo,
oscilando ou ate divergindo. Taxa muito baixa, por outro lado, faz o treino
ser lento. Por isso existem otimizadores adaptativos como Adam, que ajustam
a taxa automaticamente."


````python
# DEMONSTRACAO A4 -- gradient descent na pratica
# Minimizar f(w) = (w-3)^2 + 2  -- minimo em w=3

def f(w): return (w-3)**2 + 2
def df(w): return 2*(w-3)   # derivada

# Gradient descent
w = 10.0          # comeca longe do minimo
eta = 0.1         # taxa de aprendizado
historico = [w]

for passo in range(30):
    gradiente = df(w)
    w = w - eta * gradiente   # passo na direcao da descida
    historico.append(w)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# A funcao e o caminho percorrido
w_range = np.linspace(-2, 11, 200)
axes[0].plot(w_range, f(w_range), lw=2, color='steelblue', label='f(w) = (w-3)^2 + 2')
axes[0].scatter(historico, [f(w) for w in historico], color='red', s=30, zorder=5, alpha=0.6)
axes[0].scatter([3], [f(3)], color='green', s=200, marker='*', zorder=6, label='minimo real (w=3)')
axes[0].set_xlabel('w (parametro)'); axes[0].set_ylabel('erro')
axes[0].set_title('Gradient Descent descendo ate o minimo')
axes[0].legend()

# Convergencia
axes[1].plot(historico, marker='o', color='coral')
axes[1].axhline(3, color='green', linestyle='--', label='valor otimo (3)')
axes[1].set_xlabel('iteracao'); axes[1].set_ylabel('valor de w')
axes[1].set_title('Convergencia: w vai de 10 ate 3')
axes[1].legend()
plt.tight_layout(); plt.show()

print(f'Comecou em w=10, convergiu para w={historico[-1]:.4f} (otimo=3)')
print('Cada passo: w = w - 0.1 * derivada. O modelo APRENDEU o minimo sozinho.')
````

## A5 -- Matrizes: por que ML vive delas

### Por que matrizes existem em ML

Um dataset E uma matriz. Cada linha = um cliente. Cada coluna = uma feature.

```
            idade  renda  tempo_banco
cliente 1 [  35    5000     3        ]
cliente 2 [  42    8000     7        ]
cliente 3 [  28    3000     1        ]
```

Isso e uma matriz 3x3 (3 clientes, 3 features). Todo modelo opera sobre
essa estrutura. Por isso matriz e a linguagem nativa de ML.

### Multiplicacao de matrizes (a regra critica)

```
(m x n) . (n x p) = (m x p)
```

As dimensoes do MEIO tem que ser iguais. O resultado pega as das PONTAS.

Por que isso importa? Porque a previsao de um modelo linear E uma
multiplicacao de matrizes:

```
Previsoes = X . pesos
(n clientes x 1) = (n clientes x f features) . (f features x 1)
```

Se as dimensoes nao batem, o modelo nem roda. Erro classico de iniciante.

### Determinante e a "decuabilidade" (invertibilidade)

O determinante e um numero que diz se a matriz e "saudavel" ou "problematica".

```
det = 0  ->  matriz SINGULAR (problematica, nao tem inversa)
det != 0 ->  matriz invertivel (saudavel)
```

**Por que isso importa em regressao?** A formula da regressao linear envolve
inverter a matriz (X^T . X). Se essa matriz tem determinante zero, NAO DA
PRA INVERTER -- a regressao falha ou da resultados malucos.

**Quando isso acontece?** Quando duas features sao perfeitamente
correlacionadas (multicolinearidade). Por exemplo: "salario em reais" e
"salario em milhares de reais" -- uma e so a outra dividida por 1000. Sao
redundantes. A matriz fica singular.

**Esse e exatamente o ponto que voce mencionou:** se acontece multicolinearidade,
a matriz X^T.X fica singular (det=0), nao da pra inverter, e o OLS quebra.
A solucao e Ridge (que veremos), que ADICIONA um valorzinho na diagonal
forcando a matriz a ser invertivel de novo.

### Na sabatina

**Pergunta:** "O que acontece com a regressao linear se duas variaveis sao
perfeitamente colineares?"

**Resposta modelo:** "A matriz X^T.X fica singular -- determinante zero --
e nao pode ser invertida. Como a solucao do OLS depende dessa inversao
((X^T.X)^-1 . X^T.y), o modelo nao consegue estimar coeficientes unicos:
existem infinitas combinacoes que dao o mesmo resultado. Na pratica, o
software ou da erro ou retorna coeficientes instaveis e gigantes. A solucao
e remover uma das variaveis redundantes, ou usar regularizacao Ridge, que
adiciona lambda na diagonal de X^T.X, tornando-a invertivel novamente e
estabilizando os coeficientes."


````python
# DEMONSTRACAO A5 -- multicolinearidade quebra a regressao
import numpy as np
np.random.seed(42)

n = 500
x1 = np.random.normal(50, 10, n)
x2 = x1 * 2          # x2 e PERFEITAMENTE colinear com x1 (o dobro)
y = 3*x1 + np.random.normal(0, 5, n)

# Tentar regressao com X^T X
X = np.column_stack([np.ones(n), x1, x2])  # intercepto + x1 + x2
XtX = X.T @ X

det = np.linalg.det(XtX)
print('=== MULTICOLINEARIDADE PERFEITA ===')
print(f'  x2 = 2 * x1 (perfeitamente colinear)')
print(f'  Determinante de X^T.X: {det:.2e}')
print(f'  (praticamente ZERO -> matriz singular -> nao invertivel)')
print()

# Mostrar que a inversao falha ou da numeros instaveis
try:
    inv = np.linalg.inv(XtX)
    coefs = inv @ X.T @ y
    print(f'  Coeficientes estimados: {coefs}')
    print('  (Note como ficam instaveis/gigantes ou o numero de condicao explode)')
    cond = np.linalg.cond(XtX)
    print(f'  Numero de condicao: {cond:.2e} (saudavel seria < 1000)')
except np.linalg.LinAlgError:
    print('  ERRO: matriz singular, impossivel inverter!')
print()

# Agora SEM colinearidade
x2_bom = np.random.normal(30, 8, n)  # variavel independente
X_bom = np.column_stack([np.ones(n), x1, x2_bom])
XtX_bom = X_bom.T @ X_bom
print('=== SEM MULTICOLINEARIDADE ===')
print(f'  Determinante: {np.linalg.det(XtX_bom):.2e} (longe de zero -> saudavel)')
print(f'  Numero de condicao: {np.linalg.cond(XtX_bom):.1f} (baixo -> estavel)')
print()
print('LICAO: features redundantes quebram a regressao. Por isso checamos VIF')
print('(Variance Inflation Factor) e usamos Ridge quando ha colinearidade.')
````

## B1 -- Media, variancia, desvio: e quando cada um ENGANA

### Media (o centro)

Soma tudo, divide pela quantidade. Todo mundo sabe. Mas a media TEM UM PROBLEMA:

**A media e enganada por outliers.**

Exemplo: salarios de 5 pessoas: 2000, 2500, 3000, 3500, 100000.
- Media = 22.200 (puxada pelo CEO de 100k)
- Mediana = 3000 (o valor do meio, ignora o extremo)

A media diz "salario tipico = 22 mil", o que e MENTIRA. A mediana (3 mil) e
muito mais honesta. **Por isso, em dados assimetricos (renda, faturamento PJ),
use mediana.**

### Variancia (o espalhamento)

Mede o quao espalhados os dados estao em torno da media.

```
Variancia = media dos (desvios ao quadrado)
          = (1/n) . soma( (xi - media)^2 )
```

**Por que ao quadrado?** Duas razoes:
1. Pra desvios negativos e positivos nao se cancelarem (quadrado sempre positivo)
2. Pra penalizar mais os desvios grandes (outliers pesam mais)

### Desvio-padrao (variancia em escala original)

```
Desvio-padrao = raiz quadrada da variancia
```

**Por que tirar a raiz?** Porque a variancia esta em "unidades ao quadrado"
(reais ao quadrado??). O desvio-padrao volta pra unidade original (reais),
ficando interpretavel.

### Na sabatina

**Pergunta:** "Voce esta analisando faturamento de clientes PJ. Usaria media
ou mediana pra reportar o valor tipico? Por que?"

**Resposta modelo:** "Mediana. Faturamento de PJ e fortemente assimetrico --
poucos clientes grandes faturam muito e distorcem a media pra cima. A mediana
representa o cliente tipico de forma mais fiel, porque e robusta a outliers.
Eu reportaria a mediana como valor central e talvez os percentis (P25, P75)
pra dar nocao da dispersao, em vez de media e desvio-padrao que seriam
distorcidos pela cauda."


````python
# DEMONSTRACAO B1 -- media vs mediana com outliers
np.random.seed(42)

# Faturamento de PJ: maioria pequena, alguns gigantes
faturamento = np.concatenate([
    np.random.normal(50000, 15000, 950),   # 950 PMEs
    np.random.normal(5000000, 1000000, 50) # 50 grandes empresas
])
faturamento = faturamento[faturamento > 0]

media = faturamento.mean()
mediana = np.median(faturamento)

fig, ax = plt.subplots(figsize=(11, 5))
ax.hist(faturamento, bins=60, color='steelblue', edgecolor='black', alpha=0.7)
ax.axvline(media, color='red', linestyle='--', lw=2, label=f'Media: R$ {media:,.0f}')
ax.axvline(mediana, color='green', linestyle='--', lw=2, label=f'Mediana: R$ {mediana:,.0f}')
ax.set_xlabel('Faturamento (R$)'); ax.set_ylabel('Frequencia')
ax.set_title('Media vs Mediana em faturamento PJ (assimetrico)')
ax.legend()
plt.tight_layout(); plt.show()

print(f'Media:   R$ {media:,.0f}  <- puxada pelos gigantes')
print(f'Mediana: R$ {mediana:,.0f}  <- representa o cliente tipico')
print()
print(f'A media e {media/mediana:.1f}x maior que a mediana.')
print('Reportar a media enganaria a diretoria sobre o cliente PJ tipico.')
````

## B2 -- Distribuicoes: quando usar cada uma

Distribuicao = o "formato" que os dados tomam. Saber identificar qual usar
e diferencial na prova.

### Normal (a curva de sino)

- Formato: sino simetrico
- Quando aparece: altura, peso, erro de medicao, MEDIA de qualquer coisa (TCL)
- Parametros: media (centro) e desvio-padrao (largura)
- Regra 68-95-99.7: 68% dos dados em 1 desvio, 95% em 2, 99.7% em 3

### Binomial (contagem de sucessos)

- Quando: numero de sucessos em n tentativas SIM/NAO independentes
- Exemplo banco: quantos de 100 clientes PJ vao contratar o produto
- Parametros: n (tentativas) e p (probabilidade de sucesso)
- Media = n.p, Variancia = n.p.(1-p)

### Poisson (eventos raros no tempo)

- Quando: numero de eventos em um intervalo, quando sao raros
- Exemplo banco: quantas reclamacoes chegam por hora; quantas fraudes por dia
- Parametro: lambda (taxa media)
- Curiosidade: media = variancia = lambda

### Exponencial (tempo entre eventos)

- Quando: tempo de espera ate o proximo evento
- Exemplo banco: tempo entre dois atendimentos; tempo ate um cliente cancelar
- Relacionada a Poisson (se eventos sao Poisson, o tempo entre eles e exponencial)

### Na sabatina

**Pergunta:** "Voce quer modelar o numero de reclamacoes que chegam ao
atendimento PJ por hora. Que distribuicao usaria?"

**Resposta modelo:** "Poisson. Reclamacoes sao eventos discretos, relativamente
raros, que chegam ao longo do tempo de forma independente. A Poisson e
exatamente pra contagem de eventos em um intervalo fixo. O parametro lambda
seria a taxa media de reclamacoes por hora. Se eu quisesse modelar o TEMPO
ENTRE reclamacoes, ai usaria a Exponencial, que e a distribuicao complementar."


````python
# DEMONSTRACAO B2 -- as 4 distribuicoes principais
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Normal
x = np.linspace(-4, 4, 200)
axes[0,0].plot(x, stats.norm.pdf(x), lw=3, color='steelblue')
axes[0,0].fill_between(x, stats.norm.pdf(x), alpha=0.2)
axes[0,0].set_title('NORMAL: altura, peso, medias\nFormato de sino simetrico')

# Binomial
k = np.arange(0, 101)
axes[0,1].bar(k, stats.binom.pmf(k, 100, 0.3), color='coral')
axes[0,1].set_title('BINOMIAL: sucessos em N tentativas\nEx: quantos de 100 PJ contratam (p=0.3)')
axes[0,1].set_xlim(15, 45)

# Poisson
k2 = np.arange(0, 20)
axes[1,0].bar(k2, stats.poisson.pmf(k2, 5), color='green')
axes[1,0].set_title('POISSON: eventos raros por intervalo\nEx: reclamacoes/hora (lambda=5)')

# Exponencial
x2 = np.linspace(0, 5, 200)
axes[1,1].plot(x2, stats.expon.pdf(x2), lw=3, color='purple')
axes[1,1].fill_between(x2, stats.expon.pdf(x2), alpha=0.2, color='purple')
axes[1,1].set_title('EXPONENCIAL: tempo entre eventos\nEx: tempo ate proximo atendimento')
plt.tight_layout(); plt.show()

print('Resumo de quando usar:')
print('  NORMAL      -> medidas continuas simetricas, ou MEDIAS (TCL)')
print('  BINOMIAL    -> contagem de sucessos SIM/NAO em N tentativas')
print('  POISSON     -> contagem de eventos raros em um intervalo')
print('  EXPONENCIAL -> tempo de espera entre eventos')
````

## B3 -- Probabilidade e Bayes: o raciocinio do banco

### Probabilidade basica

```
P(evento) = casos favoraveis / casos possiveis
```

### Eventos independentes vs dependentes

**Independentes** (um nao afeta o outro):
```
P(A e B) = P(A) x P(B)
```

**Dependentes** (um afeta o outro):
```
P(A e B) = P(A) x P(B | A)
```
O `P(B|A)` le-se "probabilidade de B DADO que A aconteceu".

### Teorema de Bayes (o mais importante pra banco)

```
P(A | B) = P(B | A) x P(A) / P(B)
```

Bayes permite INVERTER probabilidades condicionais. E a base de:
- Deteccao de fraude
- Score de credito
- Filtros de spam
- Diagnostico (positivo no teste -> tem a condicao?)

### A pegadinha que SEMPRE cai: base rara

Se um evento e RARO, mesmo um teste bom gera muitos falsos positivos.

**Tecnica infalivel: imagine 10.000 casos e conte.**

Exemplo: fraude e 1% das transacoes. Modelo pega 90% das fraudes mas
sinaliza 5% das legitimas como fraude. Uma transacao foi sinalizada.
Qual a chance de ser fraude?

```
10.000 transacoes:
  Fraudes (1%):     100  ->  sinalizadas: 100 x 0.90 = 90
  Legitimas (99%):  9900 ->  sinalizadas: 9900 x 0.05 = 495

Total sinalizadas: 90 + 495 = 585
Fraudes entre elas: 90

P(fraude | sinalizada) = 90/585 = 15.4%
```

Apesar do modelo "90% bom", so 15% dos alertas sao fraude real! Porque
fraude e rara. Por isso bancos tem analistas revisando alertas.

### Na sabatina

**Pergunta:** "Seu modelo de fraude tem 95% de recall, mas o time reclama de
muitos falsos positivos. Como voce explica?"

**Resposta modelo:** "Isso e o paradoxo da base rara. Como fraude e um evento
muito raro -- digamos 1% -- mesmo um modelo com alto recall gera muitos falsos
positivos em termos absolutos, porque os 5% de erro nas transacoes legitimas
representam um numero enorme (ja que legitimas sao 99% do volume). Mesmo com
recall de 95%, a precisao pode ser baixa. A solucao nao e so 'melhorar o
modelo', mas ajustar o threshold pra equilibrar precisao e recall conforme o
custo de cada erro, e ter uma camada de revisao humana pros alertas."


````python
# DEMONSTRACAO B3 -- o paradoxo da base rara (Bayes)
def bayes_fraude(taxa_fraude, recall, falso_positivo, n=100000):
    fraudes = int(n * taxa_fraude)
    legitimas = n - fraudes
    sinaliz_fraude = fraudes * recall
    sinaliz_legit = legitimas * falso_positivo
    total_sinaliz = sinaliz_fraude + sinaliz_legit
    precisao = sinaliz_fraude / total_sinaliz if total_sinaliz > 0 else 0
    return precisao

print('=== COMO A RARIDADE DA FRAUDE AFETA A PRECISAO ===')
print('(modelo fixo: recall=95%, falso positivo=5%)')
print()
print(f'{"Taxa de fraude":<18} {"Precisao do alerta":<20}')
print('-'*40)
for taxa in [0.001, 0.005, 0.01, 0.05, 0.10, 0.30]:
    prec = bayes_fraude(taxa, 0.95, 0.05)
    print(f'{taxa*100:>6.1f}%            {prec*100:>6.1f}%')
print()
print('LICAO: com fraude a 0.1%, apenas ~2% dos alertas sao fraude real!')
print('Mesmo modelo, mas a raridade destroi a precisao.')
print('Por isso threshold e revisao humana sao essenciais em fraude.')
````

## B4 -- Testes de Hipotese: o que p-valor REALMENTE significa

### O problema que testes resolvem

Voce fez uma mudanca no atendimento PJ e o NPS subiu de 7.0 pra 7.3.
Pergunta: essa melhora e REAL ou foi sorte (variacao aleatoria)?

Teste de hipotese responde isso.

### O esquema mental

- **H0 (hipotese nula):** "nada mudou, a diferenca foi acaso". O ceticismo.
- **H1 (hipotese alternativa):** "a mudanca e real". O que voce quer provar.

Voce assume H0 verdadeira e pergunta: "se nada mudou de verdade, qual a chance
de eu ter observado uma diferenca TAO grande quanto essa, so por acaso?"

Essa chance e o **p-valor**.

### O que p-valor E e o que NAO E

**p-valor E:** a probabilidade de observar um resultado tao extremo quanto o
seu, SE a hipotese nula fosse verdadeira.

**p-valor NAO E:**
- NAO e a probabilidade de H0 ser verdadeira
- NAO e a probabilidade de voce estar errado
- NAO mede o tamanho do efeito

**Essa distincao cai em prova e em sabatina.** Decora:

> p-valor pequeno (< 0.05) = "seria muito improvavel ver isso por acaso, entao
> rejeito H0 e acredito que a mudanca e real".

### O nivel de significancia (alpha)

Geralmente 0.05 (5%). E o limite: se p < 0.05, rejeitamos H0.

Significa que aceitamos 5% de chance de erro tipo I (dizer que tem efeito
quando nao tem).

### Erros tipo I e tipo II

- **Erro tipo I (falso positivo):** rejeitar H0 quando ela e verdadeira.
  "Dizer que a mudanca funcionou quando nao funcionou." Probabilidade = alpha.
- **Erro tipo II (falso negativo):** nao rejeitar H0 quando ela e falsa.
  "Nao detectar uma melhora que era real." Probabilidade = beta.

- **Poder do teste = 1 - beta:** probabilidade de detectar um efeito que existe.

### Na sabatina

**Pergunta:** "Voce rodou um teste e deu p-valor de 0.03. O que isso significa?"

**Resposta modelo:** "Significa que, SE nao houvesse efeito real (hipotese nula
verdadeira), haveria apenas 3% de chance de observar uma diferenca tao grande
quanto a que observei. Como 3% e menor que o nivel de significancia usual de
5%, eu rejeito a hipotese nula e concluo que ha evidencia de um efeito real.
Importante: o p-valor NAO me diz a probabilidade de a hipotese nula ser
verdadeira, nem o tamanho do efeito -- pra isso eu olharia o intervalo de
confianca e a magnitude observada."

**Follow-up provavel:** "p-valor baixo significa que o efeito e grande?"

**Resposta:** "Nao. p-valor baixo significa que o efeito e estatisticamente
DETECTAVEL, nao que e GRANDE. Com amostra enorme, ate um efeito minusculo e
irrelevante na pratica pode dar p-valor baixo. Por isso sempre reporto tambem
o tamanho do efeito e o intervalo de confianca, nao so o p-valor."


````python
# DEMONSTRACAO B4 -- teste de hipotese na pratica (A/B test de NPS)
from scipy import stats
np.random.seed(42)

# Grupo controle (atendimento antigo) e tratamento (novo)
controle = np.random.normal(7.0, 1.5, 500)
tratamento = np.random.normal(7.3, 1.5, 500)  # efeito real de +0.3

# Teste t de duas amostras independentes
t_stat, p_valor = stats.ttest_ind(tratamento, controle)

print('=== A/B TEST: novo atendimento PJ melhora NPS? ===')
print()
print(f'  NPS medio controle:    {controle.mean():.3f}')
print(f'  NPS medio tratamento:  {tratamento.mean():.3f}')
print(f'  Diferenca observada:   {tratamento.mean() - controle.mean():+.3f}')
print()
print(f'  Estatistica t:  {t_stat:.3f}')
print(f'  P-valor:        {p_valor:.4f}')
print()
if p_valor < 0.05:
    print('  p < 0.05 -> Rejeitamos H0. A melhora no NPS e estatisticamente significativa.')
else:
    print('  p >= 0.05 -> Nao rejeitamos H0. Pode ter sido acaso.')
print()

# Intervalo de confianca da diferenca
diff = tratamento.mean() - controle.mean()
se = np.sqrt(tratamento.var()/len(tratamento) + controle.var()/len(controle))
ic_low, ic_high = diff - 1.96*se, diff + 1.96*se
print(f'  IC 95% da diferenca: [{ic_low:+.3f}, {ic_high:+.3f}]')
print('  Como o IC nao inclui zero, confirmamos que o efeito e real.')
````

## B5 -- Intervalo de Confianca: a parte que TODOS erram

### O que e

Em vez de dizer "o NPS medio e 7.3", voce diz "o NPS medio esta entre 7.1 e
7.5, com 95% de confianca". O intervalo captura a INCERTEZA da sua estimativa.

### A formula (caso da media)

```
IC = estimativa +/- (valor critico) x (erro padrao)
IC 95% = media +/- 1.96 x (desvio / raiz(n))
```

O 1.96 vem da Normal (95% dos dados estao a 1.96 desvios da media).

### A interpretacao CORRETA (cai em prova)

**ERRADO:** "ha 95% de probabilidade de a media verdadeira estar nesse intervalo".

**CERTO:** "se eu repetisse o experimento muitas vezes e construisse um IC a
cada vez, 95% desses intervalos conteriam a media verdadeira".

A diferenca e sutil mas importante: a media verdadeira e um numero FIXO (nao
aleatorio). O que e aleatorio e o INTERVALO (muda a cada amostra). Entao nao
faz sentido falar em "probabilidade da media estar no intervalo" -- ela esta
ou nao esta. O que tem 95% e o PROCESSO de construir intervalos.

### Por que IC e melhor que so o p-valor

O p-valor diz SE existe efeito. O IC diz QUANTO e com que incerteza.

"NPS subiu, p<0.05" -> ok, subiu. Mas subiu quanto? Pode ser +0.01 (irrelevante)
ou +2.0 (enorme). O IC responde: "subiu entre +0.1 e +0.5". Muito mais util
pra decisao de negocio.

### Largura do intervalo

- Amostra maior -> intervalo mais estreito (mais certeza)
- Mais variabilidade nos dados -> intervalo mais largo (menos certeza)
- Mais confianca exigida (99% vs 95%) -> intervalo mais largo

### Na sabatina

**Pergunta:** "Qual a diferenca entre reportar um p-valor e um intervalo de
confianca?"

**Resposta modelo:** "O p-valor responde uma pergunta binaria: o efeito e
estatisticamente significativo ou nao. Mas ele nao diz o tamanho do efeito nem
a precisao da estimativa. O intervalo de confianca da as duas coisas: mostra a
faixa provavel do valor real e, pela largura, indica quao precisa e a
estimativa. Pra decisao de negocio, o IC e muito mais informativo. Por exemplo,
saber que uma campanha aumenta conversao 'entre 2% e 8%' orienta melhor do que
so saber que 'o aumento e significativo'."


````python
# DEMONSTRACAO B5 -- intervalo de confianca e tamanho da amostra
np.random.seed(42)

fig, ax = plt.subplots(figsize=(11, 5))

tamanhos = [30, 100, 500, 2000]
cores = ['red', 'orange', 'green', 'blue']

for i, n in enumerate(tamanhos):
    amostra = np.random.normal(7.3, 1.5, n)
    media = amostra.mean()
    se = amostra.std() / np.sqrt(n)
    ic_low, ic_high = media - 1.96*se, media + 1.96*se
    ax.errorbar(media, i, xerr=1.96*se, fmt='o', color=cores[i],
                capsize=8, capthick=2, markersize=10,
                label=f'n={n}: IC = [{ic_low:.2f}, {ic_high:.2f}] (largura {2*1.96*se:.2f})')

ax.axvline(7.3, color='black', linestyle='--', label='valor real (7.3)')
ax.set_yticks(range(len(tamanhos)))
ax.set_yticklabels([f'n={n}' for n in tamanhos])
ax.set_xlabel('NPS estimado')
ax.set_title('Quanto MAIOR a amostra, mais ESTREITO o intervalo (mais certeza)')
ax.legend(loc='upper right', fontsize=9)
plt.tight_layout(); plt.show()

print('OBSERVE: o intervalo encolhe conforme a amostra cresce.')
print('Com n=30 o IC e largo (muita incerteza).')
print('Com n=2000 o IC e estreito (estimativa precisa).')
print()
print('Por isso amostras maiores dao conclusoes mais confiaveis.')
````

## B6 -- Correlacao vs Causalidade: o coracao da vaga

Esse e o tema central da vaga de Causal AI. Vou aprofundar na Parte D, mas o
fundamento estatistico vem aqui.

### Correlacao

Mede o quanto duas variaveis "andam juntas". Vai de -1 a +1.
- +1: quando uma sobe, a outra sobe (perfeitamente)
- -1: quando uma sobe, a outra desce (perfeitamente)
- 0: nao ha relacao linear

### A frase que voce repete pra sempre

> Correlacao NAO implica causalidade.

Duas variaveis podem andar juntas sem uma causar a outra. Tres motivos:

**1. Confundidor (causa comum):** uma terceira variavel causa as duas.
Exemplo: vendas de sorvete e afogamentos sobem juntos. Sorvete nao causa
afogamento. O CALOR (confundidor) causa os dois.

**2. Causalidade reversa:** voce acha que A causa B, mas e B que causa A.

**3. Coincidencia:** com dados suficientes, qualquer coisa correlaciona com
qualquer coisa por acaso.

### O exemplo do banco (vai cair na sabatina)

Voce observa: clientes que LIGAM pro banco tem NPS mais baixo. Correlacao
forte e negativa.

Conclusao ingenua: "ligacao causa insatisfacao, vamos reduzir ligacoes".

**ERRADO.** A INSATISFACAO PREVIA e o confundidor: cliente insatisfeito liga
mais (pra reclamar) E da NPS baixo. Reduzir ligacoes nao resolve a insatisfacao.

### Os dois tipos de correlacao

- **Pearson:** mede relacao LINEAR
- **Spearman:** mede relacao MONOTONICA (inclui nao-linear, desde que sempre
  cresca ou sempre decresca)

Se Pearson e baixo mas Spearman e alto, ha relacao nao-linear monotonica.

### Na sabatina

**Pergunta:** "Voce encontra que clientes que usam o app tem NPS 1.5 ponto
maior. Pode recomendar forcar todos a usar o app pra subir o NPS?"

**Resposta modelo:** "Nao posso concluir isso direto, porque e correlacao, nao
necessariamente causalidade. Clientes que JA escolhem usar o app provavelmente
sao mais engajados, mais jovens digitalmente, ou tem outro perfil que tambem
eleva o NPS -- isso e confundimento por autosselecao. Forcar quem nao quer usar
o app nao muda essas caracteristicas e pode ate gerar friccao, baixando o NPS.
Pra estabelecer causalidade eu precisaria de um experimento (A/B test) ou, se
nao for possivel, de metodos causais observacionais como propensity score ou
diff-in-diff, controlando os confundidores."


````python
# DEMONSTRACAO B6 -- confundidor cria correlacao enganosa
np.random.seed(42)
n = 1000

# Confundidor: insatisfacao previa (nao observada diretamente)
insatisfacao = np.random.normal(0, 1, n)

# Insatisfacao causa AMBOS: mais ligacoes e menor NPS
ligacoes = 3 + 2*insatisfacao + np.random.normal(0, 0.5, n)
nps = 8 - 1.5*insatisfacao + np.random.normal(0, 0.5, n)

# Correlacao observada entre ligacoes e NPS
corr = np.corrcoef(ligacoes, nps)[0,1]

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].scatter(ligacoes, nps, alpha=0.3, s=10, color='coral')
axes[0].set_xlabel('Numero de ligacoes ao banco')
axes[0].set_ylabel('NPS')
axes[0].set_title(f'Correlacao observada: {corr:.2f}\n"Ligar parece baixar o NPS" (ENGANOSO)')

# Mostrar o confundidor
sc = axes[1].scatter(ligacoes, nps, c=insatisfacao, cmap='RdYlGn_r', alpha=0.5, s=10)
axes[1].set_xlabel('Numero de ligacoes')
axes[1].set_ylabel('NPS')
axes[1].set_title('Cor = insatisfacao previa (o CONFUNDIDOR)\nEla causa AMBOS')
plt.colorbar(sc, ax=axes[1], label='insatisfacao')
plt.tight_layout(); plt.show()

print(f'Correlacao ligacoes-NPS: {corr:.2f} (forte e negativa)')
print()
print('Mas LIGAR nao CAUSA NPS baixo. A insatisfacao previa causa os dois:')
print('  insatisfacao -> mais ligacoes (pra reclamar)')
print('  insatisfacao -> menor NPS')
print()
print('Reduzir ligacoes a forca nao resolveria a insatisfacao subjacente.')
print('Esse e o erro classico de confundir correlacao com causalidade.')
````

## C1 -- Regressao Linear: a base de tudo

### O que e

Tenta achar a melhor RETA (ou plano) que descreve a relacao entre as features
(X) e o alvo (Y).

```
Y = b0 + b1.X1 + b2.X2 + ... + erro
```

- b0: intercepto (valor de Y quando todas as X sao zero)
- b1, b2...: coeficientes (quanto Y muda quando cada X aumenta 1 unidade)

### Interpretacao dos coeficientes (cai na sabatina)

Se `b1 = 500` num modelo prevendo gasto: "mantendo as outras variaveis
constantes, cada unidade a mais em X1 aumenta o gasto previsto em R$ 500".

A frase **"mantendo as outras constantes"** (ceteris paribus) e crucial.

### Quando usar

- Relacao aproximadamente linear entre X e Y
- Alvo continuo (preco, faturamento, score)
- Quando voce quer INTERPRETABILIDADE (entender o efeito de cada variavel)

### Quando NAO usar

- Relacao claramente nao-linear (use arvores ou transforme as variaveis)
- Alvo categorico/binario (use regressao logistica)
- Muitos outliers (OLS e sensivel; considere regressao robusta)

### Na sabatina

**Pergunta:** "Por que regressao linear, sendo tao simples, ainda e tao usada
no banco?"

**Resposta modelo:** "Por causa da interpretabilidade e da robustez regulatoria.
Em banco, frequentemente preciso EXPLICAR uma decisao -- por que um cliente teve
o credito negado, por exemplo. A regressao linear (ou logistica) da um
coeficiente claro por variavel, que eu consigo justificar pro regulador e pro
cliente. Modelos black-box como XGBoost podem ter performance melhor, mas a
explicabilidade da regressao costuma valer mais em contextos sensiveis. Alem
disso, ela e estavel, rapida de treinar, e serve de baseline pra comparar
modelos mais complexos."


````python
# DEMONSTRACAO C1 -- regressao linear e interpretacao
from sklearn.linear_model import LinearRegression
np.random.seed(42)
n = 500

# Dados: prever gasto mensal do cliente PJ
faturamento = np.random.normal(100, 30, n)       # em milhares
tempo_banco = np.random.normal(5, 2, n)           # anos
gasto = 200 + 3*faturamento + 50*tempo_banco + np.random.normal(0, 50, n)

X = np.column_stack([faturamento, tempo_banco])
modelo = LinearRegression().fit(X, gasto)

print('=== REGRESSAO LINEAR: prever gasto do cliente PJ ===')
print()
print(f'  Intercepto (b0):           R$ {modelo.intercept_:.2f}')
print(f'  Coef. faturamento (b1):    R$ {modelo.coef_[0]:.2f}')
print(f'  Coef. tempo_banco (b2):    R$ {modelo.coef_[1]:.2f}')
print()
print('INTERPRETACAO (ceteris paribus):')
print(f'  Cada R$ 1mil a mais de faturamento -> +R$ {modelo.coef_[0]:.2f} de gasto')
print(f'  Cada ano a mais de relacionamento -> +R$ {modelo.coef_[1]:.2f} de gasto')
print()
from sklearn.metrics import r2_score
r2 = r2_score(gasto, modelo.predict(X))
print(f'  R^2 = {r2:.3f} -> o modelo explica {r2*100:.1f}% da variacao do gasto')
````

## C2 -- Residuos e por que minimizar QUADRADOS

### O que e um residuo

Residuo = erro de previsao = (valor real - valor previsto).

```
residuo_i = y_i - y_previsto_i
```

Se o modelo previu gasto de R$ 500 e o real foi R$ 530, o residuo e +30.

### Por que minimizar a SOMA DOS QUADRADOS dos residuos?

A regressao escolhe os coeficientes que minimizam:
```
SSE = soma( (y_i - y_previsto_i)^2 )
```

Por que ao quadrado e nao o valor absoluto?

**1. Pra erros positivos e negativos nao se cancelarem.** Se voce somasse os
residuos direto, um erro de +30 e outro de -30 dariam zero, fingindo que o
modelo e perfeito. Ao quadrado, ambos viram positivos.

**2. Pra penalizar MAIS os erros grandes.** Um erro de 10 vira 100. Um erro de
20 vira 400 (4x mais, nao 2x). Isso forca o modelo a evitar erros grandes.

**3. Matematica conveniente.** A funcao quadratica e suave e tem derivada
simples, o que permite achar o minimo com formula fechada (ou gradient descent
facil).

### O custo dessa escolha

Como erros grandes sao muito penalizados, OLS e SENSIVEL A OUTLIERS. Um unico
ponto muito fora pode entortar toda a reta. Por isso, com muitos outliers,
consideramos alternativas (regressao robusta, ou MAE em vez de MSE).

### Na sabatina

**Pergunta:** "Por que a regressao minimiza o erro ao quadrado e nao o erro
absoluto?"

**Resposta modelo:** "Tres razoes. Primeiro, o quadrado evita que erros
positivos e negativos se cancelem. Segundo, penaliza desproporcionalmente os
erros grandes, o que e desejavel quando erros grandes sao especialmente ruins.
Terceiro, a funcao quadratica e diferenciavel e convexa, o que da uma solucao
analitica fechada e otimizacao estavel. O custo dessa escolha e a sensibilidade
a outliers -- se isso for um problema, eu usaria o erro absoluto (MAE), que e
mais robusto, ou regressao robusta como Huber."


````python
# DEMONSTRACAO C2 -- residuos e sensibilidade a outliers
from sklearn.linear_model import LinearRegression
np.random.seed(42)

x = np.linspace(0, 10, 50)
y = 2*x + 3 + np.random.normal(0, 1, 50)

# Versao com outlier
y_out = y.copy()
y_out[25] = 50  # outlier gigante

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Sem outlier
m1 = LinearRegression().fit(x.reshape(-1,1), y)
axes[0].scatter(x, y, alpha=0.6, color='steelblue')
axes[0].plot(x, m1.predict(x.reshape(-1,1)), 'r-', lw=2, label=f'reta (coef={m1.coef_[0]:.2f})')
# Mostrar residuos como linhas verticais
for xi, yi in zip(x, y):
    axes[0].plot([xi, xi], [yi, m1.predict([[xi]])[0]], 'gray', alpha=0.3, lw=0.5)
axes[0].set_title('Residuos = distancias verticais aos pontos')
axes[0].legend()

# Com outlier
m2 = LinearRegression().fit(x.reshape(-1,1), y_out)
axes[1].scatter(x, y_out, alpha=0.6, color='steelblue')
axes[1].scatter([x[25]], [y_out[25]], color='red', s=150, zorder=5, label='outlier')
axes[1].plot(x, m1.predict(x.reshape(-1,1)), 'g--', lw=2, label=f'reta SEM outlier (coef={m1.coef_[0]:.2f})')
axes[1].plot(x, m2.predict(x.reshape(-1,1)), 'r-', lw=2, label=f'reta COM outlier (coef={m2.coef_[0]:.2f})')
axes[1].set_title('Um outlier ENTORTA a reta (OLS e sensivel)')
axes[1].legend()
plt.tight_layout(); plt.show()

print(f'Coeficiente SEM outlier: {m1.coef_[0]:.2f}')
print(f'Coeficiente COM outlier: {m2.coef_[0]:.2f}')
print('Um unico ponto extremo mudou o coeficiente -- porque o erro e ao quadrado.')
````

## C3 -- Regularizacao: Ridge, Lasso, ElasticNet (A DECISAO)

Esse e o tema que voce pediu pra aprofundar. Vou explicar cada um, POR QUE
existem, e QUANDO escolher cada.

### O problema que regularizacao resolve

Quando o modelo tem muitas features, ou features correlacionadas, ele pode:
- Ficar com coeficientes gigantes e instaveis
- Fazer OVERFITTING (decorar o treino, falhar no teste)

Regularizacao adiciona uma PENALIDADE no tamanho dos coeficientes, forcando
eles a ficarem menores e mais estaveis.

### Ridge (L2) -- "encolhe todos"

```
Minimiza: SSE + lambda . soma(coef^2)
```

A penalidade e a soma dos coeficientes AO QUADRADO. 

**O que faz:** encolhe todos os coeficientes em direcao a zero, mas NUNCA
zera completamente. Todas as features continuam no modelo, so que com peso menor.

**Quando usar:**
- Quando todas as features sao potencialmente uteis
- Quando ha multicolinearidade (features correlacionadas) -- Ridge estabiliza
- Lembra a Parte A5? Ridge adiciona lambda na diagonal de X^T.X, tornando-a
  invertivel de novo. Resolve a singularidade.

### Lasso (L1) -- "zera os inuteis"

```
Minimiza: SSE + lambda . soma(|coef|)
```

A penalidade e a soma dos VALORES ABSOLUTOS dos coeficientes.

**O que faz:** alem de encolher, ZERA completamente os coeficientes das
features irrelevantes. Isso e SELECAO AUTOMATICA DE FEATURES.

**Quando usar:**
- Quando voce suspeita que muitas features sao inuteis e quer um modelo enxuto
- Quando quer interpretabilidade (menos variaveis = mais facil de explicar)
- Quando quer que o modelo "escolha" as features importantes sozinho

### Por que L1 zera e L2 nao? (a pergunta de ouro da sabatina)

Intuicao geometrica: a penalidade L1 (valor absoluto) tem "quinas" (formato
de losango). A solucao otima tende a cair nessas quinas, que ficam exatamente
nos eixos -- e estar no eixo significa coeficiente zero. A penalidade L2
(quadrado) e um circulo suave, sem quinas, entao a solucao raramente cai
exatamente num eixo. Por isso L1 zera e L2 so encolhe.

### ElasticNet -- "o melhor dos dois"

```
Minimiza: SSE + lambda . [ alpha.soma(|coef|) + (1-alpha).soma(coef^2) ]
```

Combina L1 e L2. O parametro `l1_ratio` (alpha) controla a mistura:
- l1_ratio = 1: puro Lasso
- l1_ratio = 0: puro Ridge
- l1_ratio = 0.5: meio a meio

**Quando usar:** quando voce tem muitas features correlacionadas E quer selecao.
Lasso sozinho, com features correlacionadas, escolhe uma e zera as outras
arbitrariamente. ElasticNet lida melhor com grupos de variaveis correlacionadas.

### O parametro lambda (forca da regularizacao)

- lambda = 0: sem regularizacao (volta ao OLS normal)
- lambda pequeno: regularizacao leve
- lambda grande: regularizacao forte (coeficientes vao a zero)
- lambda -> infinito: todos os coeficientes -> zero (modelo so com intercepto)

Escolhe lambda por validacao cruzada.

### IMPORTANTE: a convencao do sklearn (cai na prova)

No sklearn, Ridge e Lasso usam `alpha` como forca de regularizacao
(equivalente ao lambda). Mas em SVM e Regressao Logistica, usa-se `C`, que e
o INVERSO: `C = 1/lambda`.
- C grande -> regularizacao FRACA
- C pequeno -> regularizacao FORTE

Cuidado pra nao confundir na prova.

### Na sabatina (pergunta quase certa)

**Pergunta:** "Quando voce usaria Lasso em vez de Ridge?"

**Resposta modelo:** "Uso Lasso quando quero selecao automatica de features --
ou seja, quando suspeito que muitas variaveis sao irrelevantes e quero um
modelo enxuto e interpretavel. O Lasso zera os coeficientes das features
inuteis, deixando so as importantes. Ja o Ridge encolhe todos mas nao zera
nenhum, entao mantem todas as features. Uso Ridge quando acredito que a maioria
das variaveis contribui e quando ha multicolinearidade -- o Ridge estabiliza
os coeficientes nesse caso. Se tenho muitas features correlacionadas E quero
selecao, uso ElasticNet, que combina os dois e lida melhor com grupos de
variaveis correlacionadas do que o Lasso puro."

**Follow-up:** "Por que o Lasso consegue zerar coeficientes e o Ridge nao?"

**Resposta:** "E pela geometria da penalidade. A penalidade do Lasso (norma L1,
valor absoluto) tem formato de losango com quinas nos eixos. A solucao otima
tende a tocar essas quinas, e tocar uma quina no eixo significa que aquele
coeficiente e exatamente zero. Ja a penalidade do Ridge (norma L2, quadratica)
e um circulo liso, sem quinas, entao a solucao quase nunca cai exatamente sobre
um eixo -- os coeficientes ficam pequenos mas diferentes de zero."


````python
# DEMONSTRACAO C3 -- Ridge vs Lasso lado a lado
from sklearn.linear_model import Ridge, Lasso, LinearRegression
np.random.seed(42)
n = 200

# 10 features, mas SO 3 sao realmente uteis
X = np.random.randn(n, 10)
coef_real = np.array([5, -3, 2, 0, 0, 0, 0, 0, 0, 0])  # 7 sao inuteis (zero)
y = X @ coef_real + np.random.normal(0, 1, n)

# Tres modelos
ols = LinearRegression().fit(X, y)
ridge = Ridge(alpha=10).fit(X, y)
lasso = Lasso(alpha=0.5).fit(X, y)

print('=== COMPARANDO COEFICIENTES ===')
print('(features 4-10 deveriam ser ZERO -- sao inuteis)')
print()
print(f'{"Feature":<10}{"Real":>8}{"OLS":>10}{"Ridge":>10}{"Lasso":>10}')
print('-'*48)
for i in range(10):
    print(f'  X{i+1:<7}{coef_real[i]:>8}{ols.coef_[i]:>10.2f}{ridge.coef_[i]:>10.2f}{lasso.coef_[i]:>10.2f}')
print()

# Contar quantos o Lasso zerou
zerados = np.sum(np.abs(lasso.coef_) < 0.01)
print(f'Lasso zerou {zerados} coeficientes (selecao automatica de features)')
print(f'Ridge zerou {np.sum(np.abs(ridge.coef_) < 0.01)} coeficientes (so encolheu)')
print()

fig, ax = plt.subplots(figsize=(12, 5))
xpos = np.arange(10)
w = 0.25
ax.bar(xpos - w, ols.coef_, w, label='OLS', color='gray')
ax.bar(xpos, ridge.coef_, w, label='Ridge (L2)', color='steelblue')
ax.bar(xpos + w, lasso.coef_, w, label='Lasso (L1)', color='coral')
ax.axhline(0, color='black', lw=0.5)
ax.set_xticks(xpos); ax.set_xticklabels([f'X{i+1}' for i in range(10)])
ax.set_ylabel('Valor do coeficiente')
ax.set_title('Lasso ZERA features inuteis (X4-X10) | Ridge so ENCOLHE')
ax.legend()
plt.tight_layout(); plt.show()
print('Veja: Lasso (laranja) zerou X4-X10. Ridge (azul) deixou pequenos mas nao-zero.')
````

## C4 -- Regressao Logistica: classificacao de verdade

### O que e (e por que NAO e "regressao")

Apesar do nome, e um modelo de CLASSIFICACAO. Preve a PROBABILIDADE de algo
ser de uma classe (0 ou 1).

Exemplo banco: probabilidade de um cliente PJ dar default (inadimplir).

### Como funciona

1. Calcula uma combinacao linear: z = b0 + b1.X1 + b2.X2 + ...
2. Passa pela sigmoide: P(Y=1) = 1/(1 + e^(-z))
3. Se P > 0.5 (threshold), classifica como 1; senao 0

A sigmoide (lembra da Parte A2) garante que a saida fica entre 0 e 1.

### Por que nao usar regressao linear pra classificar?

Porque regressao linear pode prever valores fora de [0,1] (tipo -0.3 ou 1.8),
que nao fazem sentido como probabilidade. A logistica resolve isso com a
sigmoide.

### Interpretacao: Odds Ratio (cai na sabatina)

Os coeficientes da logistica se interpretam via ODDS (chances).

```
Odds = P(evento) / P(nao evento)
Odds Ratio = e^(coeficiente)
```

Se o coeficiente de uma variavel e 0.7, entao e^0.7 = 2.0. Significa: cada
unidade a mais nessa variavel DOBRA as chances (odds) do evento.

### A funcao de custo: Log Loss

A logistica nao minimiza erro ao quadrado. Minimiza a LOG LOSS (entropia
cruzada):

```
Log Loss = -media( y.log(p) + (1-y).log(1-p) )
```

**Por que log loss e nao erro quadratico?** Porque a log loss penaliza
SEVERAMENTE previsoes confiantes e erradas. Se voce preve 99% de chance de
nao-default e o cliente da default, a log loss explode. Isso forca o modelo a
ser bem calibrado nas probabilidades.

### Threshold e o trade-off

O padrao e 0.5, mas voce ajusta conforme o custo:
- Threshold baixo (ex: 0.3): pega mais positivos (recall alto), mas mais
  falsos positivos (precision baixa)
- Threshold alto (ex: 0.7): mais conservador

No banco, pra fraude ou default, voce ajusta conforme o custo de cada erro.

### Na sabatina

**Pergunta:** "Por que usar regressao logistica e nao linear pra prever default?"

**Resposta modelo:** "Porque default e um evento binario (acontece ou nao), e
eu quero uma PROBABILIDADE entre 0 e 1. A regressao linear poderia prever
valores negativos ou maiores que 1, que nao fazem sentido como probabilidade.
A logistica usa a funcao sigmoide pra garantir saida em [0,1]. Alem disso, no
banco a logistica e muito usada porque e interpretavel via odds ratio -- eu
consigo dizer 'essa variavel dobra as chances de default' -- o que e importante
pra explicar decisoes de credito ao regulador."


````python
# DEMONSTRACAO C4 -- regressao logistica e probabilidade
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, confusion_matrix
np.random.seed(42)
n = 1000

# Prever default de cliente PJ
score_credito = np.random.normal(600, 100, n)
endividamento = np.random.normal(0.5, 0.2, n)
# Quanto MENOR o score e MAIOR o endividamento, mais chance de default
z = -0.02*(score_credito - 600) + 3*endividamento - 1
prob_default = 1/(1+np.exp(-z))
default = np.random.binomial(1, prob_default)

X = np.column_stack([score_credito, endividamento])
modelo = LogisticRegression().fit(X, default)

print('=== REGRESSAO LOGISTICA: prever default PJ ===')
print()
print(f'  Taxa de default na base: {default.mean()*100:.1f}%')
print()
print('  Coeficientes e Odds Ratio:')
print(f'    Score de credito: coef={modelo.coef_[0][0]:+.4f}, OR={np.exp(modelo.coef_[0][0]):.3f}')
print(f'    Endividamento:    coef={modelo.coef_[0][1]:+.4f}, OR={np.exp(modelo.coef_[0][1]):.3f}')
print()
print('  INTERPRETACAO:')
print('    Score negativo -> mais score, MENOS chance de default (faz sentido)')
print('    Endividamento positivo -> mais divida, MAIS chance de default')
print()
auc = roc_auc_score(default, modelo.predict_proba(X)[:,1])
print(f'  AUC: {auc:.3f} (capacidade de separar bons de maus pagadores)')
````

## C5 -- Arvores, Random Forest, XGBoost: quando e por que

### Arvore de Decisao

Faz uma sequencia de perguntas SIM/NAO pra chegar numa decisao.

```
Faturamento > 100k?
  SIM -> Tempo de banco > 5 anos?
           SIM -> aprova
           NAO -> analisa
  NAO -> nega
```

**Vantagens:** super interpretavel (da pra desenhar), captura nao-linearidades
e interacoes, nao precisa normalizar.

**Desvantagem:** uma arvore sozinha faz OVERFITTING facil (decora o treino).

### Random Forest (floresta = muitas arvores)

Treina MUITAS arvores, cada uma:
- Num pedaco aleatorio dos dados (bootstrap)
- Vendo so um subconjunto aleatorio das features

A previsao final e a MEDIA (ou voto) de todas as arvores.

**Por que funciona?** Cada arvore erra de um jeito diferente. Na media, os
erros se cancelam. Isso reduz a VARIANCIA sem aumentar muito o vies.

**Quando usar:** quando quer boa performance sem muito tuning, dados tabulares,
e nao precisa de interpretabilidade total.

### XGBoost (Gradient Boosting)

Treina arvores em SEQUENCIA, onde cada nova arvore corrige os ERROS da anterior.

**Diferenca pro Random Forest:**
- Random Forest: arvores em PARALELO, independentes, depois faz media (bagging)
- XGBoost: arvores em SEQUENCIA, cada uma corrige a anterior (boosting)

**Quando usar:** quando quer a MELHOR performance possivel em dados tabulares.
XGBoost domina competicoes. Mas exige mais tuning e e menos interpretavel.

### A tabela de decisao

| Situacao | Modelo recomendado |
|----------|-------------------|
| Preciso explicar cada decisao | Arvore unica ou Regressao |
| Quero boa performance, pouco esforco | Random Forest |
| Quero performance maxima | XGBoost |
| Dados com relacao linear | Regressao Linear/Logistica |
| Poucos dados | Modelos lineares (arvores overfittam) |

### Por que arvores NAO precisam de normalizacao (cai na prova)

Arvores fazem splits comparando UMA feature de cada vez ("faturamento > 100k?").
A escala nao importa -- comparar 100k ou 0.1 (normalizado) da o mesmo split.
Ja modelos baseados em DISTANCIA (KNN, SVM) ou que somam features
(regressao) PRECISAM de normalizacao, senao a feature de maior escala domina.

### Na sabatina

**Pergunta:** "Qual a diferenca entre Random Forest e XGBoost, e quando usar
cada um?"

**Resposta modelo:** "Os dois sao ensembles de arvores, mas treinam de formas
opostas. Random Forest usa bagging: treina muitas arvores em paralelo, cada uma
num subconjunto aleatorio de dados e features, e faz a media. Isso reduz
variancia e e robusto, com pouco tuning. XGBoost usa boosting: treina arvores
em sequencia, cada uma focando nos erros da anterior, o que reduz o vies e
costuma dar performance superior. Eu uso Random Forest quando quero um modelo
solido rapido, com poucos hiperparametros pra ajustar. Uso XGBoost quando
performance e prioridade e tenho tempo pra tunar, aceitando menos
interpretabilidade. Pra dados tabulares de banco, XGBoost geralmente vence em
metricas, mas pesa a questao de explicabilidade."


````python
# DEMONSTRACAO C5 -- comparando os modelos
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.datasets import make_classification
np.random.seed(42)

# Dataset tabular com relacoes nao-lineares
X, y = make_classification(n_samples=2000, n_features=15, n_informative=8,
                           n_redundant=3, random_state=42)

modelos = {
    'Regressao Logistica': LogisticRegression(max_iter=1000),
    'Arvore unica': DecisionTreeClassifier(max_depth=5, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost (GradBoost)': GradientBoostingClassifier(n_estimators=100, random_state=42),
}

print('=== COMPARACAO DE MODELOS (AUC via 5-fold CV) ===')
print()
resultados = {}
for nome, modelo in modelos.items():
    scores = cross_val_score(modelo, X, y, cv=5, scoring='roc_auc')
    resultados[nome] = scores.mean()
    print(f'  {nome:<25} AUC = {scores.mean():.4f} (+/- {scores.std():.4f})')
print()

fig, ax = plt.subplots(figsize=(10, 5))
nomes = list(resultados.keys())
valores = list(resultados.values())
cores = ['gray', 'orange', 'steelblue', 'green']
ax.barh(nomes, valores, color=cores)
ax.set_xlim(0.5, 1.0)
ax.set_xlabel('AUC (maior = melhor)')
ax.set_title('Performance dos modelos em dados tabulares')
for i, v in enumerate(valores):
    ax.text(v+0.005, i, f'{v:.3f}', va='center')
plt.tight_layout(); plt.show()
print('Tipicamente: ensembles (RF, XGBoost) > arvore unica > linear,')
print('quando ha nao-linearidades. Mas linear pode vencer se a relacao for linear.')
````

## C6 -- Overfitting e o trade-off vies-variancia

### O conceito mais importante de ML

**Overfitting (sobreajuste):** o modelo DECORA o treino, incluindo o ruido, e
falha em dados novos. Como um aluno que decora as respostas da prova passada
mas nao entende a materia.

**Underfitting (subajuste):** o modelo e simples demais e nao captura nem os
padroes reais. Como um aluno que nem estudou.

### Vies e Variancia

- **Vies (bias):** erro por suposicoes simples demais. Modelo rigido. Underfitting.
- **Variancia:** erro por sensibilidade excessiva ao treino. Modelo instavel.
  Overfitting.

**O trade-off:** diminuir um geralmente aumenta o outro. O segredo e o EQUILIBRIO.

```
Modelo simples demais  -> alto vies, baixa variancia -> underfitting
Modelo complexo demais -> baixo vies, alta variancia -> overfitting
Modelo equilibrado     -> o ponto otimo
```

### Como detectar overfitting

Compare erro no TREINO vs erro no TESTE:
- Erro treino baixo + erro teste alto = OVERFITTING (decorou)
- Erro treino alto + erro teste alto = UNDERFITTING (nao aprendeu)
- Ambos baixos e proximos = bom modelo

### Como combater overfitting

1. Mais dados de treino
2. Regularizacao (Ridge, Lasso) -- penaliza complexidade
3. Reduzir complexidade (menos features, arvore mais rasa)
4. Early stopping (parar o treino antes de decorar)
5. Cross-validation pra detectar
6. Dropout (em redes neurais)

### Na sabatina

**Pergunta:** "Como voce sabe se seu modelo esta sofrendo overfitting?"

**Resposta modelo:** "Comparo a performance no conjunto de treino com a do
conjunto de teste ou validacao. Se o modelo vai muito bem no treino mas mal no
teste -- um gap grande -- e sinal claro de overfitting: ele decorou os padroes
e o ruido do treino em vez de aprender o que generaliza. Pra confirmar, uso
validacao cruzada. Pra combater, posso adicionar regularizacao, reduzir a
complexidade do modelo, conseguir mais dados, ou usar early stopping. O objetivo
e achar o ponto de equilibrio no trade-off vies-variancia, onde o erro de teste
e minimo."


````python
# DEMONSTRACAO C6 -- overfitting visual
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error
np.random.seed(42)

# Dados com relacao simples + ruido
x = np.sort(np.random.uniform(0, 1, 30))
y = np.sin(2*np.pi*x) + np.random.normal(0, 0.2, 30)
x_plot = np.linspace(0, 1, 200)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
graus = [1, 4, 15]
titulos = ['UNDERFITTING (grau 1)\nSimples demais', 
           'BOM (grau 4)\nEquilibrado',
           'OVERFITTING (grau 15)\nDecorou o ruido']

for ax, grau, titulo in zip(axes, graus, titulos):
    modelo = make_pipeline(PolynomialFeatures(grau), LinearRegression())
    modelo.fit(x.reshape(-1,1), y)
    y_pred_plot = modelo.predict(x_plot.reshape(-1,1))
    erro_treino = mean_squared_error(y, modelo.predict(x.reshape(-1,1)))
    
    ax.scatter(x, y, alpha=0.6, color='steelblue', label='dados')
    ax.plot(x_plot, np.sin(2*np.pi*x_plot), 'g--', alpha=0.5, label='verdade')
    ax.plot(x_plot, y_pred_plot, 'r-', lw=2, label='modelo')
    ax.set_ylim(-2, 2)
    ax.set_title(f'{titulo}\nErro treino: {erro_treino:.3f}')
    ax.legend(fontsize=8)
plt.tight_layout(); plt.show()

print('Grau 1: nao captura a curva (alto vies, underfitting)')
print('Grau 4: captura bem o padrao real (equilibrado)')
print('Grau 15: passa por todos os pontos, mas oscila louco (alta variancia, overfitting)')
print()
print('O grau 15 tem erro de TREINO baixissimo, mas erraria feio em dados novos.')
````

## C7 -- Validacao Cruzada: por que e como

### O problema

Voce treina um modelo e testa nos mesmos dados. Da 99% de acerto. Mas em
producao, da 70%. O que aconteceu? Voce testou no que o modelo ja viu -- nao
mediu generalizacao.

### A solucao basica: train/test split

Separa os dados: 80% pra treinar, 20% pra testar. O modelo nunca ve os 20%
durante o treino. Testar neles mede generalizacao de verdade.

### O problema do split simples

Depende de QUAIS 20% voce escolheu. Se por azar os 20% forem faceis, voce
superestima. Se forem dificeis, subestima.

### K-Fold Cross-Validation

Divide os dados em K partes (folds). Treina K vezes:
- Vez 1: treina nos folds 2-5, testa no fold 1
- Vez 2: treina nos folds 1,3,4,5, testa no fold 2
- ... e assim por diante

No fim, cada dado foi testado uma vez. A performance e a MEDIA das K rodadas.
Mais robusto que um split unico.

### Stratified K-Fold (pra classes desbalanceadas)

Garante que cada fold tem a MESMA proporcao de classes. Essencial quando uma
classe e rara (fraude, default). Sem isso, um fold poderia ficar sem nenhum
caso positivo.

### Cuidado especial: dados temporais

Se os dados tem ordem temporal (transacoes ao longo do tempo), NAO pode
embaralhar. Senao voce treina no futuro e testa no passado -- vazamento.
Use TimeSeriesSplit: treina no passado, testa no futuro.

### Cuidado: mesma entidade em varios registros

Se o mesmo cliente PJ aparece em varias linhas, ele nao pode estar no treino
E no teste -- senao ha vazamento. Use GroupKFold agrupando por cliente.

### Na sabatina

**Pergunta:** "Por que usar validacao cruzada em vez de um simples train/test
split?"

**Resposta modelo:** "Um split unico depende da sorte de quais dados cairam no
teste -- pode superestimar ou subestimar a performance dependendo da particao.
A validacao cruzada K-fold testa o modelo em K particoes diferentes e tira a
media, dando uma estimativa mais estavel e confiavel da generalizacao. Tambem
aproveita melhor os dados, ja que cada observacao e usada tanto pra treino
quanto pra teste em rodadas diferentes. Em dados desbalanceados, uso Stratified
K-Fold pra manter a proporcao de classes; em dados temporais, uso
TimeSeriesSplit pra nao vazar informacao do futuro."


````python
# DEMONSTRACAO C7 -- por que CV e mais confiavel
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification
np.random.seed(42)

X, y = make_classification(n_samples=500, n_features=10, random_state=42)

# Varios splits simples diferentes (mostra instabilidade)
print('=== SPLITS SIMPLES (cada um da um numero diferente) ===')
scores_simples = []
for seed in range(5):
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, random_state=seed)
    m = RandomForestClassifier(n_estimators=50, random_state=42).fit(Xtr, ytr)
    score = m.score(Xte, yte)
    scores_simples.append(score)
    print(f'  Split seed={seed}: acuracia = {score:.4f}')
print(f'  -> Varia de {min(scores_simples):.3f} a {max(scores_simples):.3f} (instavel!)')
print()

# Cross-validation (estavel)
cv_scores = cross_val_score(RandomForestClassifier(n_estimators=50, random_state=42), X, y, cv=5)
print('=== VALIDACAO CRUZADA 5-FOLD ===')
for i, s in enumerate(cv_scores):
    print(f'  Fold {i+1}: {s:.4f}')
print(f'  -> Media: {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})')
print()
print('A media da CV e a estimativa mais confiavel da performance real.')
````

## C8 -- Metricas: qual usar e quando cada uma MENTE

### Por que a metrica certa importa

Usar a metrica errada leva a decisoes erradas. O exemplo classico: ACURACIA
em dados desbalanceados.

### Metricas de Classificacao

**Acuracia:** % de acertos totais.
```
Acuracia = (acertos) / (total)
```
**QUANDO MENTE:** dados desbalanceados. Se 99% e nao-fraude, prever "tudo
nao-fraude" da 99% de acuracia mas e inutil (nunca pega fraude).

**Precision (precisao):** dos que previ como positivos, quantos eram mesmo?
```
Precision = TP / (TP + FP)
```
Use quando o custo de FALSO POSITIVO e alto (ex: bloquear cliente bom).

**Recall (revocacao/sensibilidade):** dos positivos reais, quantos peguei?
```
Recall = TP / (TP + FN)
```
Use quando o custo de FALSO NEGATIVO e alto (ex: deixar passar uma fraude).

**O trade-off precision-recall:** geralmente um sobe quando o outro desce.
Ajusta o threshold conforme o custo de cada erro.

**F1-Score:** media harmonica de precision e recall.
```
F1 = 2 . (P . R) / (P + R)
```
Use quando quer equilibrar os dois. A media harmonica penaliza desequilibrio.

**AUC-ROC:** capacidade do modelo de SEPARAR as classes, em todos os thresholds.
- 0.5 = aleatorio (inutil)
- 1.0 = separacao perfeita
Robusto a desbalanceamento. Muito usado em credito.

**KS (Kolmogorov-Smirnov):** separacao maxima entre as distribuicoes de bons e
maus. Padrao em credit scoring no banco. Relacao: Gini = 2.AUC - 1.

### Metricas de Regressao

**RMSE (raiz do erro quadratico medio):** erro tipico, na unidade do alvo.
Penaliza erros grandes (por causa do quadrado). Sensivel a outliers.

**MAE (erro absoluto medio):** erro medio absoluto. Mais robusto a outliers
que RMSE.

**MAPE (erro percentual):** erro em %. CUIDADO: indefinido se o alvo pode ser
zero (divide por zero).

**R-quadrado:** % da variancia do alvo explicada pelo modelo. De 0 a 1
(pode ser negativo em modelos pessimos).

### A tabela de "quando cada uma mente"

| Metrica | Mente quando... |
|---------|-----------------|
| Acuracia | classes desbalanceadas |
| RMSE | ha outliers (penaliza demais) |
| MAPE | alvo pode ser zero ou perto |
| R^2 | sempre sobe ao add variaveis (use R^2 ajustado) |
| Precision sozinha | ignora os positivos perdidos |
| Recall sozinho | ignora os falsos alarmes |

### Na sabatina

**Pergunta:** "Voce tem um modelo de fraude com 99% de acuracia. Esta bom?"

**Resposta modelo:** "Provavelmente nao, e a acuracia esta enganando. Fraude e
um evento raro -- digamos 1% das transacoes. Um modelo que simplesmente preve
'nunca e fraude' acerta 99% das vezes, mas e completamente inutil porque nunca
detecta uma fraude real. Por isso, em problemas desbalanceados, eu nao olho
acuracia. Olho precision e recall (ou F1), AUC-ROC, e principalmente o custo de
negocio de cada tipo de erro. Pra fraude, o recall importa muito -- nao quero
deixar fraude passar -- mas equilibrado com precision pra nao bloquear clientes
bons demais. E ajusto o threshold conforme esse trade-off de custo."


````python
# DEMONSTRACAO C8 -- por que acuracia mente em dados desbalanceados
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
np.random.seed(42)
n = 10000

# Base com 1% de fraude (desbalanceada)
X = np.random.randn(n, 5)
y = np.zeros(n)
fraudes_idx = np.random.choice(n, size=int(n*0.01), replace=False)
y[fraudes_idx] = 1
X[fraudes_idx] += 1.5  # fraudes tem padrao um pouco diferente

Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Modelo 1: idiota -- preve sempre 0 (nao-fraude)
pred_idiota = np.zeros(len(yte))

# Modelo 2: regressao logistica de verdade
modelo = LogisticRegression().fit(Xtr, ytr)
pred_real = modelo.predict(Xte)

print('=== POR QUE ACURACIA MENTE ===')
print(f'Base: {yte.mean()*100:.1f}% de fraude (desbalanceada)')
print()
print('MODELO IDIOTA (preve sempre nao-fraude):')
print(f'  Acuracia:  {accuracy_score(yte, pred_idiota)*100:.1f}%  <- parece otimo!')
print(f'  Recall:    {recall_score(yte, pred_idiota, zero_division=0)*100:.1f}%  <- PEGA ZERO FRAUDES!')
print(f'  Inutil: nunca detecta uma fraude, mas tem 99% de acuracia.')
print()
print('MODELO REAL (regressao logistica):')
print(f'  Acuracia:  {accuracy_score(yte, pred_real)*100:.1f}%')
print(f'  Precision: {precision_score(yte, pred_real, zero_division=0)*100:.1f}%')
print(f'  Recall:    {recall_score(yte, pred_real, zero_division=0)*100:.1f}%')
print(f'  F1:        {f1_score(yte, pred_real, zero_division=0)*100:.1f}%')
print(f'  AUC:       {roc_auc_score(yte, modelo.predict_proba(Xte)[:,1]):.3f}')
print()
print('A acuracia dos dois e parecida, mas o modelo idiota tem recall ZERO.')
print('Por isso, em fraude, olhamos recall/precision/AUC -- nunca so acuracia.')
````

# PARTE D -- Causal AI (o diferencial da vaga)

Essa parte e resumida porque voce ja tem o guia completo de Causal AI. Aqui
fica o essencial conectado com o resto.


## D1 -- Por que correlacao nao basta (revisao aplicada)

ML preditivo responde "o que vai acontecer". Causal AI responde "o que acontece
SE EU INTERVIR".

- Pergunta preditiva: "clientes que usam o app tem qual NPS?" (descreve)
- Pergunta causal: "SE eu fizer o cliente usar o app, o NPS sobe?" (intervem)

A vaga e de Causal AI porque o banco quer TOMAR ACOES (intervir), nao so prever.

## D2 -- Potential Outcomes e as medidas

Pra cada cliente existem dois "mundos":
- Y(1): NPS se ele recebesse a acao
- Y(0): NPS se nao recebesse

Efeito individual = Y(1) - Y(0). Mas so observamos UM mundo. Esse e o problema
fundamental da inferencia causal.

Medidas:
- **ATE:** efeito medio na populacao toda
- **ATT:** efeito medio nos que receberam a acao
- **CATE:** efeito por subgrupo (base pra personalizacao)

## D3 -- As 3 suposicoes (decora pra sabatina)

1. **SUTVA:** o tratamento de um nao afeta o resultado de outro
2. **Ignorability:** controlados os confundidores observaveis, o tratamento e
   "como se aleatorio" (nao ha confundidor nao medido)
3. **Positivity:** todo mundo tem chance > 0 de receber ambos os tratamentos

## D4 -- Os metodos (quando usar cada)

| Situacao | Metodo |
|----------|--------|
| Posso randomizar | A/B test (padrao ouro) |
| Tenho confundidores medidos | Propensity Score Matching, IPW, Regressao |
| Tenho antes/depois + grupo controle | Difference-in-Differences |
| Ha um cutoff/corte de decisao | Regression Discontinuity |
| Suspeito de confundidor nao medido | Instrumental Variables |
| Quero combinar ML + causal | Double ML, Causal Forests |

## D5 -- O exemplo do banco que voce DEVE saber contar

**Cenario:** o time de produto quer saber se a nova feature de cartao virtual
aumenta o engajamento.

**Abordagem que voce descreveria na sabatina:**
1. Desenharia o DAG: quais variaveis afetam tanto "ativar cartao" quanto
   "engajamento" (confundidores: idade, perfil digital, tempo de banco)
2. Se houve A/B test (alguns receberam aleatoriamente): comparo direto
3. Se nao: uso PSM ou DiD, controlando os confundidores
4. Verifico as suposicoes (parallel trends pro DiD, overlap pro PSM)
5. Reporto o efeito com intervalo de confianca
6. Analiso heterogeneidade (a feature ajuda MAIS algum segmento?)
7. Recomendo acao: rollout pra quem se beneficia, com base no CATE

Isso mostra que voce nao so sabe a teoria, mas sabe APLICAR no contexto real

Voce nao precisa ser o melhor matematico da sala. Precisa ENTENDER o suficiente
pra justificar suas decisoes com clareza. Esse guia te deu o "por que" de cada
ferramenta. Agora e praticar ate virar natural.

Boa sorte na prova e na sabatina. Voce esta muito mais preparado do que pensa.
