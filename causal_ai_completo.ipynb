---

## A pergunta que muda tudo

Imagina que voce trabalha no Itau e o time de produto te procura:

"Vimos que clientes que usam o Pix tem NPS 1.5 pontos maior que os que nao usam.
Devemos forcar todos a usar Pix para subir o NPS?"

Voce olha pros dados. A correlacao e real -- r = 0.6.
Estatisticamente significativa.

**Mas se voce fizer o que o time de produto pede, voce vai falhar.**

Por que? Porque clientes que ESCOLHEM usar Pix sao FUNDAMENTALMENTE diferentes
dos que nao usam. Eles tem mais idade digital. Tem mais educacao financeira.
Tem mais relacionamento com o banco. Essas caracteristicas TAMBEM elevam o NPS.

Forcar Pix em quem nao quer usar nao vai mudar essas caracteristicas.
Voce vai ter Pix sendo usado, mas NPS no mesmo lugar (ou pior, porque criou friccao).

---

## A diferenca conceitual entre 3 perguntas

| Pergunta | Tipo | Ferramenta |
|----------|------|------------|
| "Qual NPS esperar de cliente que usa Pix?" | Preditiva | ML classico |
| "Por que cliente X tem NPS Y?" | Explicativa | SHAP, interpretabilidade |
| "Se eu fizer cliente X usar Pix, qual sera o novo NPS?" | **Causal** | **Inferencia Causal** |

**As 3 sao validas. Mas elas respondem coisas diferentes.**

O time de produto quer a terceira -- intervencao.
Mas analise descritiva so responde a primeira.

Causal AI eh o que separa quem analisa do que quem DECIDE.

---

## A historia da medicina como analogia

Antes de 1948, medicina era observacional. Medico observava: "pacientes que tomam
X melhoram." Recomendava X.

1948 -- primeiro grande RCT (Randomized Controlled Trial): tuberculose com
estreptomicina. Pela primeira vez, randomizacao: metade tomou, metade nao.
Diferenca observada eh o efeito causal real, sem confundidores.

A medicina cresceu 100x depois disso. Tudo que voce sabe sobre medicina moderna
foi estabelecido por RCTs.

**Causal AI eh o equivalente para decisoes de negocio.**

Quando voce NAO PODE rodar A/B test (caro, etico, ja aconteceu),
Causal AI te da metodos para extrair causalidade de dados observacionais.

---

## Definicao formal: Potential Outcomes Framework

Donald Rubin (Harvard, 1974) -- considerado o pai da inferencia causal moderna.

Para cada unidade i (cliente, servidor, transacao), existem DOIS resultados
potenciais:

```
Y_i(1) = valor de Y se a unidade i FOSSE tratada
Y_i(0) = valor de Y se a unidade i NAO fosse tratada
```

**Efeito causal individual:**
```
tau_i = Y_i(1) - Y_i(0)
```

**Mas voce so observa um:**
```
Y_i^obs = T_i * Y_i(1) + (1 - T_i) * Y_i(0)
```

Onde T_i e 0 ou 1.

---

## O Problema Fundamental (Holland, 1986)

> "Voce nunca pode observar o efeito causal individual."

E como pedir para uma pessoa em dois cenarios paralelos do universo.
Voce escolhe um. O outro fica para sempre nao observado.

**Solucao:** estimar efeitos MEDIOS.

```
ATE = E[Y_i(1) - Y_i(0)] = E[Y(1)] - E[Y(0)]
```

Sob a suposicao de que o tratamento foi atribuido aleatoriamente (A/B test),
voce pode estimar isso com diferenca de medias:

```
hat{ATE} = mean(Y | T=1) - mean(Y | T=0)
```

**Sem randomizacao, essa estimativa eh ENVIESADA.** Resto do curso eh sobre
como corrigir esse vies.

---

## As 4 medidas que voce DEVE saber explicar

### ATE -- Average Treatment Effect
```
ATE = E[Y(1) - Y(0)]
```
Efeito medio se tratassemos toda a populacao vs ninguem.
Politica publica usa essa.

### ATT -- Average Treatment effect on Treated
```
ATT = E[Y(1) - Y(0) | T = 1]
```
Efeito medio nos que JA receberam tratamento.
Avaliacao de campanha ja realizada usa essa.

### ATC -- Average Treatment effect on Controls
```
ATC = E[Y(1) - Y(0) | T = 0]
```
Efeito hipotetico se controles tivessem sido tratados.
Util para decidir expandir campanha.

### CATE -- Conditional ATE (Heterogeneous Treatment Effect)
```
CATE(x) = E[Y(1) - Y(0) | X = x]
```
Efeito condicional ao perfil X. Base para personalizacao.

**Relacao matematica:**
```
ATE = ATT * P(T=1) + ATC * P(T=0)
```

Se ATT != ATC, ha **heterogeneidade no efeito**. ML causal moderno (Causal Forests)
foca exatamente nisso.


````python
# DEMONSTRACAO 1.1 -- Construindo o universo completo (com counterfactuals)
# Vamos simular um cenario IDEAL onde temos acesso aos potential outcomes
# Na vida real isso eh impossivel. Mas serve para entender as quantidades.

np.random.seed(42)
N = 10000

# Caracteristicas dos clientes (covariaveis X)
idade = np.random.normal(40, 12, N).clip(18, 80)
renda = np.random.lognormal(8, 0.6, N)
tempo_relacionamento = np.random.exponential(5, N).clip(0.1, 30)
score_digital = np.random.beta(3, 2, N) * 100   # 0-100, fluencia digital

# Y(0): NPS basal sem intervencao (cliente nao recebe campanha de educacao financeira)
Y0 = (
    5
    + 0.03 * (idade - 40)
    + 0.000002 * renda
    + 0.2 * tempo_relacionamento
    + 0.04 * score_digital
    + np.random.normal(0, 1.5, N)
)

# Y(1): NPS com campanha
# Efeito HETEROGENEO: clientes com baixo score_digital se beneficiam mais
efeito_individual = 1.5 - 0.015 * score_digital   # quem ja eh digital, ganha pouco
Y1 = Y0 + efeito_individual

df_universo = pd.DataFrame({
    'id': range(N),
    'idade': idade,
    'renda': renda,
    'tempo_relacionamento': tempo_relacionamento,
    'score_digital': score_digital,
    'Y0': Y0,
    'Y1': Y1,
    'tau': efeito_individual   # ITE -- nao observavel na vida real
})

print('=== UNIVERSO COMPLETO (simulado) ===')
print(f'N = {N:,}')
print()
print(df_universo.head())
print()

ATE_real = df_universo['tau'].mean()
print(f'ATE verdadeiro (E[Y(1)-Y(0)]):  {ATE_real:.4f}')
print(f'  Interpretacao: a campanha eleva NPS em ~{ATE_real:.2f} pontos em media')
print()

# Distribuicao do efeito individual (CATE)
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].hist(df_universo['tau'], bins=50, color='steelblue', edgecolor='black')
axes[0].axvline(ATE_real, color='red', linestyle='--', lw=2, label=f'ATE = {ATE_real:.3f}')
axes[0].set_xlabel('Efeito individual (tau_i)')
axes[0].set_title('Distribuicao do EFEITO INDIVIDUAL\n(heterogeneidade real do tratamento)')
axes[0].legend()

# Relacao entre score digital e efeito
axes[1].scatter(df_universo['score_digital'], df_universo['tau'],
                alpha=0.3, s=5, color='coral')
axes[1].set_xlabel('Score Digital'); axes[1].set_ylabel('Efeito Individual')
axes[1].set_title('CATE: Efeito decresce com fluencia digital\nQuem ja eh digital, ganha pouco com a campanha')
plt.tight_layout(); plt.show()

print()
print('OBSERVACAO CRITICA:')
print(f'  Efeito MAXIMO (score digital = 0):  {df_universo[df_universo.score_digital < 5]["tau"].mean():.3f}')
print(f'  Efeito MEDIO (score digital ~50):    {df_universo[(df_universo.score_digital > 45) & (df_universo.score_digital < 55)]["tau"].mean():.3f}')
print(f'  Efeito MINIMO (score digital > 95):  {df_universo[df_universo.score_digital > 95]["tau"].mean():.3f}')
print()
print('Decisao otima: PERSONALIZAR -- mandar campanha para quem score < 30.')
print('Decisao errada: mandar para todos (gasta budget em quem nao se beneficia).')
print('Para tomar essa decisao, voce precisa do CATE -- nao basta ATE.')
````

````python
# DEMONSTRACAO 1.2 -- O vies sem randomizacao
# Cenario realista: cliente recebe a campanha NAO aleatoriamente
# Probabilidade de receber depende das caracteristicas (selecao endogena)

np.random.seed(42)

# Mecanismo de selecao: gestores escolhem clientes MAIS DIGITAIS para a campanha
# (achando que assim teriam mais retorno -- mas e o oposto!)
logit_T = -2.0 + 0.05 * (df_universo['score_digital'] - 50) + 0.05 * (df_universo['idade'] - 40)
prob_T = 1 / (1 + np.exp(-logit_T))
df_universo['T'] = np.random.binomial(1, prob_T)

# Y observado (so o que aconteceu de fato)
df_universo['Y_obs'] = np.where(df_universo['T']==1, df_universo['Y1'], df_universo['Y0'])

# 3 estimativas
ATE_real = df_universo['tau'].mean()
ATT_real = df_universo[df_universo.T==1]['tau'].mean()
ATC_real = df_universo[df_universo.T==0]['tau'].mean()

# Estimativa NAIVE: diferenca de medias
Y_T1 = df_universo[df_universo.T==1]['Y_obs'].mean()
Y_T0 = df_universo[df_universo.T==0]['Y_obs'].mean()
diff_naive = Y_T1 - Y_T0

print('=== AS 4 MEDIDAS CAUSAIS ===')
print()
print(f'  ATE (verdadeiro):       {ATE_real:+.4f}')
print(f'  ATT (verdadeiro):       {ATT_real:+.4f}  (tratados ja eram digitais, ganham pouco)')
print(f'  ATC (verdadeiro):       {ATC_real:+.4f}  (controles NAO digitais, ganhariam mais)')
print(f'  Diff naive (observada): {diff_naive:+.4f}  ENVIESADA!')
print()
print('OBSERVACOES:')
print('  1. ATT < ATC: quem foi escolhido se beneficia MENOS que quem nao foi.')
print('     Isso eh COMUM em selecao endogena.')
print('  2. Diff naive eh enviesada porque os grupos sao DIFERENTES em X.')
print('     Tratados sao mais velhos e mais digitais -> ja teriam NPS maior.')
print()
print('PERGUNTA DE NEGOCIO REVELADA:')
print('  Voce gostaria de TROCAR de estrategia: focar em quem score < 30')
print('  (ATC alto) ao inves de continuar mandando para os ja digitais.')
print('  Mas analise descritiva nunca te diria isso.')
````

# MODULO 2 -- As 3 Suposicoes (e Por que Sao Sagradas)

---

## O grande insight

Inferencia causal observacional eh um pacto:
voce **assume** algumas coisas e, em troca, consegue extrair causalidade.

Sem assumir nada, voce so consegue correlacao.

As 3 suposicoes que vamos cobrir sao a base de TODO metodo causal classico.
Quando voce le um paper de causalidade, sempre tem uma secao 'Assumptions'.
Sao essas.

**Pegadinha de entrevista classica:** o entrevistador pergunta 'quais as
suposicoes que voce esta assumindo?'. Quem nao sabe responder, eh eliminado.

---

## Suposicao 1: SUTVA (Stable Unit Treatment Value Assumption)

Tem dois componentes que sao quase sempre confundidos:

### Componente A: No Interference
> 'O tratamento de uma unidade nao afeta o resultado de outra.'

**Violacao classica em banco:**
Voce roda campanha A para clientes selecionados. Mas os clientes pertencem
a grupos de WhatsApp e contam uns aos outros. Os nao tratados ficam sabendo
e mudam comportamento. O Y deles foi contaminado.

**Violacao classica em tecnologia:**
Voce ativa nova feature em alguns servidores. Mas servidores compartilham
banco de dados. A mudanca em um afeta o desempenho de outros.

### Componente B: No Hidden Variations
> 'Existe apenas UMA versao do tratamento.'

**Violacao classica:**
Voce trata como 'recebeu email da campanha'. Mas alguns receberam email
texto simples, outros receberam email com video, outros foram para spam.
Sao tratamentos DIFERENTES mascarados como um.

### Como verificar/mitigar

- Defina o tratamento EXATAMENTE. 'Recebeu email VERSAO X em PRIMEIRA tentativa.'
- Para interferencia: agrupe unidades que se influenciam (cluster randomization).
- Em telemetria: agrupe servidores por sub-rede ao analisar.

---

## Suposicao 2: Ignorability (Conditional Exchangeability)

**A suposicao mais importante. Tambem mais dificil de garantir.**

**Forma matematica:**
```
Y(0), Y(1) independente de T | X
```

Em portugues: depois que voce CONTROLA pelas covariaveis X medidas,
T eh efetivamente aleatorio. Como se voce tivesse feito um A/B test
dentro de cada subgrupo de X.

**Sinonimos que aparecem em papers:**
- No Unmeasured Confounders (NUC)
- Conditional Independence Assumption (CIA)
- Selection on Observables
- Strong Ignorability (quando vale com positivity)

### O que significa na pratica

Voce precisa ter MEDIDO todos os confundidores. Se existe uma variavel U
que afeta T e Y mas nao esta no dataset, ignorability viola.

**Exemplo banco:**
Voce quer saber efeito causal de 'usar Pix' em NPS.
Mediu: idade, renda, tempo de relacionamento, score de credito.
**NAO mediu:** quanto o cliente confia em tecnologia.

Confianca em tecnologia afeta uso de Pix (causal) E NPS (clientes que
confiam mais tem expectativas diferentes). Voce tem confundidor nao
medido. Ignorability viola. Suas estimativas sao enviesadas.

### Como verificar

**Voce NAO consegue verificar diretamente.** Eh impossivel testar a partir
de dados se tem confundidor nao medido. Voce so pode:

1. **Listar exaustivamente** todos os mecanismos pelos quais T e Y se relacionam
2. **Verificar** se mediu cada variavel desses mecanismos
3. **Argumentar com conhecimento de dominio**

### Sensitivity analysis (importante para entrevista!)

Pergunta: 'Quanto de confundidor nao medido seria necessario para mudar
minha conclusao?'

Se a resposta eh 'muito' (efeito muito grande), conclusao eh robusta.
Se a resposta eh 'pouco', conclusao eh fragil.

Metodos: E-value (VanderWeele), Rosenbaum bounds, Cinelli & Hazlett (2020).

---

## Suposicao 3: Positivity (Overlap)

**Forma matematica:**
```
0 < P(T=1 | X = x) < 1, para todo x no suporte de X
```

Em portugues: para cada valor das covariaveis X, deve haver chance positiva
(e nao certeza) de receber tratamento.

### Violacoes comuns

**Violacao por construcao:**
- Politica do banco: clientes < 18 anos NUNCA recebem cartao de credito.
- Estritamente: P(T=1 | idade < 18) = 0. Voce nao consegue inferir efeito do cartao em menores.

**Violacao empirica:**
- Em sua amostra, nao tem clientes acima de 80 anos no grupo controle.
- Para essa regiao de idade, voce nao tem dados para comparar.

### Como verificar

1. Estimar propensity score: `e(x) = P(T=1 | X=x)` via logistic regression
2. Plot da distribuicao de e(x) por grupo (T=0 e T=1)
3. Identificar regioes onde so um grupo aparece -> overlap viola

### Como resolver

**Opcao 1: Trim** -- remover unidades fora da regiao de overlap comum.
Voce perde generalidade (nao pode falar sobre essa regiao) mas ganha validez.

**Opcao 2: Definir estimando alvo diferente** -- em vez de ATE (toda populacao),
estimar ATT (so nos tratados) ou ATO (overlap-weighted).

---

## Quando todas as 3 valem: identificacao causal

Sob SUTVA + Ignorability + Positivity:
```
E[Y(1) - Y(0) | X = x] = E[Y | T=1, X=x] - E[Y | T=0, X=x]
```

**O ladoesquerdo eh causal (counterfactual).**
**O lado direito eh observavel.**

Essa eh a magica: SOB as suposicoes, voce pode estimar causalidade
a partir de dados observacionais.

Os metodos do resto do curso (regressao, PSM, IPW, etc.) sao MANEIRAS
DIFERENTES de fazer essa estimativa. Mas todas dependem dessas suposicoes.

---

## Tabela de violacoes comuns no Itau

| Cenario | Suposicao violada | Por que |
|---------|---------------------|--------|
| Campanha enviada por gestor sem regra clara | Ignorability | Gestor escolhe baseado em criterio nao registrado |
| Feature ativada em grupo de servidores | SUTVA (interferencia) | Servidores compartilham infraestrutura |
| Cliente PJ com receita > 10M nunca recebe certa oferta | Positivity | Probabilidade = 0 nesse grupo |
| Cliente VIP sempre recebe ligacao | Positivity | Probabilidade = 1 |
| 'Recebeu ligacao' inclui ligacao de venda e ligacao de retencao | SUTVA (hidden variations) | Dois tratamentos diferentes |


````python
# DEMONSTRACAO 2.1 -- VIOLACAO de Ignorability (Confundidor nao medido)
np.random.seed(42)
N = 5000

# Confundidor LATENTE: 'confianca em tecnologia' (nao medido)
confianca_tech = np.random.beta(2, 2, N)  # 0 a 1, nao observavel

# Caracteristicas observaveis (medidas)
idade = np.random.normal(40, 12, N).clip(18, 80)
renda = np.random.lognormal(8, 0.5, N)

# Confianca tech AFETA uso de Pix (T) e NPS (Y) -- confundidor
prob_pix = 1/(1+np.exp(-(
    -2.0 + 0.01*(idade-40) + 0.000001*renda + 3.0*confianca_tech
)))
T = np.random.binomial(1, prob_pix, N)

# Y verdadeiro: depende de T (efeito real = +0.5) e de confianca_tech
Y_observado = (
    5
    + 0.5 * T                # efeito causal real do Pix
    + 0.02*(idade-40)
    + 2.0*confianca_tech     # confundidor nao medido afeta Y
    + np.random.normal(0, 1, N)
)

# CASO 1: voce nao mede confianca_tech (situacao real)
df_obs = pd.DataFrame({
    'idade': idade, 'renda': renda, 'T': T, 'Y': Y_observado
})

# Tentativa de estimar efeito sem o confundidor
import statsmodels.formula.api as smf
mod_sem_conf = smf.ols('Y ~ T + idade + renda', data=df_obs).fit()
efeito_sem = mod_sem_conf.params['T']

# CASO 2: voce TEM o confundidor (cenario ideal)
df_completo = df_obs.copy()
df_completo['confianca_tech'] = confianca_tech
mod_com_conf = smf.ols('Y ~ T + idade + renda + confianca_tech', data=df_completo).fit()
efeito_com = mod_com_conf.params['T']

print('=== VIOLACAO DE IGNORABILITY -- CONFUNDIDOR NAO MEDIDO ===')
print()
print(f'  Efeito causal REAL do Pix:                +0.500')
print(f'  Estimativa SEM medir confianca_tech:     {efeito_sem:+.3f}  (enviesado)')
print(f'  Estimativa COM medir confianca_tech:     {efeito_com:+.3f}  (recupera o real)')
print()
print('LICAO:')
print('  Mesmo com regressao multipla controlando idade e renda, voce')
print('  superestima o efeito porque ha um CONFUNDIDOR NAO MEDIDO.')
print()
print('Eh por isso que entrevistador sempre pergunta:')
print('  "Que confundidores nao medidos podem existir?"')
print('  "Como voce sabe que mediu TODOS os confundidores?"')
print()
print('Resposta forte: "Nao sei com certeza. Por isso fiz sensitivity analysis')
print(' usando E-value. O efeito sobreviveu a ate U=1.5x mais forte que covariaveis."')
````

````python
# DEMONSTRACAO 2.2 -- VIOLACAO de Positivity (Overlap)
np.random.seed(42)
N = 3000

df_pos = pd.DataFrame({
    'idade': np.random.uniform(18, 80, N),
    'renda_anual': np.random.uniform(20000, 500000, N)
})

# CENARIO 1 -- OVERLAP BOM
# Probabilidade varia mas em todos os quadrantes
df_pos['e_bom'] = 0.4 + 0.3 * np.sin(df_pos['idade'] / 20) + np.random.normal(0, 0.05, N)
df_pos['e_bom'] = df_pos['e_bom'].clip(0.15, 0.85)
df_pos['T_bom'] = np.random.binomial(1, df_pos['e_bom'])

# CENARIO 2 -- OVERLAP RUIM
# Idosos com alta renda quase sempre tratados; jovens com baixa renda quase nunca
df_pos['e_ruim'] = 1 / (1 + np.exp(-(
    -3 + 0.05*(df_pos['idade']-50) + 0.00002*(df_pos['renda_anual']-200000)
)))
df_pos['T_ruim'] = np.random.binomial(1, df_pos['e_ruim'])

fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# Cenario bom -- propensity score por grupo
axes[0,0].hist(df_pos[df_pos.T_bom==0]['e_bom'], bins=30, alpha=0.6, color='steelblue', label='Controle')
axes[0,0].hist(df_pos[df_pos.T_bom==1]['e_bom'], bins=30, alpha=0.6, color='coral', label='Tratado')
axes[0,0].set_xlabel('Propensity Score'); axes[0,0].set_title('CENARIO BOM\nPropensity scores sobrepostos')
axes[0,0].legend()

# Cenario ruim
axes[0,1].hist(df_pos[df_pos.T_ruim==0]['e_ruim'], bins=30, alpha=0.6, color='steelblue', label='Controle')
axes[0,1].hist(df_pos[df_pos.T_ruim==1]['e_ruim'], bins=30, alpha=0.6, color='coral', label='Tratado')
axes[0,1].set_xlabel('Propensity Score'); axes[0,1].set_title('CENARIO RUIM\nPropensity scores SEPARADOS -- overlap pobre')
axes[0,1].legend()

# Visualizacao 2D
axes[1,0].scatter(df_pos[df_pos.T_bom==0]['idade'], df_pos[df_pos.T_bom==0]['renda_anual'],
                   alpha=0.3, s=8, color='steelblue', label='Controle')
axes[1,0].scatter(df_pos[df_pos.T_bom==1]['idade'], df_pos[df_pos.T_bom==1]['renda_anual'],
                   alpha=0.3, s=8, color='coral', label='Tratado')
axes[1,0].set_xlabel('Idade'); axes[1,0].set_ylabel('Renda anual')
axes[1,0].set_title('Cenario BOM: tratados/controles em todas as regioes')
axes[1,0].legend()

axes[1,1].scatter(df_pos[df_pos.T_ruim==0]['idade'], df_pos[df_pos.T_ruim==0]['renda_anual'],
                   alpha=0.3, s=8, color='steelblue', label='Controle')
axes[1,1].scatter(df_pos[df_pos.T_ruim==1]['idade'], df_pos[df_pos.T_ruim==1]['renda_anual'],
                   alpha=0.3, s=8, color='coral', label='Tratado')
axes[1,1].set_xlabel('Idade'); axes[1,1].set_ylabel('Renda anual')
axes[1,1].set_title('Cenario RUIM: regioes inteiras com apenas um grupo')
axes[1,1].legend()

plt.tight_layout(); plt.show()

print('IMPACTO PRATICO DE OVERLAP POBRE:')
print('  No cenario ruim, voce nao consegue estimar efeito para idosos ricos')
print('  ou jovens pobres -- ha pouco/nenhum dado contrafactual.')
print()
print('Solucoes:')
print('  1. Trim -- remover regioes sem overlap (perde generalidade)')
print('  2. Mudar estimando alvo: ATT em vez de ATE')
print('  3. Coletar mais dados na regiao de baixa overlap')
print('  4. Usar metodos robustos a overlap: Doubly Robust, TMLE')
````

# MODULO 3 -- DAGs (Directed Acyclic Graphs)

---

## A revolucao de Judea Pearl

Antes de Pearl, causalidade era discutida em palavras. Cada estatistico
tinha sua intuicao. Nao tinha matematica formal.

Pearl criou uma LINGUAGEM VISUAL para representar relacoes causais.
Ganhou Turing Award (Nobel da Computacao) em 2011 por isso.

DAG = **D**irected **A**cyclic **G**raph
- Direcionado: setas tem direcao (A -> B significa A causa B)
- Aciclico: sem loops (A nao pode causar A)
- Graph: nos e arestas

---

## Por que DAGs sao importantes para voce nessa vaga

Em qualquer projeto da Gerencia de Causal AI, antes de codar voce vai:
1. Desenhar o DAG do problema
2. Discutir com time/stakeholders se faz sentido
3. Identificar quais variaveis controlar

Quem chega na entrevista e desenha um DAG na hora demonstra maturidade.
Quem so fala em 'modelos' parece junior.

---

## Os 3 padroes fundamentais (revisao + aprofundamento)

### Padrao 1: Cadeia (Chain) -- Mediador

```
T -> M -> Y
```

Exemplo: Campanha (T) -> Cliente abre app (M) -> Cliente compra (Y).

**Pergunta:** quero efeito TOTAL ou efeito DIRETO de T em Y?

- **Total:** NAO controle M. Estima efeito completo da campanha.
- **Direto:** controle M. Estima quanto a campanha afeta compra DIRETAMENTE,
  alem do efeito via abrir app.

**Erro classico:** controlar M sempre. Voce esconde parte do efeito real.

### Padrao 2: Bifurcacao (Fork) -- Confundidor

```
     X
    / \
   v   v
   T   Y
```

X afeta T e Y. **SEMPRE controle X** se quer efeito causal de T em Y.

Exemplo: Idade (X) afeta uso de Pix (T) e NPS (Y).

### Padrao 3: Colidor (Collider)

```
T -> C <- Y
```

C eh afetado por T E por Y. **NUNCA controle C.**

Por que? Controlar collider abre um caminho backdoor 'fantasma' que
introduz correlacao espuria entre T e Y.

Exemplo classico:
- T = atletismo (causa fama)
- Y = beleza (causa fama)
- C = fama (afetada por ambos)

Se voce analisar apenas pessoas famosas (condicionar em C=1), vai descobrir
que atletismo e beleza tem correlacao NEGATIVA. Mas na populacao geral,
sao independentes. Eh artefato do collider.

---

## Backdoor Criterion (Pearl)

A regra de ouro para identificar efeito causal:

**Para estimar o efeito de T em Y, controle um conjunto Z de variaveis
que satisfaca:**
1. Z bloqueia TODOS os caminhos backdoor entre T e Y
2. Z NAO contem descendentes de T

**Caminho backdoor:** qualquer caminho de T para Y que comece com seta
APONTANDO PARA T (nao saindo de T).

Para bloquear: incluir confundidor; NAO incluir collider.

---

## Frontdoor Criterion (Pearl) -- avancado mas pode cair

Quando voce NAO consegue medir o confundidor U, mas tem um mediator M
que esta totalmente sob seu controle:

```
U
| \
v  v
T -> M -> Y
```

Voce pode usar Frontdoor: estimar T -> M e M -> Y, e combinar.

Aplicacao classica: efeito de fumar em cancer de pulmao, controlando
por 'quantidade de alcatrao no pulmao' como mediator (Pearl, 1995).

---

## Como desenhar um DAG no Itau

**Cenario real:** "Qual o efeito de habilitar o cartao virtual no engajamento do app?"

Variaveis:
- T: habilitou cartao virtual
- Y: engajamento (uso do app em 30 dias)
- Confundidores potenciais: idade, tempo de relacionamento, score de credito digital, segmento (PF/PJ)
- Mediator possivel: cliente passa a usar Pix mais (pode estar no caminho)
- Collider: 'cliente baixou app novamente' (afeitada por T e por Y)

DAG:
```
idade ---+
         |
         v
score---T ---> uso_pix ---> Y(engajamento)
         |             ^
         |             |
         +-------------+
```

Decisoes:
- Controlar idade, score (confundidores)
- NAO controlar 'cliente baixou app' (collider)
- Decidir se controla uso_pix dependendo se quer efeito total ou direto

Essa eh exatamente a discussao que voce vai ter na vaga.


````python
# DEMONSTRACAO 3.1 -- visualizacao dos 3 padroes
import networkx as nx

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# CHAIN
G1 = nx.DiGraph()
G1.add_edges_from([('T','M'), ('M','Y')])
pos1 = {'T':(0,0), 'M':(1,0), 'Y':(2,0)}
nx.draw(G1, pos1, ax=axes[0], with_labels=True, node_size=3000,
        node_color='lightblue', font_size=18, font_weight='bold',
        arrowsize=30, width=2.5, edge_color='black')
axes[0].set_title('CHAIN: Mediator\n\nNAO controle M -> efeito TOTAL\nControle M -> efeito DIRETO',
                  fontsize=12, pad=20)

# FORK
G2 = nx.DiGraph()
G2.add_edges_from([('X','T'), ('X','Y'), ('T','Y')])
pos2 = {'X':(1,1), 'T':(0,0), 'Y':(2,0)}
nx.draw(G2, pos2, ax=axes[1], with_labels=True, node_size=3000,
        node_color='lightyellow', font_size=18, font_weight='bold',
        arrowsize=30, width=2.5, edge_color='black')
axes[1].set_title('FORK: Confundidor\n\nSEMPRE controle X\n(bloqueia caminho backdoor)',
                  fontsize=12, pad=20)

# COLLIDER
G3 = nx.DiGraph()
G3.add_edges_from([('T','C'), ('Y','C')])
pos3 = {'T':(0,1), 'Y':(2,1), 'C':(1,0)}
nx.draw(G3, pos3, ax=axes[2], with_labels=True, node_size=3000,
        node_color='lightcoral', font_size=18, font_weight='bold',
        arrowsize=30, width=2.5, edge_color='black')
axes[2].set_title('COLLIDER\n\nNUNCA controle C\n(abre caminho backdoor falso)',
                  fontsize=12, pad=20)

plt.tight_layout(); plt.show()
````

````python
# DEMONSTRACAO 3.2 -- vies de collider numericamente
np.random.seed(42)
N = 5000

# T e Y SAO INDEPENDENTES (geramos assim)
T = np.random.binomial(1, 0.5, N)
Y = np.random.normal(0, 1, N)

# C eh collider: depende de T E Y
C = 1.5 * T + 1.5 * Y + np.random.normal(0, 0.3, N)

df_col = pd.DataFrame({'T': T, 'Y': Y, 'C': C})

# Correlacao real entre T e Y
corr_raw = df_col[['T','Y']].corr().iloc[0,1]

# Estratificando em quartis de C
df_col['C_quartil'] = pd.qcut(df_col['C'], 4, labels=['Q1','Q2','Q3','Q4'])

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Dados sem condicionar
axes[0].scatter(df_col['T'] + np.random.normal(0, 0.05, N), df_col['Y'],
                alpha=0.2, s=5, color='steelblue')
axes[0].set_xlabel('T'); axes[0].set_ylabel('Y')
axes[0].set_title(f'Sem condicionar em C\nCorrelacao T-Y: {corr_raw:+.4f}  (proxima de 0, como deveria)')

# Correlacao por quartil de C
correlacoes = []
for q in ['Q1','Q2','Q3','Q4']:
    sub = df_col[df_col['C_quartil']==q]
    c = sub[['T','Y']].corr().iloc[0,1]
    correlacoes.append(c)

axes[1].bar(['Q1','Q2','Q3','Q4'], correlacoes, color=['steelblue','steelblue','coral','coral'])
axes[1].axhline(0, color='black', lw=1)
axes[1].set_xlabel('Quartil de C (condicionando)')
axes[1].set_ylabel('Correlacao T-Y')
axes[1].set_title(f'Condicionando em C\nCorrelacao ESPURIA aparece -- collider bias!')
plt.tight_layout(); plt.show()

print('=== VIES DE COLLIDER -- DEMONSTRACAO NUMERICA ===')
print()
print(f'Geracao: T ~ Binomial(0.5), Y ~ Normal(0,1) -- INDEPENDENTES por construcao')
print(f'C = 1.5T + 1.5Y + ruido -- C eh collider')
print()
print(f'  Correlacao T-Y sem condicionar:           {corr_raw:+.4f}  (correto, proximo de 0)')
print(f'  Correlacao em Q1 de C (baixo):            {correlacoes[0]:+.4f}')
print(f'  Correlacao em Q4 de C (alto):             {correlacoes[3]:+.4f}')
print()
print('Voce ve correlacao NEGATIVA aparecer quando condicionou em C alto.')
print('Eh ARTEFATO -- nao existe na populacao real.')
print()
print('IMPACTO PRATICO:')
print('  Se voce inclui collider como controle em regressao causal,')
print('  voce INTRODUZ vies. Mais variaveis nao significa melhor analise.')
````

````python
# DEMONSTRACAO 3.3 -- DAG de um problema real do Itau
import networkx as nx

fig, ax = plt.subplots(figsize=(13, 9))

# Nos do DAG
G = nx.DiGraph()
edges = [
    ('Idade', 'Score_credito'),
    ('Idade', 'Habilitou_cartao'),
    ('Idade', 'Engajamento'),
    ('Score_credito', 'Habilitou_cartao'),
    ('Score_credito', 'Engajamento'),
    ('Tempo_relac', 'Habilitou_cartao'),
    ('Tempo_relac', 'Engajamento'),
    ('Habilitou_cartao', 'Uso_Pix'),
    ('Habilitou_cartao', 'Engajamento'),
    ('Uso_Pix', 'Engajamento'),
    ('Engajamento', 'Baixou_app_novo'),
    ('Habilitou_cartao', 'Baixou_app_novo'),
]
G.add_edges_from(edges)

pos = {
    'Idade': (0, 2),
    'Score_credito': (0, 1),
    'Tempo_relac': (0, 0),
    'Habilitou_cartao': (2, 2),
    'Uso_Pix': (3, 1),
    'Engajamento': (4, 2),
    'Baixou_app_novo': (5, 1),
}

# Cores indicando papel
node_colors = {
    'Idade': 'lightyellow',           # confundidor
    'Score_credito': 'lightyellow',   # confundidor
    'Tempo_relac': 'lightyellow',     # confundidor
    'Habilitou_cartao': 'lightgreen', # TRATAMENTO
    'Uso_Pix': 'lightblue',           # mediator
    'Engajamento': 'pink',            # RESPOSTA
    'Baixou_app_novo': 'lightcoral',  # collider
}
colors = [node_colors[n] for n in G.nodes()]

nx.draw(G, pos, ax=ax, with_labels=True, node_size=4500,
        node_color=colors, font_size=9, font_weight='bold',
        arrowsize=25, width=2, edge_color='gray')

# Legenda manual
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='lightyellow', edgecolor='black', label='Confundidor -> CONTROLAR'),
    Patch(facecolor='lightgreen', edgecolor='black', label='Tratamento (T)'),
    Patch(facecolor='lightblue', edgecolor='black', label='Mediator -> decida se controla'),
    Patch(facecolor='pink', edgecolor='black', label='Resposta (Y)'),
    Patch(facecolor='lightcoral', edgecolor='black', label='Collider -> NUNCA CONTROLAR'),
]
ax.legend(handles=legend_elements, loc='upper left', fontsize=10)
ax.set_title('DAG: Efeito de habilitar cartao virtual no engajamento (Itau)', fontsize=13, pad=20)
plt.tight_layout(); plt.show()

print('ANALISE DO DAG:')
print()
print('Quero estimar efeito CAUSAL de Habilitou_cartao -> Engajamento.')
print()
print('Caminhos backdoor (devem ser bloqueados):')
print('  1. Habilitou_cartao <- Idade -> Engajamento')
print('  2. Habilitou_cartao <- Score_credito -> Engajamento')
print('  3. Habilitou_cartao <- Tempo_relac -> Engajamento')
print()
print('Conjunto MINIMO de controle: {Idade, Score_credito, Tempo_relac}')
print()
print('NAO controlar:')
print('  - Uso_Pix (mediator) se quer efeito TOTAL')
print('  - Baixou_app_novo (COLLIDER) -- jamais!')
print()
print('Efeito ESTIMADO sera o efeito causal TOTAL de habilitar cartao em engajamento.')
````

# MODULO 4 -- Regressao para Controle de Confundidores

---

## O metodo mais antigo e ainda usado

Regressao linear ou logistica, com confundidores como controles,
eh o metodo causal mais antigo. Eh ensinado em econometria desde os anos 70.

Apesar de ser 'simples', ainda eh o ponto de partida de qualquer analise.
Em muitos casos, eh tudo que voce precisa.

---

## A logica matematica

Voce quer estimar E[Y(1) - Y(0)].

Sob ignorability, isso eh igual a:
```
E[Y | T=1, X] - E[Y | T=0, X], medio sobre X
```

Se voce especifica E[Y | T, X] como uma funcao LINEAR:
```
E[Y | T, X] = alpha + tau*T + beta*X
```

Entao o coeficiente tau eh exatamente o efeito causal -- sob a suposicao
de que o modelo linear esta correto.

---

## Quando regressao funciona BEM

1. Voce tem TODOS os confundidores medidos (Ignorability)
2. A relacao Y ~ X eh aproximadamente linear
3. O efeito eh homogeneo (constante em todos os subgrupos)
4. Boa cobertura: overlap entre tratados e controles

---

## Quando regressao FALHA

1. **Especificacao errada da forma funcional:** se Y ~ X eh nao-linear e voce assume linear, vies.
2. **Heterogeneidade do efeito:** se efeito varia por X, tau eh uma media ponderada estranha.
3. **Falta de overlap:** regressao extrapola para regioes sem dado -- perigoso.
4. **Confundidores nao medidos:** sempre que tem, sempre vies.

---

## Tip pratico: 'doublyrobust' simples

Voce pode controlar confundidores via:
- **Outcome regression:** modelar E[Y | T, X] e calcular diferenca de previsoes
- **Treatment modeling (propensity):** modelar P(T=1 | X) e usar para reponderar

Quando voce faz AMBOS e combina inteligentemente, voce ganha 'double robustness':
o estimador eh consistente se UM dos dois modelos esta correto. Veremos depois.


````python
# DEMONSTRACAO 4.1 -- Regressao para controlar confundidores (caso linear)
np.random.seed(42)
N = 3000

# Cenario: efeito de 'adesao a programa premium' (T) em 'gasto mensal' (Y)
# Confundidor: renda
renda = np.random.lognormal(8, 0.6, N)

# Probabilidade de aderir depende da renda
prob_T = 1/(1+np.exp(-((renda - 4000) / 2000)))
T = np.random.binomial(1, prob_T)

# Y verdadeiro: efeito = +500 do programa, + efeito da renda
Y = (
    500 * T            # efeito causal real
    + 0.3 * renda      # confundidor
    + np.random.normal(0, 200, N)
)

df_reg = pd.DataFrame({'renda': renda, 'T': T, 'Y': Y})

import statsmodels.formula.api as smf

# Modelo 1: SEM controlar renda (enviesado)
mod1 = smf.ols('Y ~ T', data=df_reg).fit()
tau1 = mod1.params['T']
se1  = mod1.bse['T']

# Modelo 2: COM controle de renda
mod2 = smf.ols('Y ~ T + renda', data=df_reg).fit()
tau2 = mod2.params['T']
se2  = mod2.bse['T']

print('=== REGRESSAO PARA CONTROLE DE CONFUNDIDORES ===')
print()
print(f'Efeito causal REAL: +500')
print()
print(f'Modelo 1 (SEM controle):  tau = {tau1:.1f} +/- {se1:.1f}  ENVIESADO')
print(f'Modelo 2 (COM controle):  tau = {tau2:.1f} +/- {se2:.1f}  CORRETO')
print()
print('Diferenca: o modelo sem controle CONFUNDE o efeito do programa')
print('com a influencia da renda (renda alta -> mais adesao + mais gasto).')
print()
print('IC 95% do efeito (modelo com controle):')
print(f'  [{tau2 - 1.96*se2:.1f}, {tau2 + 1.96*se2:.1f}]')
print()

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Visualizar o confundidor
axes[0].scatter(df_reg[df_reg.T==0]['renda'], df_reg[df_reg.T==0]['Y'],
                alpha=0.4, s=10, color='steelblue', label='Controle')
axes[0].scatter(df_reg[df_reg.T==1]['renda'], df_reg[df_reg.T==1]['Y'],
                alpha=0.4, s=10, color='coral', label='Tratado')
axes[0].set_xlabel('Renda'); axes[0].set_ylabel('Gasto mensal')
axes[0].set_title('Confundimento: tratados sao concentrados em renda alta')
axes[0].legend()

# Comparacao de estimativas
axes[1].bar(['Sem\ncontrole', 'Com\ncontrole'], [tau1, tau2],
             yerr=[1.96*se1, 1.96*se2], capsize=10,
             color=['red', 'green'], alpha=0.6)
axes[1].axhline(500, color='black', linestyle='--', lw=2, label='Efeito real')
axes[1].set_ylabel('Efeito estimado (R$)')
axes[1].set_title('Sem controle eh ENVIESADO\nIC 95%')
axes[1].legend()
plt.tight_layout(); plt.show()
````

# MODULO 5 -- Propensity Score Matching (PSM)

---

## A intuicao geometrica

Em A/B test, randomizacao garante que tratados e controles sao
estatisticamente identicos em TODAS as covariaveis.

Em dados observacionais, eles NAO sao. Como fazer eles parecerem similares?

**Resposta de Rosenbaum & Rubin (1983):** se voce condiciona no propensity score
P(T=1|X), os grupos ficam balanceados em X.

Em vez de balancear em 10 dimensoes, balanceia em 1 (o score).
**Pareou Rosenbaum & Rubin: Nobel deveria ter sido para eles.**

---

## Algoritmo PSM completo

1. **Modelar propensity score:** P(T=1|X) via logistic regression ou ML
2. **Diagnostico:** verificar overlap (positivity)
3. **Matching:** para cada tratado, encontrar controle com score similar
4. **Diagnostico apos match:** verificar balanco de covariaveis
5. **Estimar:** ATT (efeito nos tratados) como diferenca de medias na amostra pareada

---

## Decisoes praticas

### Tipo de matching
- **1:1 sem reposicao:** cada controle so pareado uma vez. Conservador.
- **1:k:** cada tratado pareado com k controles. Mais dados, mais vies.
- **Caliper:** so parear se diferenca de scores < c. Aumenta qualidade do match mas perde dados.
- **Nearest neighbor:** vizinho mais proximo (default)
- **Optimal matching:** minimiza distancia TOTAL (mais complexo, melhor balanco)

### Avaliacao do match
- **Standardized Mean Difference (SMD):** deve ser < 0.1 apos match
- **KS test:** distribuicoes devem ser similares
- **Love plot:** visualizacao do balanco antes/depois

---

## Vantagens e desvantagens

**Vantagens:**
- Intuitivo (mimica RCT)
- Robusto a especificacao incorreta de E[Y|X]
- Cria pareamento -- analise subsequente eh simples

**Desvantagens:**
- Perde dados (controles nao pareados sao descartados)
- Match exato eh impossivel com muitas covariaveis
- So estima ATT (nao ATE) se voce parea so tratados


````python
# DEMONSTRACAO 5.1 -- IMPLEMENTACAO PSM DO ZERO
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import NearestNeighbors

np.random.seed(42)
N = 4000

# Geramos dados com confundidores
df_psm = pd.DataFrame({
    'idade': np.random.normal(40, 12, N).clip(18, 80),
    'renda': np.random.lognormal(8, 0.6, N),
    'tempo_relac': np.random.exponential(5, N).clip(0.1, 30),
    'score': np.random.normal(600, 100, N).clip(300, 900),
})

# T depende dos confundidores
logit_T = (
    -3.0
    + 0.04*(df_psm['idade']-40)
    + 0.0002*(df_psm['renda']-3000)
    + 0.1*df_psm['tempo_relac']
    + 0.005*(df_psm['score']-600)
)
df_psm['T'] = np.random.binomial(1, 1/(1+np.exp(-logit_T)))

# Y verdadeiro: efeito real = +30, + confundidores
df_psm['Y'] = (
    50
    + 30 * df_psm['T']                         # efeito real = 30
    + 0.5 * (df_psm['idade'] - 40)
    + 0.005 * (df_psm['renda'] - 3000)
    + 2 * df_psm['tempo_relac']
    + 0.05 * (df_psm['score'] - 600)
    + np.random.normal(0, 10, N)
)

print(f'Tratados: {df_psm.T.sum()} ({df_psm.T.mean()*100:.1f}%)')
print(f'Controles: {(1-df_psm.T).sum()}')
print()

# ESTIMATIVA NAIVE
naive = df_psm[df_psm.T==1]['Y'].mean() - df_psm[df_psm.T==0]['Y'].mean()
print(f'Diferenca naive de medias: {naive:.2f}')
print(f'Efeito real:               30.00')
print()

# PASSO 1 -- ESTIMAR PROPENSITY SCORE
X_cols = ['idade', 'renda', 'tempo_relac', 'score']
X = df_psm[X_cols].values
T = df_psm['T'].values

ps_model = LogisticRegression(max_iter=1000)
ps_model.fit(X, T)
df_psm['ps'] = ps_model.predict_proba(X)[:, 1]

print('PASSO 1: Propensity scores estimados')
print(df_psm[['idade','renda','tempo_relac','score','T','ps']].head())
print()

# PASSO 2 -- DIAGNOSTICO DE OVERLAP
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(df_psm[df_psm.T==0]['ps'], bins=40, alpha=0.6, color='steelblue', label='Controle', density=True)
ax.hist(df_psm[df_psm.T==1]['ps'], bins=40, alpha=0.6, color='coral', label='Tratado', density=True)
ax.set_xlabel('Propensity Score'); ax.set_ylabel('Densidade')
ax.set_title('Diagnostico de Overlap (Positivity)')
ax.legend()
plt.tight_layout(); plt.show()

print('Bom overlap: ambos os grupos cobrem a maioria do intervalo [0,1].')
````

````python
# DEMONSTRACAO 5.2 -- MATCHING 1:1 com NearestNeighbors
from sklearn.neighbors import NearestNeighbors

# PASSO 3 -- MATCHING
tratados = df_psm[df_psm['T']==1].copy()
controles = df_psm[df_psm['T']==0].copy()

# Para cada tratado, encontrar o controle com PS mais proximo
nn = NearestNeighbors(n_neighbors=1)
nn.fit(controles[['ps']].values)

distances, indices = nn.kneighbors(tratados[['ps']].values)
tratados['match_id'] = controles.iloc[indices.flatten()].index.values
tratados['distance'] = distances.flatten()

# Aplicar caliper: descartar matches com distancia > 0.05
caliper = 0.05
tratados_good = tratados[tratados['distance'] < caliper].copy()

n_descartados = len(tratados) - len(tratados_good)
print(f'Tratados originais:           {len(tratados)}')
print(f'Descartados por caliper:      {n_descartados} ({n_descartados/len(tratados)*100:.1f}%)')
print(f'Tratados pareados:            {len(tratados_good)}')
print()

# Construir amostra pareada
controles_matched = controles.loc[tratados_good['match_id'].values].copy()
controles_matched['T_match'] = 0
tratados_good['T_match'] = 1

amostra_pareada = pd.concat([tratados_good, controles_matched], axis=0)

# PASSO 4 -- DIAGNOSTICO POS-MATCH (balanco de covariaveis)
def smd(grupo1, grupo2):
    """Standardized Mean Difference"""
    mean_diff = grupo1.mean() - grupo2.mean()
    pooled_sd = np.sqrt((grupo1.var() + grupo2.var()) / 2)
    return mean_diff / pooled_sd if pooled_sd > 0 else 0

print('BALANCO DE COVARIAVEIS:')
print(f"{'Variavel':<15} {'SMD ANTES':>12} {'SMD DEPOIS':>13}")
print('-'*45)
for col in X_cols:
    smd_antes = smd(df_psm[df_psm.T==1][col], df_psm[df_psm.T==0][col])
    smd_depois = smd(tratados_good[col], controles_matched[col])
    status = 'OK' if abs(smd_depois) < 0.1 else 'CHECAR'
    print(f'{col:<15} {smd_antes:>12.3f} {smd_depois:>13.3f}  {status}')
print()
print('Regra: SMD < 0.1 apos match indica BOM balanco.')
print()

# PASSO 5 -- ESTIMAR ATT
att_psm = tratados_good['Y'].mean() - controles_matched['Y'].mean()
se_att = np.sqrt(tratados_good['Y'].var()/len(tratados_good) +
                  controles_matched['Y'].var()/len(controles_matched))

print('=== ESTIMATIVAS COMPARADAS ===')
print(f'  Efeito real:                  +30.00')
print(f'  Diferenca naive de medias:    {naive:+.2f}  enviesado')
print(f'  ATT via PSM:                  {att_psm:+.2f} +/- {1.96*se_att:.2f}  CORRIGIDO')
print()
print('PSM recuperou o efeito causal verdadeiro.')
print('Eh um dos metodos mais usados quando voce nao tem A/B test.')
````

# MODULO 6 -- Inverse Propensity Weighting (IPW)

---

## A intuicao da reponderacao

Em vez de DESCARTAR controles nao-pareados (como PSM), por que nao
RECONSIDERAR todas as observacoes com pesos diferentes?

IPW: pondera cada observacao pelo INVERSO da probabilidade de receber
o tratamento que recebeu.

Resultado: voce cria uma 'pseudo-populacao' onde tratados e controles
sao balanceados.

---

## A formula

Para cada observacao i:
```
peso_i = T_i / e(X_i)  +  (1-T_i) / (1-e(X_i))
```

Onde e(X_i) = P(T=1 | X_i) eh o propensity score.

**Interpretacao:**
- Tratado raro (baixa probabilidade) recebe peso ALTO -- ele 'representa' muitos similares no controle
- Controle raro (alta probabilidade de ser tratado) recebe peso ALTO -- representa muitos similares no tratamento

Estimador IPW do ATE:
```
ATE_hat = (1/N) * sum( T_i*Y_i/e(X_i) - (1-T_i)*Y_i/(1-e(X_i)) )
```

---

## Comparacao com PSM

| Aspecto | PSM | IPW |
|---------|-----|-----|
| Mecanismo | Pareamento | Reponderacao |
| Perda de dados | Sim (controles nao pareados) | Nao |
| Estimando | ATT (geralmente) | ATE (geralmente) |
| Sensibilidade a overlap | Baixa | Alta (pesos extremos) |
| Variancia | Maior | Menor (se overlap bom) |

---

## A pegadinha critica do IPW: pesos extremos

Se e(X_i) eh proximo de 0 ou 1, o peso explode.

Exemplo: cliente com e(X) = 0.02 que foi tratado recebe peso 1/0.02 = 50.
Uma observacao 'vale' 50. Se essa observacao for atipica, contamina tudo.

**Solucao: trimming or stabilized weights**

Stabilized weights:
```
sw_i = P(T=1) * T_i / e(X_i) + P(T=0) * (1-T_i) / (1-e(X_i))
```

Trimming: descartar observacoes com e(X) fora de [0.05, 0.95].


````python
# DEMONSTRACAO 6 -- IPW
# Usando os mesmos dados do PSM

# Propensity scores ja foram estimados (df_psm['ps'])

# IPW classico (vanilla)
df_psm['peso_ipw'] = np.where(
    df_psm['T']==1,
    1 / df_psm['ps'],
    1 / (1 - df_psm['ps'])
)

# Stabilized weights
p_t = df_psm['T'].mean()
df_psm['peso_stab'] = np.where(
    df_psm['T']==1,
    p_t / df_psm['ps'],
    (1 - p_t) / (1 - df_psm['ps'])
)

# Estimar ATE via IPW
ipw_ate = (
    (df_psm['T'] * df_psm['Y'] / df_psm['ps']).sum() / len(df_psm)
    - ((1 - df_psm['T']) * df_psm['Y'] / (1 - df_psm['ps'])).sum() / len(df_psm)
)

# Estimar com stabilized
ipw_stab_ate = (
    np.sum(df_psm['T'] * df_psm['Y'] * df_psm['peso_stab']) / np.sum(df_psm['T'] * df_psm['peso_stab']) -
    np.sum((1-df_psm['T']) * df_psm['Y'] * df_psm['peso_stab']) / np.sum((1-df_psm['T']) * df_psm['peso_stab'])
)

print('=== IPW ===')
print()
print(f'Efeito real:                       +30.00')
print(f'Diferenca naive:                   {naive:+.2f}')
print(f'PSM (ATT):                         {att_psm:+.2f}')
print(f'IPW classico (ATE):                {ipw_ate:+.2f}')
print(f'IPW stabilized (ATE):              {ipw_stab_ate:+.2f}')
print()

# Diagnostico: distribuicao dos pesos
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df_psm['peso_ipw'], bins=50, color='coral', edgecolor='black')
axes[0].set_xlabel('Peso IPW classico')
axes[0].set_title(f'Pesos classicos -- pode ter outliers\nMax: {df_psm.peso_ipw.max():.2f}')

axes[1].hist(df_psm['peso_stab'], bins=50, color='steelblue', edgecolor='black')
axes[1].set_xlabel('Peso estabilizado')
axes[1].set_title(f'Pesos estabilizados -- mais comportados\nMax: {df_psm.peso_stab.max():.2f}')
plt.tight_layout(); plt.show()

print('Pesos extremos sao perigosos. Cheque sempre o histograma!')
print('Se max > 10-20, considere trimming ou stabilized weights.')
````

# MODULO 7 -- Doubly Robust (DR) Estimation

---

## A genialidade do estimador AIPW

AIPW = Augmented Inverse Probability Weighting

Combina DOIS modelos:
1. **Outcome model:** mu(X, T) = E[Y | X, T] -- previsao de Y
2. **Treatment model:** e(X) = P(T=1 | X) -- propensity score

E entrega 'double robustness': **o estimador eh consistente se UM dos dois
modelos estiver correto**, nao precisa que ambos sejam.

Isso eh seguro. Se voce errou na especificacao do outcome model mas o
propensity esta certo, ok. Se errou no propensity mas o outcome esta certo, ok.
So falha se errar nos DOIS.

---

## A formula AIPW

```
ATE = (1/N) * sum(
    [T_i*(Y_i - mu_1(X_i))/e(X_i) + mu_1(X_i)]
  - [(1-T_i)*(Y_i - mu_0(X_i))/(1-e(X_i)) + mu_0(X_i)]
)
```

Onde:
- mu_1(X_i) = E[Y | T=1, X_i] (predicao se tratado)
- mu_0(X_i) = E[Y | T=0, X_i] (predicao se controle)

---

## Por que double robustness funciona

- Se outcome model esta correto: o ajuste mu(X, T) por si so estima certo
- Se propensity esta correto: o IPW por si so estima certo
- AIPW combina os dois: se um falha, o outro corrige

Eh a base do **Targeted Maximum Likelihood Estimation (TMLE)** e do
**Double Machine Learning (DML)** que veremos depois.


````python
# DEMONSTRACAO 7 -- DOUBLY ROBUST
from sklearn.ensemble import GradientBoostingRegressor

# Etapa 1: outcome models por grupo
df_psm_X = df_psm[X_cols].values
df_psm_Y = df_psm['Y'].values
df_psm_T = df_psm['T'].values

# Modelo para tratados
mu1_model = GradientBoostingRegressor(n_estimators=100, max_depth=3, random_state=42)
mu1_model.fit(df_psm_X[df_psm_T==1], df_psm_Y[df_psm_T==1])
mu1_hat = mu1_model.predict(df_psm_X)

# Modelo para controles
mu0_model = GradientBoostingRegressor(n_estimators=100, max_depth=3, random_state=42)
mu0_model.fit(df_psm_X[df_psm_T==0], df_psm_Y[df_psm_T==0])
mu0_hat = mu0_model.predict(df_psm_X)

# Propensity score ja temos
ps_hat = df_psm['ps'].values

# Estimador AIPW
term1 = df_psm_T * (df_psm_Y - mu1_hat) / ps_hat + mu1_hat
term0 = (1 - df_psm_T) * (df_psm_Y - mu0_hat) / (1 - ps_hat) + mu0_hat
ate_dr = (term1 - term0).mean()

# Bootstrap para SE
n_boot = 200
ates_boot = []
for _ in range(n_boot):
    idx = np.random.choice(len(df_psm), len(df_psm), replace=True)
    t = df_psm_T[idx]; y = df_psm_Y[idx]; p = ps_hat[idx]
    m1 = mu1_hat[idx]; m0 = mu0_hat[idx]
    a = (t*(y-m1)/p + m1 - (1-t)*(y-m0)/(1-p) - m0).mean()
    ates_boot.append(a)

ci_low, ci_high = np.percentile(ates_boot, [2.5, 97.5])

print('=== ESTIMADOR DOUBLY ROBUST (AIPW) ===')
print()
print(f'  ATE real:                 +30.00')
print(f'  Naive:                    {naive:+.2f}')
print(f'  Regressao com controle:   ~30.0')
print(f'  PSM (ATT):                {att_psm:+.2f}')
print(f'  IPW:                      {ipw_ate:+.2f}')
print(f'  Doubly Robust (AIPW):     {ate_dr:+.2f}  [{ci_low:+.2f}, {ci_high:+.2f}]')
print()
print('VANTAGEM DO DR: se voce errar UM dos modelos (outcome OU propensity)')
print('o estimador ainda eh consistente. Os outros nao tem essa garantia.')
print()
print('Por isso DR (e variantes como TMLE, DML) sao o padrao moderno.')
````

# MODULO 8 -- Difference-in-Differences (DiD)

---

## Quando voce nao consegue randomizar

Imagina: O Banco Central muda a regulacao de credito so em algumas regioes
(politica nao randomizada). Voce quer saber o impacto na inadimplencia.

Voce nao pode randomizar (regulacao eh imposta). Voce nao pode controlar
todos os confundidores (regiao tem cultura, economia, hist diferente).

**Solucao: DiD.**

Compara a MUDANCA na regiao tratada com a MUDANCA na regiao controle.
Se as regioes tinham TENDENCIAS PARALELAS antes do tratamento,
a diferenca nas mudancas eh o efeito causal.

---

## A formula

```
DiD = (Y_T_depois - Y_T_antes) - (Y_C_depois - Y_C_antes)
```

Onde T = tratado, C = controle, antes/depois do evento.

---

## A premissa critica: Parallel Trends

**Sem o tratamento, tratados e controles teriam evoluido identicamente.**

Voce NUNCA observa esse contrafactual diretamente. Mas pode verificar
**indiretamente** olhando os periodos ANTERIORES ao tratamento.

Se as series sao paralelas antes, eh razoavel supor que continuariam
paralelas depois (na ausencia do tratamento).

Se as series ja divergiam antes, parallel trends viola e DiD eh enviesado.

---

## Implementacao via regressao

A forma classica eh com interacao:
```
Y = beta0 + beta1*Tratado + beta2*Pos + beta3*(Tratado*Pos) + erro
```

O coeficiente beta3 da interacao **eh** o DiD.

---

## Aplicacoes no Itau

- Lancamento de feature em alguns segmentos (mas nao todos)
- Mudanca regulatoria por regiao
- Nova politica de credito que afeta determinado setor
- Atualizacao do app que so foi para um grupo (canary release)
- Telemetria: feature ativada em servidores de uma regiao especifica

---

## Variantes modernas

- **Event Study DiD:** estima efeito em multiplos momentos pre/pos
- **Staggered DiD:** tratamento ocorre em momentos diferentes para diferentes grupos
- **Synthetic Control:** quando voce so tem UMA unidade tratada (regiao, empresa)
- **Callaway & Sant'Anna (2021):** moderna correcao para DiD com timing diferente


````python
# DEMONSTRACAO 8.1 -- DiD CLASSICO
np.random.seed(42)

# Cenario: nova politica em SP afeta inadimplencia. Comparar com RJ.
# Dados mensais, 24 meses (12 antes, 12 depois)

n_meses = 24
T_mes = 11  # mes 12 (indice 11) eh quando entrou a politica

# Tendencia base (mesma para ambos)
tendencia = 0.005 * np.arange(n_meses)  # leve aumento ao longo do tempo

# SP (tratado)
sp_inadim = 0.08 + tendencia + np.random.normal(0, 0.005, n_meses)
# A politica reduz inadim em 0.02 pontos (efeito real)
sp_inadim[T_mes+1:] -= 0.02

# RJ (controle, sem politica)
rj_inadim = 0.075 + tendencia + np.random.normal(0, 0.005, n_meses)

# Plotar
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(range(n_meses), sp_inadim, 'b-o', ms=6, lw=2, label='SP (tratado)')
ax.plot(range(n_meses), rj_inadim, 'r-s', ms=6, lw=2, label='RJ (controle)')
ax.axvline(T_mes, color='gray', linestyle='--', lw=2, label='Politica inicia')
ax.set_xlabel('Mes'); ax.set_ylabel('Taxa de inadimplencia')
ax.set_title('DiD: politica em SP vs RJ -- tendencias paralelas ANTES')
ax.legend()
plt.tight_layout(); plt.show()

# DiD MANUAL
sp_antes = sp_inadim[:T_mes+1].mean()
sp_depois = sp_inadim[T_mes+1:].mean()
rj_antes = rj_inadim[:T_mes+1].mean()
rj_depois = rj_inadim[T_mes+1:].mean()

did = (sp_depois - sp_antes) - (rj_depois - rj_antes)

print('=== DIFFERENCE-IN-DIFFERENCES ===')
print()
print(f'{"":>15} {"Antes":>10} {"Depois":>10} {"Delta":>10}')
print('-'*50)
print(f'{"SP (tratado)":>15} {sp_antes:>10.4f} {sp_depois:>10.4f} {sp_depois-sp_antes:>+10.4f}')
print(f'{"RJ (controle)":>15} {rj_antes:>10.4f} {rj_depois:>10.4f} {rj_depois-rj_antes:>+10.4f}')
print(f'{"DIFERENCA":>15}                       {did:>+10.4f}')
print()
print(f'Efeito causal REAL:  -0.020')
print(f'DiD estimou:         {did:+.4f}')
print()
print('DiD recupera o efeito causal. Bonito!')
````

````python
# DEMONSTRACAO 8.2 -- DiD via REGRESSAO (forma profissional)
import statsmodels.formula.api as smf

# Criar dataframe long
df_did = pd.DataFrame({
    'mes': np.tile(np.arange(n_meses), 2),
    'estado': ['SP']*n_meses + ['RJ']*n_meses,
    'tratado': [1]*n_meses + [0]*n_meses,
    'pos': [1 if m > T_mes else 0 for m in np.tile(np.arange(n_meses), 2)],
    'inadim': np.concatenate([sp_inadim, rj_inadim])
})

df_did['tratado_x_pos'] = df_did['tratado'] * df_did['pos']

modelo_did = smf.ols('inadim ~ tratado + pos + tratado_x_pos', data=df_did).fit()
print(modelo_did.summary().tables[1])
print()
did_coef = modelo_did.params['tratado_x_pos']
did_se = modelo_did.bse['tratado_x_pos']
print(f'Coeficiente do termo de interacao (DiD): {did_coef:.4f}')
print(f'Erro padrao:                              {did_se:.4f}')
print(f'IC 95%: [{did_coef - 1.96*did_se:.4f}, {did_coef + 1.96*did_se:.4f}]')
print()
print('A interacao tratado_x_pos eh o DiD estimado.')
print('Este eh o jeito profissional de fazer DiD.')
print()
print('Erros padrao: em DiD real, deve-se usar erros AGRUPADOS (clustered SE) por')
print('unidade (estado), porque observacoes da mesma unidade nao sao independentes.')
````

````python
# DEMONSTRACAO 8.3 -- TESTE DE PARALLEL TRENDS
# Antes de aceitar DiD, voce DEVE checar se as tendencias sao paralelas pre-tratamento

# Plot ja foi feito. Mas teste formal:
df_pre = df_did[df_did['pos']==0].copy()  # so antes do tratamento

# Testar interacao 'tratado * mes' no periodo pre
mod_pt = smf.ols('inadim ~ tratado + mes + tratado:mes', data=df_pre).fit()

print('=== TESTE DE PARALLEL TRENDS ===')
print()
print(f'Coeficiente da interacao tratado:mes (PRE-tratamento):')
print(f'  Valor:    {mod_pt.params["tratado:mes"]:.5f}')
print(f'  P-valor:  {mod_pt.pvalues["tratado:mes"]:.4f}')
print()
if mod_pt.pvalues['tratado:mes'] > 0.05:
    print('Nao se rejeita parallel trends. SEGURO usar DiD.')
else:
    print('Parallel trends REJEITADA. DiD eh enviesado.')
    print('Considerar Synthetic Control ou outros metodos.')
print()
print('REGRA:')
print('  Coeficiente da interacao perto de 0 + p-valor alto = parallel trends ok.')
print('  Coeficiente da interacao significativo = SP e RJ ja divergiam antes -- viola.')
````

# MODULO 9 -- Regression Discontinuity Design (RDD)

---

## A intuicao do desconto natural

Imagina: o Itau aprova credito acima de score 600 e nega abaixo.

- Cliente com score 599: nega
- Cliente com score 601: aprova

Esses dois clientes sao PRATICAMENTE IDENTICOS (mesmo perfil de risco).
A unica diferenca eh que um ficou logo abaixo do cutoff e outro logo acima.

Se voce comparar a evolucao do score de inadimplencia desses dois grupos,
**voce esta proximo de um experimento natural**. Como se voce tivesse
randomizado quem fica abaixo e quem fica acima.

Esse eh o coracao do RDD.

---

## Pressupostos do RDD

1. **Existe um cutoff claro:** acima de c, recebe; abaixo, nao.
2. **A variavel de atribuicao (running variable) eh continua perto do cutoff:** nao ha 'salto' nas covariaveis ao redor.
3. **Indivíduos nao manipulam a posicao:** voce nao consegue mudar artificialmente sua score para ficar acima.

---

## Tipos de RDD

### Sharp RDD
Tratamento eh DETERMINISTICO: acima do cutoff sempre recebe, abaixo nunca.

### Fuzzy RDD
Acima do cutoff aumenta a PROBABILIDADE de receber, mas nao garante.
Estima-se via 2SLS (Two-Stage Least Squares).

---

## Implementacao

1. Restringir analise a janela estreita ao redor do cutoff (bandwidth)
2. Estimar regressao com running variable + dummy de tratamento
3. Calcular salto descontinuo no cutoff
4. Esse salto eh o efeito causal local (LATE -- Local ATE)

---

## Vantagens

- Premissas mais fracas que outros metodos observacionais
- Inferencia VALIDA mesmo com confundidores nao medidos (no cutoff)
- Visual: voce VE o salto no grafico

## Desvantagens

- Estima efeito LOCAL (so vale perto do cutoff)
- Precisa de muitos dados perto do cutoff
- Sensivel a especificacao da forma funcional


````python
# DEMONSTRACAO 9 -- RDD
np.random.seed(42)
N = 5000

# Running variable: score de credito (300-900)
score = np.random.normal(600, 100, N).clip(300, 900)

# Cutoff em 600
cutoff = 600
T = (score >= cutoff).astype(int)  # recebe credito

# Y: taxa de inadimplencia
# Sem tratamento (Y0): cresce com score baixo
Y0 = 0.30 - 0.0003*score + np.random.normal(0, 0.03, N)
Y0 = Y0.clip(0, 1)

# Tratamento (credito aprovado) reduz inadim em 0.08 (efeito real)
Y1 = Y0 - 0.08

# Observado
Y = np.where(T==1, Y1, Y0)

df_rdd = pd.DataFrame({'score': score, 'T': T, 'Y': Y})

# RDD: focar em janela estreita
bandwidth = 50
df_local = df_rdd[(df_rdd.score > cutoff - bandwidth) & (df_rdd.score < cutoff + bandwidth)].copy()
df_local['score_centered'] = df_local['score'] - cutoff

import statsmodels.formula.api as smf
mod_rdd = smf.ols('Y ~ T + score_centered + T:score_centered', data=df_local).fit()

salto = mod_rdd.params['T']

print('=== REGRESSION DISCONTINUITY DESIGN ===')
print()
print(f'Cutoff:       {cutoff}')
print(f'Bandwidth:    {bandwidth}')
print(f'N na janela:  {len(df_local)}')
print()
print(f'Salto estimado no cutoff (LATE):  {salto:.4f}')
print(f'Efeito real:                       -0.0800')
print(f'Erro padrao:                       {mod_rdd.bse["T"]:.4f}')
print()

# Plot
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Dados completos
axes[0].scatter(df_rdd[df_rdd.T==0]['score'], df_rdd[df_rdd.T==0]['Y'],
                alpha=0.3, s=8, color='red', label='Negado')
axes[0].scatter(df_rdd[df_rdd.T==1]['score'], df_rdd[df_rdd.T==1]['Y'],
                alpha=0.3, s=8, color='green', label='Aprovado')
axes[0].axvline(cutoff, color='black', linestyle='--', lw=2, label=f'Cutoff = {cutoff}')
axes[0].set_xlabel('Score'); axes[0].set_ylabel('Inadimplencia')
axes[0].set_title('Dados completos -- veja o salto no cutoff')
axes[0].legend()

# Zoom na janela
axes[1].scatter(df_local[df_local.T==0]['score'], df_local[df_local.T==0]['Y'],
                alpha=0.4, s=20, color='red', label='Negado')
axes[1].scatter(df_local[df_local.T==1]['score'], df_local[df_local.T==1]['Y'],
                alpha=0.4, s=20, color='green', label='Aprovado')
axes[1].axvline(cutoff, color='black', linestyle='--', lw=2)

# Linhas de regressao locais
x_left = np.linspace(cutoff - bandwidth, cutoff, 50)
y_left = mod_rdd.params['Intercept'] + mod_rdd.params['score_centered']*(x_left - cutoff)
x_right = np.linspace(cutoff, cutoff + bandwidth, 50)
y_right = (mod_rdd.params['Intercept'] + mod_rdd.params['T'] +
           (mod_rdd.params['score_centered'] + mod_rdd.params['T:score_centered'])*(x_right - cutoff))

axes[1].plot(x_left, y_left, 'r-', lw=2.5)
axes[1].plot(x_right, y_right, 'g-', lw=2.5)
axes[1].set_xlabel('Score (zoom no cutoff)'); axes[1].set_ylabel('Inadimplencia')
axes[1].set_title(f'RDD na janela -- salto = {salto:.4f}')
axes[1].legend()
plt.tight_layout(); plt.show()
````

# MODULO 10 -- Instrumental Variables (IV)

---

## A situacao mais dificil em causalidade

Voce nao tem RCT. Voce suspeita de confundidores nao medidos.
Voce nao tem cutoff para RDD. Voce nao tem antes/depois para DiD.

Nessa situacao desesperadora, **IV pode salvar voce.**

---

## A ideia genial

Encontre uma variavel Z que:
1. AFETA T (causalmente)
2. NAO afeta Y diretamente (so via T)
3. NAO eh afetada pelos confundidores

Z chama-se INSTRUMENTO.

Exemplo classico (Angrist & Krueger 1991): efeito de educacao em salario.
- T = anos de educacao
- Y = salario
- Z = trimestre do nascimento (afeta quando voce sai da escola compulsoria, sem afetar salario diretamente)

---

## A formula via 2SLS

1. **Primeiro estagio:** regredir T em Z e covariaveis. Pegar T_hat (predicao).
2. **Segundo estagio:** regredir Y em T_hat e covariaveis. Coeficiente eh o efeito causal.

Por isso 2SLS = Two-Stage Least Squares.

---

## Aplicacao em telemetria

Cenario hipotetico:
- T = cliente usa nova feature
- Y = engajamento
- Confundidor nao medido: 'cliente eh tech-savvy'
- Instrumento Z: receber recomendacao de uma campanha randomizada que sugere a feature

Z afeta T (recomendacao aumenta uso) mas nao Y diretamente (so via T).
2SLS recupera o efeito causal mesmo com confundidor latente.

---

## Pegadinhas

- **Weak instrument:** se Z afeta T fracamente, estimativa eh instavel
- **Exclusion restriction:** voce assume que Z so afeta Y via T -- NAO testavel diretamente
- **Heterogeneity:** IV estima LATE (Local ATE) -- so para 'compliers'


````python
# DEMONSTRACAO 10 -- IV via 2SLS
np.random.seed(42)
N = 3000

# Confundidor LATENTE
U = np.random.normal(0, 1, N)

# Instrumento Z: randomizado, afeta T mas nao Y diretamente
Z = np.random.binomial(1, 0.5, N)

# T depende de Z (instrumento forte) e do confundidor U
T = np.random.binomial(1, 1/(1+np.exp(-(-0.5 + 1.5*Z + 1.0*U))))

# Y depende de T (efeito real = 2) e U (confundidor)
Y = 2*T + 1.5*U + np.random.normal(0, 1, N)

df_iv = pd.DataFrame({'Y': Y, 'T': T, 'Z': Z, 'U': U})

# Estimativas
# 1) NAIVE: OLS de Y em T (ignorando confundidor)
import statsmodels.formula.api as smf
mod_naive = smf.ols('Y ~ T', data=df_iv).fit()
ate_naive = mod_naive.params['T']

# 2) Com confundidor U (ideal mas impossivel na realidade)
mod_ideal = smf.ols('Y ~ T + U', data=df_iv).fit()
ate_ideal = mod_ideal.params['T']

# 3) IV (2SLS) sem precisar de U
# Primeiro estagio: T ~ Z
first = smf.ols('T ~ Z', data=df_iv).fit()
df_iv['T_hat'] = first.fittedvalues

# Segundo estagio: Y ~ T_hat
second = smf.ols('Y ~ T_hat', data=df_iv).fit()
ate_iv = second.params['T_hat']

# Tambem com linearmodels para erro padrao correto
try:
    from linearmodels.iv import IV2SLS
    iv_model = IV2SLS.from_formula('Y ~ 1 + [T ~ Z]', df_iv).fit()
    ate_iv_correct = iv_model.params['T']
    se_iv = iv_model.std_errors['T']
except:
    ate_iv_correct = ate_iv
    se_iv = second.bse['T_hat']

print('=== INSTRUMENTAL VARIABLES (IV / 2SLS) ===')
print()
print(f'  Efeito causal REAL:                       2.000')
print(f'  OLS naive (sem controlar U):              {ate_naive:.4f}  ENVIESADO')
print(f'  OLS com U (impossivel na realidade):      {ate_ideal:.4f}')
print(f'  IV 2SLS (sem precisar de U):              {ate_iv_correct:.4f} +/- {se_iv:.4f}')
print()
print('IV recupera o efeito causal MESMO com confundidor nao medido,')
print('desde que o instrumento Z seja valido.')
print()

# Diagnostico: forca do instrumento
from scipy import stats as scipy_stats
f_stat = first.fvalue
print('DIAGNOSTICOS:')
print(f'  F-statistic do primeiro estagio: {f_stat:.2f}')
if f_stat > 10:
    print('  Instrumento FORTE (regra de bolso: F > 10)')
else:
    print('  Instrumento FRACO -- cuidado!')
print()
print('A exclusion restriction (Z afeta Y so via T) NAO eh testavel.')
print('Voce justifica com conhecimento de dominio.')
````

# MODULO 11 -- Double Machine Learning (DML)

---

## A combinacao perfeita: ML + causalidade

Voce viu:
- Regressao causal eh limitada (so funciona bem se forma funcional eh certa)
- ML eh otimo para previsao (mas nao eh causal naturalmente)

**Double Machine Learning (Chernozhukov et al. 2018)** combina os dois.
Usa ML para AJUDAR a inferencia causal sem sacrificar a validez.

Esta no top de papers mais citados em econometria moderna.

---

## A receita basica

Para estimar o efeito causal tau de T em Y, com confundidores X:

1. **Modelo de Y:** treina ML para prever Y a partir de X (sem T)
   - Obtem residuos: e_Y = Y - mu(X)
2. **Modelo de T:** treina ML para prever T a partir de X
   - Obtem residuos: e_T = T - e(X)
3. **Regressao residual:** regredir e_Y em e_T
   - O coeficiente eh tau

Por que residuos? Frisch-Waugh-Lovell theorem: regressao parcial.

---

## O detalhe critico: Cross-Fitting

Se voce usa o MESMO dado para treinar mu e e_X, voce overfitt.
Cross-fitting: divide os dados em K folds. Para cada fold:
- Treina mu e e_X nos OUTROS K-1 folds
- Calcula residuos no fold atual

Resultado: residuos honestos, sem overfitting.

---

## Vantagens do DML

- Permite ML poderoso (Random Forest, XGBoost) sem perder propriedades estatisticas
- Funciona com muitos confundidores (high-dimensional X)
- Tem propriedades teoricas: distribuicao normal, IC valido
- Eh o estado da arte para estimacao de ATE em observacional

---

## Onde usar

- Voce tem MUITOS confundidores (>10)
- Forma funcional Y ~ X eh provavelmente nao-linear
- Quer combinar rigor causal com poder de ML


````python
# DEMONSTRACAO 11 -- DOUBLE ML
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.model_selection import KFold

np.random.seed(42)
N = 3000

# Multi-dim confounders
X = np.random.randn(N, 8)

# T depende de X (nao linear)
logit_T = 0.5*X[:,0] + 0.3*X[:,1]**2 - 0.4*X[:,2]*X[:,3] + 0.2*X[:,4]
T = np.random.binomial(1, 1/(1+np.exp(-logit_T)))

# Y depende de T (efeito real = 3) e X (nao linear)
Y = (
    3 * T
    + 0.5*X[:,0]
    + 0.3*X[:,1]**2
    + 0.7*X[:,2]*X[:,3]
    + 0.4*X[:,4]
    + np.random.normal(0, 1, N)
)

# OLS simples (mal especificado)
import statsmodels.api as sm
X_design = np.column_stack([T, X])
X_design = sm.add_constant(X_design)
mod_ols = sm.OLS(Y, X_design).fit()
ate_ols = mod_ols.params[1]  # coeficiente de T

# Double Machine Learning manual (cross-fitting)
kf = KFold(n_splits=5, shuffle=True, random_state=42)
e_Y = np.zeros(N)
e_T = np.zeros(N)

for tr_idx, va_idx in kf.split(X):
    # Treinar nos K-1 folds, prever no K-esimo
    rf_Y = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42)
    rf_Y.fit(X[tr_idx], Y[tr_idx])
    mu_Y = rf_Y.predict(X[va_idx])
    e_Y[va_idx] = Y[va_idx] - mu_Y

    rf_T = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
    rf_T.fit(X[tr_idx], T[tr_idx])
    mu_T = rf_T.predict_proba(X[va_idx])[:, 1]
    e_T[va_idx] = T[va_idx] - mu_T

# Regressao residual
ate_dml = (e_Y * e_T).sum() / (e_T * e_T).sum()
se_dml = np.sqrt(((e_Y - ate_dml * e_T)**2).sum() / ((e_T*e_T).sum()**2) * N)

print('=== DOUBLE MACHINE LEARNING ===')
print()
print(f'  Efeito real:               +3.0000')
print(f'  OLS classico:              {ate_ols:.4f}  (especificacao linear errada)')
print(f'  DML com Random Forest:     {ate_dml:.4f}  +/-  {1.96*se_dml:.4f}')
print()
print('DML captura nao-linearidades nos confundidores via ML, sem perder validade.')
print('Eh o metodo padrao para causal ML hoje.')
````

# MODULO 12 -- Causal Forests (Athey & Wager)

---

## De ATE para CATE

Causal Forest (Wager & Athey, 2018) eh Random Forest adaptado para
estimar **efeitos heterogeneos (CATE)**.

Em vez de uma previsao 'qual o ATE', voce obtem:
'qual o efeito tau(x) para cada perfil de cliente x?'

---

## Como funciona (intuicao)

Random Forest classico: cada split de arvore minimiza erro de previsao de Y.

Causal Forest: cada split maximiza a HETEROGENEIDADE do efeito tau.

Isso significa: a arvore eh construida para identificar SUBGRUPOS com
efeitos diferentes.

Resultado: previsao do efeito causal por individuo.

---

## Por que isso eh poderoso no Itau

Permite personalizacao causal. Em vez de:
- 'Esta campanha eleva NPS em +0.5 em media'

Voce diz:
- 'Esta campanha eleva NPS em +1.5 para PJ comercio, em +0.2 para industria,
  e REDUZ em -0.3 para servicos'

Voce so manda campanha para quem se beneficia. ROI explode.

---

## Implementacao via EconML

```python
from econml.dml import CausalForestDML

cf = CausalForestDML(
    model_y=RandomForestRegressor(),
    model_t=RandomForestClassifier(),
    n_estimators=500,
    random_state=42
)
cf.fit(Y, T, X=X, W=W)  # X: features para CATE, W: controles

# CATE individual
cate_individual = cf.effect(X_novo)
```

---

## Diagnosticos importantes

1. **Heterogeneity test:** o efeito varia significativamente? Calibration test (Chernozhukov 2018)
2. **GATE -- Group ATE:** dividir em quintis de CATE e ver se o efeito real corresponde
3. **Best Linear Projection:** quais features mais influenciam o CATE?

---

## Quando NAO usar Causal Forest

- N pequeno (<2000): RF precisa de dados
- Poucos confundidores (<5): regressao eh suficiente
- Voce quer ATE global: DML simples eh mais eficiente


````python
# DEMONSTRACAO 12 -- CAUSAL FOREST via EconML
try:
    from econml.dml import CausalForestDML
    from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier

    np.random.seed(42)
    N = 5000

    # Features que MODIFICAM o efeito do tratamento (Heterogeneity)
    X_het = np.random.randn(N, 5)

    # Confundidores comuns
    W_conf = np.random.randn(N, 5)

    # T depende dos confundidores
    logit_T = 0.5*W_conf[:,0] + 0.3*W_conf[:,1]
    T = np.random.binomial(1, 1/(1+np.exp(-logit_T)))

    # Efeito HETEROGENEO: depende de X[:,0]
    # Quem tem X[:,0] alto se beneficia mais
    tau_individual = 2 + 1.5 * X_het[:,0]

    Y = tau_individual * T + W_conf @ np.array([0.5, 0.3, 0.4, 0.2, 0.1]) + np.random.normal(0, 1, N)

    # Causal Forest
    cf = CausalForestDML(
        model_y=RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42),
        model_t=RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42),
        n_estimators=300,
        random_state=42
    )
    cf.fit(Y, T, X=X_het, W=W_conf)

    cate_estimado = cf.effect(X_het)

    # Comparar CATE real vs estimado
    print('=== CAUSAL FOREST DML ===')
    print()
    print(f'ATE real (media): {tau_individual.mean():.3f}')
    print(f'ATE estimado:     {cate_estimado.mean():.3f}')
    print()
    print('Distribuicao do efeito individual estimado:')
    print(f'  Min:    {cate_estimado.min():.3f}')
    print(f'  P25:    {np.percentile(cate_estimado, 25):.3f}')
    print(f'  Mediana:{np.median(cate_estimado):.3f}')
    print(f'  P75:    {np.percentile(cate_estimado, 75):.3f}')
    print(f'  Max:    {cate_estimado.max():.3f}')
    print()

    # Visualizar
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    # CATE estimado vs CATE real
    axes[0].scatter(tau_individual, cate_estimado, alpha=0.3, s=5)
    axes[0].plot([tau_individual.min(), tau_individual.max()],
                  [tau_individual.min(), tau_individual.max()],
                  'k--', lw=2, label='Perfeito')
    axes[0].set_xlabel('CATE real'); axes[0].set_ylabel('CATE estimado')
    axes[0].set_title('Causal Forest: estimativa por individuo')
    axes[0].legend()

    # Heterogeneidade em X[:,0]
    axes[1].scatter(X_het[:,0], cate_estimado, alpha=0.3, s=5, color='steelblue')
    axes[1].set_xlabel('X[:,0] (feature que modifica efeito)')
    axes[1].set_ylabel('CATE estimado')
    axes[1].set_title('Heterogeneidade detectada\nClientes com X[0] alto tem efeito maior')
    plt.tight_layout(); plt.show()

    print('UTILIDADE NO ITAU:')
    print('  Identificar SUBGRUPOS com maior efeito de campanha')
    print('  Personalizar acoes: so mandar para quem tau > limiar')
    print('  ROI da estrategia: somar tau dos selecionados, comparar com custo')
except ImportError:
    print('EconML nao instalado. Execute: pip install econml')
    print()
    print('Causal Forest seria o estado da arte para CATE heterogeneo.')
````

# MODULO 13 -- Meta-Learners (S, T, X, R, DR)

---

## A familia de meta-learners

Frameworks que combinam qualquer ML model para estimar CATE.

### S-Learner (Single Model)
- Treina UM modelo: Y ~ X + T (com T como feature)
- CATE(x) = mu(x, T=1) - mu(x, T=0)
- Simples mas pode subestimar efeitos pequenos se T tem feature importance baixa

### T-Learner (Two Models)
- Treina DOIS modelos: um para tratados, outro para controles
- mu1(x): Y ~ X em T=1
- mu0(x): Y ~ X em T=0
- CATE(x) = mu1(x) - mu0(x)
- Captura interacoes complexas mas perde dados em cada modelo

### X-Learner (Cross-Learner) -- Kunzel et al. 2019
- Combina T-Learner com correcao via propensity score
- Funciona bem quando ha imbalance entre tratados e controles
- Mais sofisticado, melhor performance em geral

### R-Learner -- Nie & Wager 2017
- Usa orthogonalizacao tipo DML
- Robusto a especificacao errada dos nuisance models
- Padrao moderno

### DR-Learner -- Kennedy 2020
- Combina propensity score + outcome model
- Doubly robust por construcao
- Estado da arte

---

## Qual escolher?

Regra de bolso:
- **Inicio:** T-Learner (simples, interpretavel)
- **Imbalance grande:** X-Learner
- **Confundidores complexos:** R-Learner ou DR-Learner
- **Producao real:** Causal Forest DML (combina ML com inferencia)


````python
# DEMONSTRACAO 13 -- T-LEARNER VS S-LEARNER
from sklearn.ensemble import GradientBoostingRegressor

np.random.seed(42)
N = 3000
X = np.random.randn(N, 5)
T = np.random.binomial(1, 0.5, N)

# Efeito heterogeneo: depende de X[:,0]
tau = 2 + 1.5 * X[:,0]
Y = tau * T + 0.5*X[:,0] + 0.3*X[:,1] + np.random.normal(0, 1, N)

# Treino-teste
X_tr, X_te, T_tr, T_te, Y_tr, Y_te = train_test_split(
    X, T, Y, test_size=0.3, random_state=42
)
tau_te = 2 + 1.5 * X_te[:,0]  # CATE real no teste

# S-LEARNER: um unico modelo Y ~ X, T
X_T_tr = np.column_stack([X_tr, T_tr])
s_model = GradientBoostingRegressor(n_estimators=200, max_depth=3, random_state=42)
s_model.fit(X_T_tr, Y_tr)
X_T1 = np.column_stack([X_te, np.ones(len(X_te))])
X_T0 = np.column_stack([X_te, np.zeros(len(X_te))])
tau_s = s_model.predict(X_T1) - s_model.predict(X_T0)

# T-LEARNER: dois modelos
mu1 = GradientBoostingRegressor(n_estimators=200, max_depth=3, random_state=42)
mu1.fit(X_tr[T_tr==1], Y_tr[T_tr==1])
mu0 = GradientBoostingRegressor(n_estimators=200, max_depth=3, random_state=42)
mu0.fit(X_tr[T_tr==0], Y_tr[T_tr==0])
tau_t = mu1.predict(X_te) - mu0.predict(X_te)

# Comparar
from sklearn.metrics import mean_squared_error
rmse_s = np.sqrt(mean_squared_error(tau_te, tau_s))
rmse_t = np.sqrt(mean_squared_error(tau_te, tau_t))

print('=== COMPARACAO DE META-LEARNERS ===')
print()
print(f'  S-Learner -- RMSE no CATE: {rmse_s:.4f}')
print(f'  T-Learner -- RMSE no CATE: {rmse_t:.4f}')
print()
print(f'  Media CATE real:           {tau_te.mean():.3f}')
print(f'  Media CATE S-Learner:      {tau_s.mean():.3f}')
print(f'  Media CATE T-Learner:      {tau_t.mean():.3f}')
print()

# Plot
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].scatter(tau_te, tau_s, alpha=0.3, s=5, color='steelblue')
axes[0].plot([tau_te.min(), tau_te.max()], [tau_te.min(), tau_te.max()], 'k--', lw=2)
axes[0].set_xlabel('CATE real'); axes[0].set_ylabel('CATE S-Learner')
axes[0].set_title(f'S-Learner -- RMSE = {rmse_s:.3f}')

axes[1].scatter(tau_te, tau_t, alpha=0.3, s=5, color='coral')
axes[1].plot([tau_te.min(), tau_te.max()], [tau_te.min(), tau_te.max()], 'k--', lw=2)
axes[1].set_xlabel('CATE real'); axes[1].set_ylabel('CATE T-Learner')
axes[1].set_title(f'T-Learner -- RMSE = {rmse_t:.3f}')
plt.tight_layout(); plt.show()
````

# MODULO 14 -- Aplicacao em Telemetria

---

## O contexto da vaga

A vaga #823124 e clara: voce vai analisar dados de telemetria,
observabilidade, infraestrutura. Esses dados tem caracteristicas
especificas que mudam como voce faz causal AI.

**Caracteristicas dos dados de telemetria:**
- Volume enorme (milhoes de eventos por minuto)
- Series temporais com sazonalidade complexa
- Multiplas dimensoes (servidor, regiao, versao, usuario)
- Interferencia natural (servidores compartilham infra)
- Eventos raros importantes (incidentes, picos)

---

## Tipos de problemas causais em telemetria

### 1. Efeito de feature flag
Time de tecnologia ativa nova feature em alguns servidores.
Voce quer saber: a feature CAUSA reducao de latencia?
**Metodo:** DiD se ativou em algumas regioes; A/B se feature flag por usuario.

### 2. Efeito de deploy
Versao nova do app foi deployada. Conversao subiu 5%.
Foi pela versao ou foi por sazonalidade?
**Metodo:** Synthetic Control com unidades similares como controles.

### 3. Causa raiz de incidente
Servico X teve latencia alta as 14h. Mudancas concorrentes:
- Deploy as 13:45
- Pico de trafego as 14:00
- Banco de dados manutencao as 14:15
Qual foi a causa?
**Metodo:** Granger causality + DAGs.

### 4. Otimizacao de infraestrutura
Aumentamos memoria RAM em alguns servidores. Reduziu erros 5xx?
**Metodo:** RDD se ha cutoff de upgrade; PSM se foi gradual.

---

## Pegadinhas especificas de telemetria

### 1. Sazonalidade mascara efeitos
Latencia naturalmente sobe em horario de pico. Se voce comparar
ANTES e DEPOIS de deploy sem controlar horario, vai achar 'pioria'
quando na verdade eh so o horario.

### 2. Auto-correlation viola independencia
Latencia no minuto T eh correlacionada com minuto T-1.
Erros padrao classicos (sem cluster ou Newey-West) sao errados.

### 3. Treatment spillover via dependencias
Voce muda servidor A. Servidor B chama A via API. B tambem eh
afetado, contaminando 'controle'. SUTVA viola.

### 4. Selection bias em logs
So loga 'erros que causaram chamada de suporte'. Voce nao tem
contrafactual para 'erros que o usuario nao reportou'.

## A receita aplicada para a vaga


````python
# CASO COMPLETO -- SIMULACAO REALISTA DE TELEMETRIA NO ITAU
# Cenario: feature de cache foi ativada em alguns servidores em 15/05/2026.
# Voce eh o cientista de Causal AI. A pergunta: qual o efeito CAUSAL na latencia?

np.random.seed(42)

# Definicao do experimento
n_servidores = 100
n_dias = 60      # 30 antes + 30 depois
dia_intervencao = 30

# Caracteristicas dos servidores (covariaveis)
servidores = pd.DataFrame({
    'servidor_id': range(n_servidores),
    'memoria_gb': np.random.choice([16, 32, 64, 128], n_servidores, p=[0.4, 0.3, 0.2, 0.1]),
    'cpu_cores': np.random.choice([4, 8, 16, 32], n_servidores, p=[0.3, 0.4, 0.2, 0.1]),
    'regiao': np.random.choice(['SP', 'RJ', 'NE'], n_servidores, p=[0.5, 0.3, 0.2]),
    'versao_so': np.random.choice(['v1', 'v2', 'v3'], n_servidores, p=[0.2, 0.5, 0.3]),
})

# Mecanismo de selecao: servidores com MAIS memoria foram tratados primeiro
# (vies endogeno!) -- gestor de plataforma achou que 'melhor servidor merece feature primeiro'
prob_tratamento = 1 / (1 + np.exp(-((servidores['memoria_gb'] - 48) / 20)))
servidores['tratado'] = np.random.binomial(1, prob_tratamento)

# Gerar dados diarios de latencia
registros = []
for _, sv in servidores.iterrows():
    for dia in range(n_dias):
        # Latencia base depende de memoria, CPU, regiao
        lat_base = (
            120
            - 0.8 * sv['memoria_gb']     # mais memoria = menos latencia
            - 1.5 * sv['cpu_cores']      # mais CPU = menos latencia
            + (15 if sv['regiao'] == 'NE' else 0)
            + (5 if sv['versao_so'] == 'v1' else 0)
        )

        # Sazonalidade fim de semana
        sazon = 8 * (dia % 7 in [5, 6])

        # Tendencia geral
        tendencia = 0.2 * dia

        # Ruido diario
        ruido = np.random.normal(0, 5)

        # EFEITO REAL DA FEATURE: -15ms (mas so apos dia_intervencao)
        if sv['tratado'] == 1 and dia >= dia_intervencao:
            efeito_feature = -15
        else:
            efeito_feature = 0

        latencia = lat_base + sazon + tendencia + efeito_feature + ruido

        registros.append({
            'servidor_id': sv['servidor_id'],
            'dia': dia,
            'pos_intervencao': int(dia >= dia_intervencao),
            'latencia_ms': latencia,
            'memoria_gb': sv['memoria_gb'],
            'cpu_cores': sv['cpu_cores'],
            'regiao': sv['regiao'],
            'versao_so': sv['versao_so'],
            'tratado': sv['tratado'],
        })

df_tel = pd.DataFrame(registros)
print(f'Dataset: {len(df_tel):,} registros')
print(f'Servidores: {df_tel.servidor_id.nunique()}')
print(f'Dias: {df_tel.dia.nunique()}')
print(f'Tratados: {servidores.tratado.sum()} servidores ({servidores.tratado.mean()*100:.0f}%)')
print()
print(df_tel.head())
````

````python
# PASSO 1 -- ANALISE EXPLORATORIA
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Series temporais agregadas
media_diaria = df_tel.groupby(['dia', 'tratado'])['latencia_ms'].mean().unstack()
axes[0,0].plot(media_diaria.index, media_diaria[0], 'b-o', ms=4, label='Controle', alpha=0.7)
axes[0,0].plot(media_diaria.index, media_diaria[1], 'r-s', ms=4, label='Tratado', alpha=0.7)
axes[0,0].axvline(dia_intervencao, color='black', linestyle='--', lw=2, label='Feature ativada')
axes[0,0].set_xlabel('Dia'); axes[0,0].set_ylabel('Latencia media (ms)')
axes[0,0].set_title('Series temporais por grupo')
axes[0,0].legend()

# Distribuicao de latencia pre vs pos
for i, periodo in enumerate(['Pre-intervencao', 'Pos-intervencao']):
    sub = df_tel[df_tel['pos_intervencao'] == i]
    axes[0,1].hist(sub[sub['tratado']==1]['latencia_ms'], bins=40, alpha=0.4,
                    color=['red','darkred'][i], label=f'Tratado {periodo}')
axes[0,1].set_xlabel('Latencia (ms)'); axes[0,1].set_title('Distribuicao de latencia -- tratados')
axes[0,1].legend()

# Distribuicao do confundidor (memoria) por grupo
axes[1,0].hist(servidores[servidores.tratado==0]['memoria_gb'], bins=10,
                alpha=0.6, color='steelblue', label='Controle')
axes[1,0].hist(servidores[servidores.tratado==1]['memoria_gb'], bins=10,
                alpha=0.6, color='coral', label='Tratado')
axes[1,0].set_xlabel('Memoria (GB)'); axes[1,0].set_title('DESBALANCO: tratados tem mais memoria')
axes[1,0].legend()

# Latencia por memoria
for mem in [16, 32, 64, 128]:
    sub = df_tel[(df_tel.memoria_gb==mem) & (df_tel.pos_intervencao==0)]
    axes[1,1].plot([mem]*len(sub.tratado.unique()), [sub[sub.tratado==t]['latencia_ms'].mean() for t in sub.tratado.unique()], 'o', ms=10, label=f'{mem}GB' if mem==16 else None)
axes[1,1].set_xlabel('Memoria (GB)'); axes[1,1].set_ylabel('Latencia (ms)')
axes[1,1].set_title('Latencia decresce com memoria (confundidor)')

plt.tight_layout(); plt.show()

print('DIAGNOSTICOS:')
print('  1. Tratados estao em servidores com mais memoria (desbalanco)')
print('  2. Mais memoria -> menos latencia (confundimento)')
print('  3. Aparente reducao de latencia em tratados pode ser por memoria, nao feature')
print()
print('Comparar medias direto seria ENGANOSO. Precisamos de metodo causal.')
````

````python
# PASSO 2 -- ESTIMATIVA NAIVE (incorreta)
lat_T1_pos = df_tel[(df_tel.tratado==1) & (df_tel.pos_intervencao==1)]['latencia_ms'].mean()
lat_T0_pos = df_tel[(df_tel.tratado==0) & (df_tel.pos_intervencao==1)]['latencia_ms'].mean()
naive_pos = lat_T1_pos - lat_T0_pos

print('ESTIMATIVA NAIVE (comparar tratados vs controles POS-intervencao):')
print(f'  Latencia tratados pos:    {lat_T1_pos:.2f} ms')
print(f'  Latencia controles pos:   {lat_T0_pos:.2f} ms')
print(f'  Diferenca:                {naive_pos:+.2f} ms')
print()
print(f'  Efeito real:              -15.00 ms')
print()
print('A diferenca naive nao representa o efeito real -- confundimento por memoria.')
````

````python
# PASSO 3 -- DiD (metodo correto para esse caso)
import statsmodels.formula.api as smf

df_tel['tratado_x_pos'] = df_tel['tratado'] * df_tel['pos_intervencao']

# DiD via regressao com clustered SE (importante em telemetria)
mod_did = smf.ols(
    'latencia_ms ~ tratado + pos_intervencao + tratado_x_pos + memoria_gb + cpu_cores + C(regiao) + C(versao_so)',
    data=df_tel
).fit(cov_type='cluster', cov_kwds={'groups': df_tel['servidor_id']})

did_efeito = mod_did.params['tratado_x_pos']
did_se = mod_did.bse['tratado_x_pos']

print('=== DIFFERENCE-IN-DIFFERENCES ===')
print()
print(f'Coeficiente de tratado*pos (DiD):  {did_efeito:+.3f}')
print(f'Erro padrao (clustered):           {did_se:.3f}')
print(f'IC 95%: [{did_efeito - 1.96*did_se:+.3f}, {did_efeito + 1.96*did_se:+.3f}]')
print(f'P-valor:                           {mod_did.pvalues["tratado_x_pos"]:.4f}')
print()
print(f'Efeito real:                       -15.00 ms')
print()
print('DiD recupera o efeito causal com IC apertado. ESTA EH SUA ESTIMATIVA OFICIAL.')
print()
print('NOTA TECNICA: usamos clustered SE por servidor_id porque observacoes do')
print('mesmo servidor sao correlacionadas no tempo. Padrao em telemetria.')
````

````python
# PASSO 4 -- VERIFICAR PARALLEL TRENDS (validacao essencial do DiD)
df_pre = df_tel[df_tel.pos_intervencao == 0].copy()

mod_pt = smf.ols(
    'latencia_ms ~ tratado + dia + tratado:dia + memoria_gb + cpu_cores',
    data=df_pre
).fit(cov_type='cluster', cov_kwds={'groups': df_pre['servidor_id']})

print('=== TESTE DE PARALLEL TRENDS (PRE-INTERVENCAO) ===')
print()
print(f'Coef. tratado:dia (pre):  {mod_pt.params["tratado:dia"]:+.4f}')
print(f'P-valor:                  {mod_pt.pvalues["tratado:dia"]:.4f}')
print()
if mod_pt.pvalues['tratado:dia'] > 0.05:
    print('Nao se rejeita parallel trends. DiD eh seguro.')
else:
    print('Parallel trends rejeitada -- DiD eh suspeito. Considere Synthetic Control.')
````

````python
# PASSO 5 -- ROBUSTEZ: PSM + DiD
# Para reforcar a credibilidade, vamos parear servidores tratados com controles similares
# e DEPOIS rodar DiD na amostra pareada

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import OneHotEncoder

# Propensity score por SERVIDOR (nao por linha)
servidores_X = pd.get_dummies(servidores, columns=['regiao', 'versao_so'], drop_first=True)
X_ps = servidores_X[[c for c in servidores_X.columns if c not in ['servidor_id', 'tratado']]].values

ps_model = LogisticRegression(max_iter=1000)
ps_model.fit(X_ps, servidores_X['tratado'])
servidores_X['ps'] = ps_model.predict_proba(X_ps)[:, 1]

# Matching 1:1 com caliper
tratados_sv = servidores_X[servidores_X.tratado==1]
controles_sv = servidores_X[servidores_X.tratado==0]

nn = NearestNeighbors(n_neighbors=1)
nn.fit(controles_sv[['ps']].values)
dists, idxs = nn.kneighbors(tratados_sv[['ps']].values)

caliper = 0.10
mask = dists.flatten() < caliper
tratados_match = tratados_sv[mask]['servidor_id'].values
controles_match_idx = idxs.flatten()[mask]
controles_match = controles_sv.iloc[controles_match_idx]['servidor_id'].values

ids_match = list(tratados_match) + list(controles_match)
df_match = df_tel[df_tel.servidor_id.isin(ids_match)].copy()

print(f'Servidores pareados: {len(ids_match)} ({len(ids_match)/len(servidores)*100:.0f}% do total)')
print()

# DiD na amostra pareada
mod_did_match = smf.ols(
    'latencia_ms ~ tratado + pos_intervencao + tratado_x_pos',
    data=df_match
).fit(cov_type='cluster', cov_kwds={'groups': df_match['servidor_id']})

did_match_efeito = mod_did_match.params['tratado_x_pos']
did_match_se = mod_did_match.bse['tratado_x_pos']

print('=== DiD APOS PSM (robustez) ===')
print(f'Efeito estimado: {did_match_efeito:+.3f} +/- {1.96*did_match_se:.3f}')
print()
````

````python
# PASSO 6 -- HETEROGENEIDADE -- a feature beneficia mais ALGUNS servidores?
# Vamos rodar DiD separado por categorias

print('=== HETEROGENEIDADE POR CATEGORIA ===')
print()

# Por regiao
print('Por regiao:')
for regiao in ['SP', 'RJ', 'NE']:
    sub = df_tel[df_tel.regiao == regiao]
    mod = smf.ols('latencia_ms ~ tratado + pos_intervencao + tratado_x_pos + memoria_gb + cpu_cores', data=sub).fit()
    eff = mod.params['tratado_x_pos']
    se = mod.bse['tratado_x_pos']
    print(f'  {regiao}: efeito = {eff:+.2f} ms [{eff-1.96*se:+.2f}, {eff+1.96*se:+.2f}]')
print()

# Por nivel de memoria
print('Por nivel de memoria:')
for mem in [16, 32, 64, 128]:
    sub = df_tel[df_tel.memoria_gb == mem]
    if len(sub) > 100:
        try:
            mod = smf.ols('latencia_ms ~ tratado + pos_intervencao + tratado_x_pos + cpu_cores', data=sub).fit()
            eff = mod.params['tratado_x_pos']
            se = mod.bse['tratado_x_pos']
            print(f'  {mem}GB: efeito = {eff:+.2f} ms [{eff-1.96*se:+.2f}, {eff+1.96*se:+.2f}]')
        except: pass
print()
print('INTERPRETACAO PARA O GESTOR:')
print('  - Feature efetiva no agregado: ~-15 ms')
print('  - Efeito relativamente homogeneo entre regioes')
print('  - ROI: depende da carga em cada servidor; servidores de baixa memoria')
print('    poderiam se beneficiar MAIS proporcionalmente')
print('  - Recomendacao: rollout completo, monitorando deteccao de outliers')
````

## Bloco A -- Fundamentos (Q1-15)

---

### Q1 (F) -- Qual a diferenca entre correlacao e causalidade?

**Resposta modelo:**
Correlacao mede associacao estatistica entre duas variaveis -- elas variam juntas.
Causalidade mede o efeito de uma INTERVENCAO em uma variavel sobre outra.

Correlacao pode existir sem causalidade por: (1) confundimento -- terceira
variavel afeta as duas; (2) reverso -- Y causa X; (3) selecao -- amostra
filtrada.

Exemplo: clientes que ligam para o banco tem NPS mais baixo. Correlacao
forte (r=-0.92). Mas LIGAR nao causa NPS baixo. Insatisfacao previa causa
ambos: cliente liga porque esta insatisfeito, e da NPS baixo porque esta
insatisfeito.

---

### Q2 (F) -- O que eh o problema fundamental da inferencia causal?

**Resposta modelo:**
Para cada unidade, existem dois resultados potenciais: Y(0) se nao recebe
tratamento e Y(1) se recebe. O efeito causal individual eh Y(1) - Y(0).

O problema fundamental: voce so observa um dos dois. Se a pessoa foi
tratada, voce ve Y(1) e perde Y(0). Voce nunca observa o contrafactual.

Solucao pratica: estimar EFEITOS MEDIOS (ATE, ATT) sob suposicoes
verificaveis -- nunca efeitos individuais sem mais estrutura.

---

### Q3 (F) -- Defina ATE, ATT e CATE.

**Resposta modelo:**
- ATE (Average Treatment Effect): efeito medio na populacao toda.
  E[Y(1) - Y(0)]. 'Se tratassemos todos vs ninguem, qual a diferenca media?'

- ATT (ATE on Treated): efeito medio so nos que receberam tratamento.
  E[Y(1) - Y(0) | T=1]. 'Nos que receberam, qual o efeito?'

- CATE (Conditional ATE): efeito condicional a caracteristicas X.
  E[Y(1) - Y(0) | X=x]. Base para personalizacao.

ATE = ATT * P(T=1) + ATC * P(T=0).

---

### Q4 (F) -- Quais sao as 3 suposicoes para inferencia causal observacional?

**Resposta modelo:**
1. **SUTVA**: Stable Unit Treatment Value Assumption. Tratamento de uma
   unidade nao afeta outra, e existe so uma versao do tratamento.
2. **Ignorability** (Conditional Exchangeability): apos controlar pelos
   confundidores observaveis X, tratamento eh efetivamente aleatorio.
   Y(0), Y(1) ⊥ T | X. Equivalente a 'no unmeasured confounders'.
3. **Positivity** (Overlap): 0 < P(T=1|X=x) < 1 para todo x. Cada unidade
   tem chance positiva de receber ambos os tratamentos.

Sem essas suposicoes, voce nao tem causalidade -- so correlacao.

---

### Q5 (M) -- Como verificar Positivity na pratica?

**Resposta modelo:**
1. Estimar propensity score e(x) = P(T=1|X=x) via logistic ou ML.
2. Plotar distribuicoes de e(x) por grupo T=0 e T=1.
3. Identificar regioes onde so um grupo aparece -- overlap viola.

Solucoes:
- **Trim:** restringir analise a regiao de overlap comum (perde generalidade).
- **Mudar estimando:** ATT em vez de ATE se controles sao escassos.
- **Stabilized weights:** ajuda em IPW quando ps proximo de extremos.

---

### Q6 (D) -- Como argumentar que Ignorability vale na pratica?

**Resposta modelo:**
Ignorability eh INVERIFICAVEL diretamente -- voce nao consegue testar se
falta confundidor com os dados.

Estrategia:
1. Listar exaustivamente todos os mecanismos pelos quais T e Y se relacionam.
2. Verificar se mediu cada confundidor desses mecanismos.
3. Usar conhecimento de dominio (entrevistar gestores que decidiram T).
4. Fazer **sensitivity analysis** (E-value de VanderWeele, Rosenbaum bounds,
   Cinelli & Hazlett 2020) para quantificar quao forte um confundidor nao
   medido precisaria ser para anular meu efeito.

Resposta forte em entrevista: 'Nao tenho certeza absoluta. Por isso fiz
sensitivity analysis. Meu efeito sobrevive a confundidor nao medido com
forca de ate U=2x as covariaveis observadas. Eh robusto.'

---

### Q7 (M) -- Em um DAG: X -> T -> Y e X -> Y. Quem eh X?

**Resposta modelo:**
X eh um **confundidor** (fork). Afeta T e Y diretamente.

Para identificar efeito causal de T em Y, devemos controlar X --
bloquear o caminho backdoor T <- X -> Y.

Metodos: regressao com X como covariavel; PSM em X; IPW com e(X).

---

### Q8 (M) -- Em um DAG: T -> M -> Y. O que eh M? Devo controlar?

**Resposta modelo:**
M eh um **mediator** -- esta no caminho causal de T para Y.

Depende do que eu quero estimar:
- Efeito TOTAL de T em Y: NAO controlar M (controlar 'desliga' o caminho).
- Efeito DIRETO de T em Y (ignorando o que passa por M): controlar M.

Erro classico: cientista de dados controla 'todas as variaveis' incluindo
mediators, e fica confuso sobre por que efeito parece pequeno.

---

### Q9 (M) -- O que eh um collider? O que acontece se eu controlar?

**Resposta modelo:**
Collider eh uma variavel afetada por T E por Y (T -> C <- Y).

Se voce controla collider (condiciona em C), abre um caminho backdoor falso
e INTRODUZ correlacao espuria entre T e Y.

Exemplo classico: T = atletismo, Y = beleza, C = fama de Hollywood.
Atletas + belos -> mais famosos. Se voce analisar SO famosos, descobre
correlacao negativa entre atletismo e beleza. Eh artefato.

Em ML: incluir collider como feature em modelo causal DEGRADA a estimativa.

---

### Q10 (D) -- Diferencie Backdoor e Frontdoor criterion.

**Resposta modelo:**
**Backdoor:** para estimar T -> Y, identificar conjunto Z que bloqueia
todos os caminhos backdoor (que comecam com seta entrando em T), sem
incluir descendentes de T.

**Frontdoor:** quando voce NAO consegue medir o confundidor U, mas tem
um mediator M que esta totalmente sob seu controle (T -> M -> Y, e U
afeta T mas nao M diretamente). Estima T -> M e M -> Y separadamente,
e combina.

Aplicacao classica do Frontdoor: Pearl (1995) -- efeito de fumar em
cancer de pulmao, usando alcatrao como mediator quando 'genetica' (U)
nao eh medida.

---

### Q11 (F) -- Quando RCT (A/B test) eh impossivel?

**Resposta modelo:**
1. **Etico:** nao posso forcar pacientes a fumar para testar cancer.
2. **Caro:** experimento exige amostra grande, tempo, custo.
3. **Ja aconteceu:** o evento ja ocorreu (mudanca regulatoria), nao tem
   como voltar atras.
4. **Imposto:** tratamento eh decisao do governo/cliente, nao do pesquisador.
5. **Interferencia:** se tratamentos se contaminam (redes sociais), RCT
   classico nao funciona.

Nesses casos, Causal AI observacional eh a unica opcao.

---

### Q12 (M) -- Por que randomizacao eh tao poderosa?

**Resposta modelo:**
Randomizacao garante que tratados e controles sao IDENTICOS em expectativa --
nao so nas covariaveis MEDIDAS, mas tambem nas NAO MEDIDAS.

Isso significa que toda variavel U (mesmo as que nao sei que existem)
fica balanceada entre os grupos. Comparacao de medias eh estimativa
nao-viesada do efeito causal.

Eh por isso que RCT eh o padrao ouro: nenhum metodo observacional
oferece essa garantia.

---

### Q13 (D) -- O que eh do-calculus?

**Resposta modelo:**
Sistema formal de Judea Pearl para manipular expressoes causais.

Distingue:
- P(Y|X) -- conditional: voce OBSERVA X = x e ve a distribuicao de Y.
- P(Y|do(X)) -- intervention: voce INTERVEM, forcando X = x, e ve Y.

Sao geralmente DIFERENTES. P(NPS | liga para banco) eh -0.92 (correlacao).
P(NPS | do(forcar a ligar)) eh ~0 (nao causa).

Do-calculus tem 3 regras (Pearl, 1995) que permitem converter expressoes
intervencionais em condicionais, usando o DAG. Quando consegue converter,
o efeito eh IDENTIFICAVEL a partir de dados observacionais.

---

### Q14 (M) -- O que eh um experimento natural?

**Resposta modelo:**
Situacao da vida real onde a 'natureza' (lei, evento externo, regra
arbitraria) gera variacao quase-aleatoria no tratamento.

Exemplos:
- Loteria militar nos EUA (Angrist 1990): efeito de servir em Vietnam.
- Inundacoes: efeito de mudanca de endereco em filhos.
- Cutoffs administrativos: aprovacao de credito acima de score X.
- Limites de idade: 18 anos para direito a algo.

Voce usa esses 'experimentos da natureza' como instrumentos ou em RDD.

---

### Q15 (D) -- Diferencie identificacao e estimacao.

**Resposta modelo:**
**Identificacao:** sob que suposicoes o efeito causal eh expressavel a
partir de quantidades observaveis? Eh uma questao TEORICA.

**Estimacao:** dado que esta identificado, como estimo na pratica com
dados finitos? Eh uma questao COMPUTACIONAL.

Voce pode identificar mas nao estimar bem (amostra pequena).
Voce pode estimar bem (modelo flexivel) mas nao estar identificado
(suposicoes violadas) -- e ai a estimativa eh enviesada por ML poderoso
que seja.

---


## Bloco B -- Metodos Classicos (Q16-30)

---

### Q16 (F) -- Como funciona PSM?

**Resposta modelo:**
1. Estimar propensity score e(x) = P(T=1|X) via logistic regression ou ML.
2. Para cada tratado, encontrar controle com e(x) similar (nearest neighbor).
3. Aplicar caliper (ex: 0.05) para descartar pares com distancia grande.
4. Avaliar balanco apos match: SMD < 0.1 nas covariaveis.
5. Estimar ATT como diferenca de medias na amostra pareada.

PSM mimica RCT a posteriori: cria grupos comparaveis.

---

### Q17 (M) -- Quais limitacoes do PSM?

**Resposta modelo:**
1. Perde dados (controles nao pareados sao descartados).
2. So balanceia em X MEDIDO -- se ha confundidor nao medido, falha.
3. Match em multiplas dimensoes eh impossivel (curse of dimensionality);
   por isso reduzimos ao propensity score 1D.
4. Sensivel a especificacao do propensity model.
5. Estima ATT, nao ATE.
6. Variancia inferiorcia: ignorar incerteza do propensity score subestima
   SE -- usar bootstrap.

---

### Q18 (M) -- O que eh IPW e quando usar?

**Resposta modelo:**
Inverse Propensity Weighting. Pondera cada observacao pelo inverso da
probabilidade de receber o tratamento que recebeu.

```
ATE = (1/N) * sum( T_i*Y_i/e(X_i) - (1-T_i)*Y_i/(1-e(X_i)) )
```

Usar quando:
- Quer ATE (nao ATT).
- Nao quer descartar dados (PSM perde).
- Overlap razoavel (sem propensity score perto de 0 ou 1).

Pegadinha: pesos extremos. Use stabilized weights ou trim em [0.05, 0.95].

---

### Q19 (D) -- O que torna AIPW (DR) doubly robust?

**Resposta modelo:**
AIPW combina outcome model (mu(X,T)) com propensity model (e(X)).

```
ATE = (1/N) * sum(
    [T*(Y-mu_1(X))/e(X) + mu_1(X)]
  - [(1-T)*(Y-mu_0(X))/(1-e(X)) + mu_0(X)]
)
```

Double robustness: o estimador eh consistente se UM dos dois modelos
estiver corretamente especificado, nao precisa que ambos sejam.

Isso eh seguranca: se erra outcome mas acerta propensity, ok; se erra
propensity mas acerta outcome, ok; so falha se errar nos dois.

Base de TMLE e Double Machine Learning -- estado da arte hoje.

---

### Q20 (M) -- Diferencie PSM e IPW.

**Resposta modelo:**

| Aspecto | PSM | IPW |
|---------|-----|-----|
| Mecanismo | Pareamento | Reponderacao |
| Perda de dados | Sim | Nao |
| Estimando | ATT | ATE |
| Sensivel a overlap pobre | Menos | Mais |
| Variancia | Maior | Menor (se overlap bom) |
| Intuicao | Mimica RCT | Cria pseudo-populacao |

Em pratica: combinar com DR (AIPW) eh melhor que um isolado.

---

### Q21 (M) -- O que eh SMD e como interpretar?

**Resposta modelo:**
Standardized Mean Difference. Mede balanco de uma covariavel entre
tratados e controles, normalizado pelo desvio padrao agrupado.

```
SMD = (mean_T1 - mean_T0) / pooled_SD
```

Regra de bolso:
- SMD < 0.1: bom balanco.
- 0.1 < SMD < 0.25: aceitavel para alguns autores.
- SMD > 0.25: desbalanco preocupante.

Calcular APOS matching: deve estar abaixo do limite em todas as
covariaveis relevantes.

---

### Q22 (D) -- Quando voce escolheria Doubly Robust em vez de regressao simples?

**Resposta modelo:**
Quando:
1. Voce nao tem certeza da forma funcional correta (linear? polinomial? interacoes?)
2. Voce quer protecao contra erro de especificacao
3. Voce tem muitos confundidores e quer usar ML neles (extensao DML)
4. Voce quer reportar 'protecao dupla' para audiencia tecnica

Quando NAO:
- N pequeno (DR tem variancia maior em pequenas amostras)
- Voce tem certeza do modelo correto (regressao eh mais eficiente)

---

### Q23 (M) -- Como detectar weak overlap?

**Resposta modelo:**
1. Estimar propensity scores.
2. Plotar histogramas de e(X) por grupo.
3. Identificar regioes onde so um grupo aparece (densidade ~0 do outro).
4. Calcular percentis: minimo de e(X) entre tratados deve ser > 0.05;
   maximo entre controles deve ser < 0.95.

Indicadores de problema:
- Max(peso IPW) > 20.
- Variancia dos pesos > 10.
- Mais de 5% de observacoes com e(X) < 0.01 ou > 0.99.

---

### Q24 (M) -- O que eh trimming?

**Resposta modelo:**
Descartar observacoes com propensity score em extremos (ex: e(X) < 0.05
ou e(X) > 0.95).

Vantagem: reduz variancia, controla pesos extremos.
Desvantagem: perde generalidade -- a estimativa nao se aplica a regioes
descartadas.

Em paper, sempre reportar: 'aplicamos trimming em [0.05, 0.95]; resultados
robustos ao limite escolhido (sensitivity analysis).'

---

### Q25 (D) -- Qual o problema com regressao quando ha heterogeneidade do efeito?

**Resposta modelo:**
Regressao linear assume efeito CONSTANTE: Y = alpha + tau*T + beta*X.

Se efeito real varia por X (CATE), tau eh uma MEDIA PONDERADA estranha --
nao eh ATE, nao eh ATT.

Solucao: incluir interacoes T*X. Ou usar metodos que estimam CATE
diretamente (T-Learner, Causal Forest).

Em entrevista: 'Regressao com interacao captura heterogeneidade ate certo
ponto. Para padroes complexos, prefiro Causal Forest DML.'

---

### Q26 (M) -- Por que controlar 'todas as variaveis' eh ruim?

**Resposta modelo:**
Tres razoes:
1. Pode incluir COLLIDERS -- abre vies espurio.
2. Pode incluir MEDIATORS -- esconde efeito real.
3. Pode incluir descendentes de T -- viola backdoor criterion.

A escolha de variaveis vem do DAG, nao do dataset.

Exemplo: 'campanha aumenta vendas' -- incluir 'cliente comprou produto premium'
como controle? NAO. Eh descendente da campanha, viola.

---

### Q27 (M) -- Como lidar com tratamento continuo (nao binario)?

**Resposta modelo:**
Opcoes:
1. **Discretizar** em quartis e tratar como categorico (perde info).
2. **Dose-response function:** estimar E[Y(t)] para varios niveis de t.
3. **Generalized Propensity Score (GPS)**: extensao de PSM para T continuo.
4. **DML continuo:** Chernozhukov tem extensao.

Em telemetria, T continuo eh comum (ex: 'quanto de memoria adicional').

---

### Q28 (D) -- Como tratar SUTVA violation por interferencia?

**Resposta modelo:**
1. **Cluster randomization:** unidade de aleatorizacao eh o cluster
   (regiao, agencia), nao individuo.
2. **Network analysis:** modelar grafo de interferencia explicitamente
   (Aronow & Samii 2017).
3. **Saturation design:** variar a INTENSIDADE do tratamento em clusters.
4. **Two-stage randomization:** sortear cluster, depois unidades dentro.

Em telemetria: agrupar servidores por sub-rede, tratar como cluster.

---

### Q29 (M) -- Estabilidade do propensity score em ML.

**Resposta modelo:**
ML para propensity score (RF, XGBoost) tem vantagens:
- Captura nao-linearidades automaticamente.
- Lida com interacoes complexas.

Mas tem riscos:
- Overfitting: propensity inflado leva a IPW instavel.
- Calibration: ML preditivo pode otimizar AUC sem calibrar bem.

Solucao: cross-fitting (Double ML). Treinar em folds, prever em holdout.
Reduz overfitting e mantem validacao estatistica.

---

### Q30 (D) -- Por que cross-fitting eh importante em DML?

**Resposta modelo:**
Sem cross-fitting, o mesmo dado eh usado para treinar nuisance models
(mu, e) E para estimar tau. Isso causa overfitting bias.

Cross-fitting:
1. Divide dados em K folds.
2. Para fold k: treina mu, e nos OUTROS K-1 folds.
3. Calcula residuos no fold k.
4. Agrega residuos de todos os folds.

Garantia teorica (Chernozhukov 2018): com cross-fitting, ML poderoso
pode ser usado SEM perder propriedades estatisticas (normalidade, IC valido).


## Bloco C -- Quase-Experimentais (Q31-40)

---

### Q31 (M) -- Quando usar DiD em vez de PSM/IPW?

**Resposta modelo:**
DiD requer:
1. Dados em painel (mesma unidade observada antes e depois).
2. Grupo controle nao tratado.
3. Tendencias paralelas pre-tratamento.

DiD vence PSM/IPW quando:
- Voce TEM dados antes e depois -- aproveite!
- Confundidores INVARIANTES no tempo (raca, geografia) -- DiD elimina por construcao.
- Voce nao consegue medir todos os confundidores estaticos.

Em telemetria: A/B test forneceu controle, mas voce tem dados longitudinais.
DiD eh natural.

---

### Q32 (M) -- Como testar parallel trends?

**Resposta modelo:**
1. **Visual:** plotar series temporais por grupo. Se paralelas pre-evento,
   intuitivamente ok.
2. **Estatistico:** regredir Y em T*tempo no periodo PRE-tratamento. Se
   coeficiente da interacao nao eh significativo, nao se rejeita paralelas.
3. **Placebo test:** rodar DiD em um periodo onde nao houve intervencao
   real. Deve dar zero. Se da efeito, paralelas estao quebradas.
4. **Event study:** estimar efeitos em multiplos meses pre/pos. Pre-coeficientes
   devem ser nao significativos.

---

### Q33 (D) -- O que fazer se parallel trends viola?

**Resposta modelo:**
Opcoes:
1. **Synthetic Control:** constroi controle como combinacao ponderada de
   nao-tratados (Abadie 2010).
2. **Controles temporais:** incluir interacoes tempo*X para permitir
   tendencias diferentes por covariavel.
3. **DiD com matching:** parear unidades antes de DiD para garantir
   trajetorias similares.
4. **Modelar a divergencia:** se tendencia diverge linearmente, modelar
   tendencia diferente por grupo.
5. **Aceitar limite:** reportar como suspeito, fazer sensitivity analysis.

---

### Q34 (M) -- Diferencie Sharp e Fuzzy RDD.

**Resposta modelo:**
**Sharp:** tratamento eh DETERMINADO pelo cutoff. Score >= 600 sempre
recebe; score < 600 nunca recebe. Compliance perfeita.

**Fuzzy:** o cutoff AUMENTA a probabilidade mas nao garante. Score >= 600
tem 80% de chance; score < 600 tem 20%. Outros fatores influenciam.

Sharp: comparacao direta perto do cutoff.
Fuzzy: 2SLS, instrumento = dummy(score >= cutoff).

---

### Q35 (D) -- Como detectar manipulacao em RDD?

**Resposta modelo:**
Teste de **McCrary (2008):** se ha manipulacao, voce observa um SALTO
na densidade da running variable no cutoff.

Exemplo: pessoas ajustam tempo de trabalho para ficar acima de cutoff
de aposentadoria. Densidade dispara logo acima.

Visualizacao: histograma da running variable. Verifica se ha pico ou
vale no cutoff.

Se manipulacao detectada, RDD eh suspeito -- pessoas selecionam onde
ficam, viola exogeneidade.

---

### Q36 (M) -- Como escolher bandwidth em RDD?

**Resposta modelo:**
Trade-off:
- Bandwidth pequeno: pouco vies (unidades muito similares ao cutoff)
  mas pouca amostra (alta variancia).
- Bandwidth grande: mais dados (baixa variancia) mas pode extrapolar
  longe do cutoff (vies).

Metodos:
1. **Bandwidth otimo MSE-minimizing** (Imbens & Kalyanaraman 2012;
   Calonico, Cattaneo & Titiunik 2014).
2. **Cross-validation:** testar diferentes bandwidths e ver estabilidade.
3. **Reportar varios:** mostrar resultados para bandwidth pequeno, medio,
   grande -- se estaveis, robusto.

---

### Q37 (M) -- Quando IV eh apropriado?

**Resposta modelo:**
IV eh util quando:
1. Voce SUSPEITA de confundidor nao medido.
2. Voce tem uma variavel Z que afeta T mas nao Y diretamente.
3. Voce esta disposto a aceitar suposicao de exclusion restriction
   (nao testavel).

Exemplos de instrumentos:
- Loteria militar (Vietnam).
- Distancia geografica (acesso a servico).
- Mes de nascimento (regra de matricula escolar).
- Randomizacao parcial (ofertas randomizadas).

---

### Q38 (D) -- O que eh weak instrument problem?

**Resposta modelo:**
Quando o instrumento Z afeta T fracamente (baixo poder no primeiro estagio),
o estimador IV tem:
- Vies amplificado em direcao ao OLS naive.
- Variancia explodida.
- IC inutilizavel.

Diagnostico: F-statistic do primeiro estagio. Regra de bolso (Staiger &
Stock 1997): F > 10. Se F < 10, instrumento eh fraco.

Solucao: encontrar instrumento melhor; combinar varios; reportar limitacao.

---

### Q39 (D) -- IV identifica ATE ou LATE? Por que?

**Resposta modelo:**
IV identifica **LATE** -- Local Average Treatment Effect -- definido por
Angrist & Imbens (1994).

LATE eh o efeito nos COMPLIERS: pessoas cujo tratamento foi MUDADO pelo
instrumento.

Outras categorias:
- Always-takers: receberiam tratamento independente do instrumento.
- Never-takers: nunca receberiam, mesmo com instrumento.
- Defiers: instrumento causa o oposto (excluidos pela suposicao monotonicidade).

LATE pode ser diferente do ATE se compliers nao sao representativos da
populacao.

---

### Q40 (M) -- Synthetic Control: quando usar?

**Resposta modelo:**
Quando voce tem APENAS UMA unidade tratada (estado, empresa, regiao) e
muitos potenciais controles.

Metodo: constroi 'controle sintetico' como combinacao ponderada dos
nao-tratados que melhor reproduz o pre-tratamento da unidade tratada.

Compara: unidade tratada real vs controle sintetico apos intervencao.
Diferenca eh o efeito causal.

Exemplo classico: Abadie & Gardeazabal (2003) -- impacto do terrorismo
no PIB do Pais Basco. Sem outro 'pais basco' para comparar, sintetizam um.

Em telemetria: 1 servidor tratado e 100 controles. Construir servidor
sintetico ponderando os controles.


## Bloco D -- Causal ML Moderno (Q41-50)

---

### Q41 (F) -- O que eh Double Machine Learning?

**Resposta modelo:**
Framework de Chernozhukov et al. (2018) que combina ML com inferencia
causal sem sacrificar propriedades estatisticas.

Receita:
1. Treinar ML para prever Y dado X (sem T): obter residuos e_Y.
2. Treinar ML para prever T dado X: obter residuos e_T.
3. Regredir e_Y em e_T: coeficiente eh o efeito causal tau.

Inclui cross-fitting para evitar overfitting bias.

Vantagens: usa ML poderoso, lida com muitos confundidores, mantem IC valido.

---

### Q42 (M) -- Por que residualizacao no DML funciona?

**Resposta modelo:**
**Frisch-Waugh-Lovell theorem:** o coeficiente de T em Y = a + tau*T + b*X
eh identico ao coeficiente de regredir (Y - E[Y|X]) em (T - E[T|X]).

Ao residualizar (remover o que X explica de Y e de T), o que sobra eh
a relacao 'limpa' entre T e Y -- o efeito causal sob ignorability.

DML usa ML para estimar E[Y|X] e E[T|X] flexivelmente, e depois aplica
FWL nos residuos.

---

### Q43 (M) -- Causal Forest vs Random Forest classico: qual a diferenca?

**Resposta modelo:**
**Random Forest classico:** cada split minimiza erro de previsao de Y.
Objetivo: prever Y bem.

**Causal Forest (Wager & Athey 2018):** cada split maximiza HETEROGENEIDADE
do efeito tau entre folhas. Objetivo: identificar subgrupos com efeitos
diferentes.

Adicionalmente: 'honesty' -- divide amostra: metade para encontrar splits,
outra metade para estimar efeito (evita overfitting).

Saida: estimativa de CATE para cada individuo -- nao so previsao de Y.

---

### Q44 (D) -- O que sao Meta-Learners?

**Resposta modelo:**
Frameworks que usam ML para estimar CATE, plug-and-play com qualquer modelo:

- **S-Learner:** um modelo Y ~ X + T. CATE = mu(x,1) - mu(x,0).
- **T-Learner:** dois modelos, um por grupo. CATE = mu_1(x) - mu_0(x).
- **X-Learner (Kunzel 2019):** combina T-Learner com correcao via PS;
  melhor para grupos desbalanceados.
- **R-Learner (Nie & Wager):** orthogonalizacao tipo DML; robusto.
- **DR-Learner (Kennedy 2020):** doubly robust por construcao.

Trade-off bias-variance: T-Learner perde dados; S-Learner pode subestimar
efeitos pequenos. X e R-Learner sao melhores em geral.

---

### Q45 (M) -- DoWhy: o que eh e como usar?

**Resposta modelo:**
Framework Python da Microsoft com API unificada para causal inference.
4 etapas:

1. **Modelar:** declarar DAG explicitamente.
2. **Identificar:** algoritmo verifica se efeito eh identificavel.
3. **Estimar:** roda metodo (regression, PSM, IPW, IV, DML, etc.).
4. **Refutar:** testa robustez (placebo, sensitivity, dummy outcome).

```python
import dowhy
model = dowhy.CausalModel(data, treatment='T', outcome='Y', graph='DAG')
estimand = model.identify_effect()
estimate = model.estimate_effect(estimand, method_name='backdoor.linear_regression')
refutation = model.refute_estimate(estimand, estimate, method_name='placebo_treatment_refuter')
```

Forte para mostrar em entrevista: 'uso DoWhy para padronizar processo
e fazer refutation tests sistematicos.'

---

### Q46 (M) -- EconML: o que eh e quando usar?

**Resposta modelo:**
Framework Python da Microsoft focado em metodos modernos de estimativa
heterogenea (CATE).

Inclui:
- DML (Double Machine Learning).
- Causal Forest DML.
- Meta-learners (S, T, X, R, DR).
- DR-IV (Doubly Robust Instrumental Variables).
- DeepIV (neural IV).

Usar quando: voce quer CATE personalizado e ML moderno em producao.

### Q47 (D) -- Qual a diferenca entre DoWhy e EconML?

**Resposta modelo:**
**DoWhy:** infraestrutura conceitual completa. Foco em identificacao,
estimacao e refutacao. Tem muitos metodos basicos.

**EconML:** metodos modernos heterogeneos (CATE). Foco em qualidade
estatistica e performance. Integracao com sklearn.

Eles SE COMPLEMENTAM. Workflow comum: DoWhy para identificacao e refutacao,
EconML para estimacao avancada de CATE.

### Q48 (D) -- Quando usar TMLE (Targeted Maximum Likelihood)?

**Resposta modelo:**
TMLE (van der Laan, 2006) eh uma classe de estimadores DR que 'targeta'
explicitamente o parametro de interesse durante o ajuste.

Diferenca para AIPW: TMLE faz UM update direcionado ao parametro de
interesse, em vez de apenas combinar dois modelos.

Vantagens:
- Mais eficiente que AIPW em alguns cenarios.
- Garantia teorica de eficiencia (com Super Learner).
- Integra bem com Super Learner (ensemble de ML).

Usar quando: voce quer maxima eficiencia e robustez teorica.
Geralmente em medicina e politica publica.

---

### Q49 (D) -- O que sao Highly Adaptive Lasso (HAL) e Super Learner?

**Resposta modelo:**
**HAL:** estimador nao-parametrico com taxas de convergencia muito boas.
Usado dentro de TMLE para garantir eficiencia teorica.

**Super Learner:** ensemble que combina varios algoritmos (RF, GBM, GLM, etc.)
com pesos otimos via cross-validation. Garantia: nao pior que melhor
componente isolado.

Combinacao Super Learner + TMLE eh chamada 'TMLE de van der Laan' e eh
considerada estado da arte em causal inference rigorosa.

### Q50 (D) -- Como avaliar heterogeneidade do efeito honestamente?

**Resposta modelo:**
1. **Best Linear Projection (BLP)** (Semenova & Chernozhukov 2021):
   projetar CATE em features observaveis. Coeficientes mostram quais
   features modificam o efeito.

2. **GATES (Group ATE):** dividir individuos em quintis de CATE estimado.
   Comparar GATE estimado vs GATE real em validacao. Se monotonico e
   significativo, heterogeneidade eh REAL.

3. **Calibration test (Chernozhukov et al.):** se estimativa de CATE eh
   util, deve haver correlacao entre prediction e outcome em validacao.

4. **Bonferroni correction:** se voce procurar efeito em muitos subgrupos,
   eventualmente acha por acaso. Corrigir multiplos testes.

Em entrevista: 'Reporto heterogeneidade SO se sobrevive a calibration test
e BLP. Caso contrario, reporto ATE.'

### Q51 -- Cenario: Avaliacao de Campanha

> 'Voce eh contratado(a). Time de Marketing diz: aumentamos a abertura
> de email em 25% apos lancar nova linha de assunto. Quero saber se foi
> a linha de assunto ou foi sazonalidade.'

**Sua abordagem:**

1. **Clarificar o cenario:**
   - Quando a mudanca foi implementada? Foi A/B test ou rollout total?
   - Que dados tenho do periodo pre-mudanca?
   - Quantos emails enviados antes/depois?

2. **Se foi A/B test puro:** comparar grupos. Cuidado com sazonalidade
   se A/B nao roda em paralelo no tempo.

3. **Se foi rollout total:** problematico. Sem grupo controle simultaneo.
   Opcoes:
   - DiD com SEGMENTOS que receberam a nova vs antiga (se houve)
   - Synthetic Control: criar 'controle' a partir do historico
   - Interrupted Time Series: modelar tendencia e sazonalidade,
     comparar com previsto vs observado

4. **Premissas que VOU verificar:**
   - Sazonalidade: comparar abertura em mesmo periodo do ano anterior
   - Tendencias parallelas (se DiD)
   - Outros eventos concorrentes (mudanca de produto, evento midiatico)

5. **Reportar:**
   - Efeito estimado com IC
   - Premissas e sua plausibilidade
   - Limitacoes (nao posso provar causalidade absoluta sem RCT)

### Q52 -- Cenario: Deploy em Producao

> 'Deploy v2.3 no app foi feito ontem. Crashes subiram 40%. Foi o deploy?'

2. **Confundidores possiveis:**

   - **Comparar plataformas:** se Android v2.3 e iOS v2.3 lancaram em

   - **Trend extrapolation:** taxa de crashes pre-deploy continuaria?
     Diferenca eh efeito do deploy.

4. **Validacao:**
   - Mesma analise por plataforma
   - Logs de stack trace -- se erros sao tipicamente de v2.3, evidencia

5. **Acao recomendada:** rollback se efeito > X, ou hotfix.

Para isolar o efeito causal:
1. PSM/IPW: parear/ponderar clientes nao-premium com clientes premium

2. DiD: comparar receita PRE e POS campanha em premium vs nao-premium.

3. Idealmente: se possivel, randomizar dentro de premium quem recebe

### Q54 -- Cenario: Pricing Experiment

> 'Reduzimos preco de produto em 10% em SP. Vendas subiram 30% em SP vs
> 5% em RJ. Conclusao: reducao causa aumento.'

**Sua abordagem:**

1. **Premissa: tendencias paralelas.** Vendas em SP e RJ tinham mesma
   tendencia antes da mudanca? Plotar serie pre-intervencao.

2. **Confundidores temporais:**
   - Houve outro evento em SP simultaneamente (campanha local,
     concorrente, evento)?
   - Mudanca economica regional?

3. **Metodo: DiD.**
   - Estimar diferenca SP_pos - SP_pre vs RJ_pos - RJ_pre.
   - Modelo: vendas ~ SP + pos + SP*pos + controles
   - Coeficiente da interacao eh o efeito causal incremental.

4. **Cuidado com seasonalidade local:** SP tem padrao mensal diferente?

5. **Reportar:**
   - DiD: efeito X (+IC). Diferente dos 30% naive.
   - Sensitivity: quanto de viesado nao medido para anular efeito?

### Q55 -- Cenario: Mudanca de Politica Interna

**Sua abordagem:**

1. **Confundidores temporais:**
   - Sazonalidade: janeiro/fevereiro tem padrao diferente de inadim

   - Mudancas macro: taxa de juros, desemprego, IPCA?

2. **Comparacao com periodo equivalente do ano anterior.**

### Q56 -- Cenario: Confundidor nao medido

> 'Eu mostrei que usar Pix esta associado com +1.5 pontos de NPS. Voce
> diz que pode ser confundidor nao medido. Como argumentar?'

**Sua abordagem (resposta forte):**

'O confundidor nao medido provavel eh "afinidade com tecnologia" -- nao
posso medir diretamente. Mas posso fazer 3 coisas:

1. **Proxies:** usar IDADE, EDUCACAO, INTENSIDADE de uso de app como
   proxies imperfeitas. Controlar reduz o vies.

2. **Sensitivity analysis (E-value):** calcular quao forte o confundidor
   nao medido precisaria ser (em odds ratio) para anular meu efeito.

3. **IV se possivel:** se houve campanha randomizada incentivando Pix,
   usar como instrumento. Mesmo com confundidor U nao medido, 2SLS
   recupera o efeito.

Se E-value for alto (ex: 2.5), eh implausivel que tal confundidor exista.
Se baixo (ex: 1.1), a conclusao eh fragil e devo reportar como tal.'

### Q57 -- Cenario: Heterogeneidade

> 'Voce estima que campanha de educacao financeira aumenta NPS em 0.5
> pontos em media. O VP de Marketing fica decepcionado. O que voce diz?'

**Sua abordagem:**

'O ATE de 0.5 esconde heterogeneidade. Vou estimar CATE com Causal
Forest e identificar subgrupos.

Resultado provavel: efeito grande (>2.0) em segmento X (baixo letramento
financeiro) e nulo/negativo em outros.

Recomendacao: nao gastar com a campanha para todos -- so para o segmento
X. ROI sobe drasticamente.

Em vez de gastar R$ 100 por cliente x 100k clientes = R$ 10M para ganhar
0.5 pontos NPS medio, gasta R$ 100 x 20k clientes (segmento X) = R$ 2M

Isso eh o que diferencia causalidade clássica de Causal AI moderna:
heterogeneidade leva a ROI 10x maior.

### Q58 -- Cenario: Custos de Erro

> 'Eu prefiro errar para mais ou para menos no efeito causal?'

**Sua abordagem:**

'Depende da DECISAO. Em causal AI a estimativa nao eh o fim -- a decisao
eh.

Se a decisao eh APROVAR uma campanha cara:
- Errar para MAIS (superestimar): aprovo campanha que nao funciona, gasto.
- Errar para MENOS (subestimar): rejeito campanha boa, perda de
  oportunidade.

Custos diferentes para erros opostos.

Em causal AI moderna usamos:
1. IC ampla -- ser conservador na decisao.
2. Loss function assimetrica -- penalizar mais o erro custoso.
3. Decision Theory: combinar estimativa com custos para escolher acao otima.

Por exemplo: ATE = +0.5, IC [0.0, 1.0]. Se custo de campanha alto, decido
pelo limite INFERIOR do IC. Se houvesse decisao reversivel, decido pela
MEDIA.'

2. **RDD** se houve cutoff de elegibilidade: comparar logo acima e abaixo.

4. **Synthetic Control** se ha apenas 1 unidade tratada e muitos
   nao-tratados similares.

5. **Reportar honestamente:** se nada disso funcionar, dizer 'sem dados
   pre nao posso isolar efeito causal -- analise eh exploratoria/descritiva,
   nao causal.'

Eh melhor admitir limitacao que reportar pseudo-causalidade.'
